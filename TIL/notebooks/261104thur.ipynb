##### 오늘 한 일
- DB에 저장된 회원 정보를 Python 객체로 읽어 오는 방법을 학습했다.
- 기존에 직접 작성하던 SQL 조회를 SQLAlchemy ORM을 활용한 Python 코드로 표현해 봤다.

##### 새롭게 이해한 것
- Repository는 DB 작업을 모아 두는 역할을 한다. ORM은 그 안에서 개발자가 직접 처리하던 SQL 작성과 조회 결과 변환의 일부를 도와준다.
- Pydantic 모델은 API가 데이터를 받고 보내는 형식을 정의한다. SQLAlchemy 모델은 실제 DB 테이블과 Python 클래스를 연결한다. 둘 다 모델이라고 부르지만 역할은 다르다.
- `__init__.py`는 `app` 폴더를 Python 패키지로 다루기 위해 사용하는 파일이다.
- 클래스는 타입이다: 클래스를 정의하면 새로운 종류의 데이터 타입을 만들 수 있다. 예를 들어 `Member` 클래스는 회원이라는 타입을 정의하고, 회원 한 명의 정보를 담은 객체는 `Member` 타입에 해당한다.

##### 오늘 느낀 점
