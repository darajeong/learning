##### 오늘 한 일

- `input()`을 사용해 사용자에게 데이터를 입력받는 방법을 학습했다.
- 입력받은 문자열을 `int()` 또는 `float()`로 형 변환하는 방법을 알아봤다.
- `if`, `elif`, `else`를 사용해 조건에 따라 다른 코드를 실행했다.
- `and`, `or`, `not` 논리 연산자의 결과를 확인했다.
- 사용자 입력과 조건문을 활용하는 계산 및 조건 판단 문제를 살펴봤다.

##### 새롭게 이해한 것

- `input()`으로 입력받은 값은 항상 문자열이므로 계산에 사용하려면 목적에 맞는 데이터형으로 변환해야 한다.
- `input()` 안에 안내 문구를 작성하면 사용자에게 입력해야 할 내용을 알려 줄 수 있다.
- 문자열과 숫자를 `+`로 연결할 때는 숫자를 `str()`로 변환해야 한다.
- f-string을 사용하면 `str()`을 일일이 사용하지 않아도 `{}` 안에 변수를 넣어 문자열과 함께 출력할 수 있다.
- 문자열 `"1" + "2"`는 숫자 덧셈이 아니라 문자열 연결이 되어 `"12"`가 된다.
- 문자열 `"15" * 2`는 숫자 곱셈이 아니라 문자열을 두 번 반복하여 `"1515"`가 된다.
- `>=`, `<`, `==`, `!=` 같은 비교 연산자의 결과는 `True` 또는 `False`로 나온다.
- `if`와 `else` 뒤에는 콜론(`:`)을 작성하고, 조건에 포함되는 코드는 들여쓰기해야 한다.
- `elif`는 앞의 조건이 거짓일 때 다음 조건을 검사하고, `else`는 앞의 모든 조건에 해당하지 않는 나머지 경우를 처리한다.
- `and`는 두 조건이 모두 참일 때, `or`는 하나 이상의 조건이 참일 때 `True`가 된다. `not`은 참과 거짓을 반대로 바꾼다.
- `if-elif-else`는 위에서부터 순서대로 조건을 확인하므로 조건을 작성하는 순서가 중요하다.
- 여러 범위로 조건을 나눌 때는 조건이 바뀌는 지점의 경계값을 테스트하는 것이 중요하다.
- 계산한 결과를 변수에 저장할 때는 대입 연산자 `=`를 사용한다.
- 조건식의 결과인 `True` 또는 `False`를 변수에 저장한 뒤 `if`문의 조건으로 사용할 수도 있다.
- 로그인처럼 여러 조건이 동시에 맞아야 하는 경우에는 각 조건을 비교한 뒤 `and`로 연결할 수 있다.
