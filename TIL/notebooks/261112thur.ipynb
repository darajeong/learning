##### 새롭게 이해한 것

* 조회 버튼을 누르면 상태를 `loading`으로 바꾸고, `await getMembers(...)`로 응답을 기다린다. 그동안 브라우저에는 “불러오는 중”이 표시된다. 응답이 오면 회원 수에 따라 `success` 또는 `empty`를 표시하고, 요청이나 응답 검사에서 오류가 발생하면 `error`를 표시한다.
* 백엔드 API를 만들 때 사용할 수 있는 언어와 프레임워크에는 Python(Flask, FastAPI, Django), JavaScript/TypeScript(Express, NestJS), Java(Spring Boot)가 있다.
* 자주 접하는 포트 번호와 용도는 다음과 같다.

  | 포트 | 용도 |
  |---|---|
  | 22 | SSH로 다른 컴퓨터에 원격 접속할 때 쓰는 기본 포트 |
  | 80 | HTTP 웹사이트의 기본 포트 |
  | 443 | HTTPS 웹사이트의 기본 포트 |
  | 3000 | Next.js 개발 서버의 기본 포트 |
  | 3306 | MySQL 데이터베이스의 기본 포트 |
  | 5000 | Flask 개발 서버의 기본 포트 |
  | 5432 | PostgreSQL 데이터베이스의 기본 포트 |
  | 8000 | FastAPI를 실행할 때 쓰는 Uvicorn 서버의 기본 포트 |
  | 8080 | 개발용 웹 서버에서 자주 쓰는 포트 |

* 엔터티는 ‘회원’처럼 데이터로 표현하려는 대상이다. 객체는 ‘홍길동’, ‘김철수’처럼 구체적인 회원을 표현한 것이고, 인스턴스는 그렇게 만들어진 회원 객체 하나하나를 가리킨다. ORM은 Python 객체를 DB에 저장하거나 불러올 수 있도록 연결해 주는 도구다. SQL을 직접 많이 작성하지 않아도 DB를 다룰 수 있다.
* 호출은 프론트엔드에서 하고, `main.py`는 백엔드의 진입점이다.
* `localhost`는 내 컴퓨터에서 띄운 웹사이트에 접속할 때 쓰는 주소이고, 컴포넌트는 웹사이트 화면을 이루는 코드 부품이다.
