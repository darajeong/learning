* React와 FastAPI를 활용한 풀스택 방식이 아니라 Apache·PHP·MySQL을 사용하는 LAMP 계열 방식으로 웹사이트를 만드는 과정을 학습하였다. 브라우저의 요청을 PHP가 처리하고, SQL로 MySQL의 데이터를 조회한 뒤 결과를 화면에 보여주는 흐름을 익혔다.

##### 새롭게 이해한 것
- PHP는 웹 서버에서 사용자 요청을 처리하고 필요하면 데이터베이스를 조회하여 브라우저에 보낼 HTML을 만드는 프로그래밍 언어이다.
- 웹 주소를 입력하면 브라우저가 DNS로 서버를 찾고 Apache에 요청을 보내며, PHP가 MySQL의 데이터를 처리해 만든 결과를 Apache가 브라우저에 돌려준다.
- DNS는 서버 주소를 찾고, Apache는 웹 요청을 받으며, PHP는 요청을 처리하고, MySQL은 데이터를 보관하며, SQL은 MySQL에 데이터 작업을 지시한다.
- 전형적인 LAMP는 PHP가 데이터 처리와 HTML 생성을 함께 담당하지만, Modern Full Stack은 React 같은 화면 부분과 FastAPI 같은 서버 부분을 분리해 주로 API로 통신한다.
- LAMP와 Modern Full Stack은 구조는 다르지만 모두 사용자 요청을 서버에서 처리하고 데이터베이스 결과를 화면에 보여주는 같은 웹의 기본 원리를 사용하므로 함께 비교한다.

##### 오늘 느낀 점
* 아직 웹사이트를 만드는 과정이 와닿지 않는다. 풀스택 방식과 LAMP 방식이 있다는 것을 처음 알았고, 각각 차이에 대해 좀 더 깊게 공부해 봐야 할 것 같다.
* 파이썬을 활용하여 여거가지로 활용할 수 있다는 것을 알게되었다. 그 중 하나가 인덱스로도 비효율적인 부분을 해결할 수 있다는 것을 알 수 있었다.
