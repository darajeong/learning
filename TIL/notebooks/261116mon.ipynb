##### 새롭게 이해한 것

- `.env`에 설정한 `text-embedding-3-small`은 텍스트를 숫자 묶음으로 바꿔 주는 임베딩 모델이다. 예를 들어 “파이썬 숙제 언제까지 내?”라고 물으면 질문과 공지 내용을 각각 숫자 묶음으로 바꾼다. 이를 비교하면 질문과 의미가 가까운 공지를 찾을 수 있다.
