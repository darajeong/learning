##### 새롭게 이해한 것

- 인덱싱은 질문이 오기 전에 자료를 찾기 좋게 준비하는 일. 긴 문서를 작은 조각으로 나누고, 각 조각의 의미를 비교할 수 있도록 숫자 묶음으로 바꿔 저장함
- 청크 오버랩은 문서를 나눌 때 이웃한 조각에 내용을 조금씩 겹쳐 넣는 방법. 경계에서 잘려도 ‘금요일 오후 6시’가 무엇의 마감인지 문맥을 함께 찾기 쉬워짐
- 질의 단계는 질문을 받은 뒤 관련 자료를 찾고 답변을 만드는 전체 과정
- 리트리버는 질문에 맞는 자료 조각을 골라오는 검색 담당. 도서관에서 필요한 페이지를 찾아주는 사서에 해당
- 생성은 모델이 찾은 자료를 읽고 질문에 맞는 답변을 쓰는 단계. 공지에는 금요일이라고 적혀 있는데 모델이 토요일이라고 답했다면, 자료는 제대로 찾았고 답변을 만드는 단계에서 오류가 생긴 것
