##### 새롭게 이해한 것
* Path Parameter는 조회·수정·삭제할 대상을 주소 경로에서 식별하는 값이다. 예를 들어 `/members/1`의 `1`은 대상 회원의 ID다.
* Query Parameter는 검색·필터·정렬 등의 조건을 전달하는 값이다. 예를 들어 `/members?name=김민수`의 `name=김민수`는 검색 조건이다.
* psycopg는 Python에서 PostgreSQL에 연결하고 SQL을 실행하며 결과를 받아올 수 있게 해 주는 라이브러리다. 요청은 `프론트엔드 → FastAPI 백엔드 → psycopg → PostgreSQL` 순서로 전달된다.
* RDBMS는 행과 열로 구성된 테이블을 중심으로 데이터를 관리하는 관계형 데이터베이스 관리 시스템이다. PostgreSQL과 MySQL이 대표적이다.
* NoSQL은 문서형, 키-값형 등 관계형 테이블 이외의 데이터 모델을 사용하는 데이터베이스 계열이다. MongoDB는 JSON과 비슷한 문서 형태로 데이터를 저장하는 문서형 NoSQL 데이터베이스다.
* Pydantic(파이댄틱)은 Python에서 데이터가 정해진 형식과 조건에 맞는지 검사하는 라이브러리다. Request 모델은 서버가 받을 데이터의 형식을, Response 모델은 서버가 돌려줄 데이터의 형식을 정의한다.
* 이번 실습에서 `204`는 삭제 성공과 응답 본문 없음을, `404`는 해당 회원이 없음을 의미한다. `422`는 입력값 검증 실패를, `500`은 DB 연결이나 SQL 실행 등 서버 내부 처리 오류를 의미한다.
