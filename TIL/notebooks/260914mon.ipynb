- p-value는 두 집단에 차이가 없다는 귀무가설 아래 현재와 같은 결과가 우연히 나타날 가능성을 의미하며, 일반적으로 0.05보다 작으면 귀무가설을 기각한다.
- 과적합은 학습 데이터만 지나치게 잘 맞히는 상태이고 과소적합은 학습과 테스트 데이터 모두 잘 맞히지 못하는 상태이다.
