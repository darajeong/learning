##### 새롭게 이해한 것

- **Top-K:** 검색 결과를 몇 개 가져올지 정하는 값이다. K=3이면 유사도가 높은 순서대로 최대 3개를 가져온다.
- **Hit@K:** 가져온 결과 안에 정답 근거가 있는지 확인하는 지표다. Hit@3=1은 상위 3개 중 하나 이상에 정답 근거가 있다는 뜻이다.
- **코사인 유사도와 거리:** 코사인 유사도는 -1~1 범위이며 클수록 비슷하다. 코사인 거리는 `1 - 유사도`로 계산하며, 0~2 범위에서 작을수록 비슷하다.
- **유사도:** 컴퓨터가 질문과 글이 얼마나 비슷한지 계산한 점수다.
- **RR:** 정답 근거가 처음 등장한 순위의 역수다. 첫 번째에 나오면 1, 두 번째에 나오면 0.5다.
- **Gold:** 평가를 위해 사람이 미리 정해둔 정답 문구다.
