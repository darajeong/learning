##### 새롭게 이해한 것
* 웹사이트의 기본 화면은 `HTML`로 구조를 만들고, `CSS`로 꾸미며, `JavaScript`로 클릭이나 입력 같은 동작을 구현한다.
* 기능이 많은 웹서비스는 화면을 관리하는 `React`, 서버 기능을 처리하는 `Python`이나 `Java`, 데이터를 저장하고 조회하는 `SQL` 등을 함께 사용할 수 있다.
* 하나의 사이트에는 역할이 다른 여러 코드가 들어가며, 사용자의 요청을 프론트엔드가 받아 백엔드로 전달하고 데이터베이스에 저장한 뒤 그 결과를 다시 화면에 보여준다.
* `React`는 별도의 프로그래밍 언어가 아니라 JavaScript를 이용해 화면을 컴포넌트라는 작은 부품으로 나누고 데이터 변화를 편리하게 관리하는 라이브러리이다.
* `Figma`는 개발 전에 웹사이트 화면을 설계하는 디자인 도구이고, `Bootstrap`은 실제 웹페이지의 버튼이나 입력창을 빠르게 꾸밀 수 있도록 미리 만들어진 디자인을 제공하는 도구이다.
* 웹사이트는 필요에 따라 `Figma`로 화면을 설계하고, `React`로 화면과 기능을 구현하며, `Bootstrap`으로 디자인을 빠르게 적용할 수 있다.
