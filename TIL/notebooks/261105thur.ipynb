##### 새롭게 이해한 것

1. ORM과 SQL의 차이
- 회원 가입이나 정보 수정이 많은 앱에서는 **ORM**을 사용하는 것이 편리하다.
- 반면, 여러 테이블을 복잡하게 합쳐 매출 통계를 내는 작업에서는 **SQL을 직접 작성하는 것**이 더 명확할 수 있다.

2. FastAPI와 SQLAlchemy ORM의 역할
- **FastAPI**: 사용자의 요청(Request)을 받고 응답(Response)을 반환하는 역할
- **SQLAlchemy ORM**: Python의 Model(클래스)과 DB의 Table(테이블)을 연결하고, 데이터를 다룰 수 있도록 도와주는 역할

3. Commit과 Rollback
- **Commit**: DB의 변경 사항을 확정하고 영구적으로 저장하는 것(영속화)
- **Rollback**: 아직 Commit하지 않은 변경 사항을 되돌리는 것
`autocommit=False`는 변경 사항을 자동으로 확정하지 않겠다는 설정이다.
따라서 `commit()`을 직접 호출해 저장을 확정해야 한다.
**주의:** 이미 `commit()`으로 확정한 변경 사항은 `rollback()`으로 되돌릴 수 없다.

4. Session과 db의 개념
**Session은 DB 작업을 진행하는 '작업 공간'이다.**
Session에는 DB 작업에 필요한 여러 기능이 있으며, 이러한 기능을 **메서드(Method)** 라고 부른다.
`db`가 Session 객체라면 점(`.`)을 찍어서 해당 기능을 사용할 수 있다.

예시:

```python
db.add(member)       # 회원 추가
db.get(Member, 1)    # 1번 회원 조회
db.commit()          # 변경 사항 저장 확정
db.rollback()        # 변경 사항 되돌리기
```

**`get_db()`의 역할**
- 요청(Request)마다 Session을 열어준다.
- 요청 처리가 끝나면 Session을 정리한다.

**`db`의 의미**
`db`는 DB 작업을 할 때 사용하는 Session 객체에 붙인 이름이다.

```python
def find_all(db: Session):
```

이 코드의 의미는 다음과 같다.
> "이 함수에 Session이라는 DB 작업 도구를 전달해 줘. 여기서는 그 도구를 `db`라고 부를게."
즉, **`db`는 Session 객체이며, 오늘 학습의 가장 중요한 핵심 개념이다!**

5. 데이터 무결성(Integrity)
**Integrity를 유지한다**는 것은 데이터의 정확성과 일관성을 지킨다는 의미다.
쉽게 말해, 엉뚱하거나 잘못된 값이 DB에 들어가지 않도록 관리하는 것이다.

6. Request Body, Response Body, Schema
- **Request Body**: 사용자가 서버에 보내는 데이터  
  예: 회원 가입 시 입력하는 이름, 이메일, 나이
- **Response Body**: 서버가 사용자에게 돌려주는 데이터  
  예: 생성된 회원 번호와 회원 정보
- **Schema**: 데이터의 형식과 허용 값을 정의하는 Python 클래스

7. HTTP 상태 코드
- **500 Internal Server Error**: 서버 내부에서 오류가 발생했다는 의미
