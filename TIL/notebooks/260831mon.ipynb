##### 새롭게 이해한 것
- **디버깅 전략**은 오류를 감으로 고치지 않고 트레이스백과 변수 값을 확인하면서 문제의 원인을 단계적으로 좁혀 가는 방법입니다.
- **스모크 테스트**는 프로그램을 자세히 검사하기 전에 실행 여부와 핵심 기능이 정상적으로 작동하는지 빠르게 확인하는 기본 테스트입니다.
- **Django**는 Python으로 서버, 데이터베이스, 로그인 등의 백엔드 기능을 만드는 웹 프레임워크입니다.
- **React**는 JavaScript로 버튼, 입력창, 메뉴 등 사용자가 보는 화면을 만드는 라이브러리입니다.
- Figma에서 만든 코드를 GPT나 Codex에서 이어 개발할 때는 디자인과 파일 구조가 깨지지 않도록 전체 프로젝트를 전달하고 기능을 하나씩 추가해야 합니다.
- **엔터티**는 회원·상품·주문처럼 관리하려는 대상이고, **테이블**은 해당 대상의 실제 데이터를 행과 열로 저장하는 표입니다.
- 데이터베이스는 PostgreSQL에 데이터를 저장하고 DBeaver와 SQL로 관리하며, 모델링·정규화·무결성·트랜잭션·인덱스·보안·백업 등의 방법으로 데이터를 정확하고 안전하게 관리합니다.
