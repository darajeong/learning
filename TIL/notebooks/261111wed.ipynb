##### 새롭게 이해한 것
* **CORS 설정**은 브라우저에서 어떤 출처의 요청을 허용할지 정하는 설정이다. 예를 들어 Next.js 화면이 `http://localhost:3000`에서 실행되고 FastAPI 서버가 `http://127.0.0.1:8000`에서 실행된다면, 두 주소의 출처가 다르므로 FastAPI에서 `http://localhost:3000`의 요청을 허용해야 한다.
* **Soft Delete**는 데이터를 DB에 남겨둔 채 삭제된 상태로 표시하는 방식이다.
* **`models.py`**는 DB에 저장하는 데이터의 구조를 정의한다. 예를 들어 회원 한 명의 데이터가 회원번호, 이름, 이메일, 나이 같은 항목으로 구성된다는 것을 정한다.
* **`schemas.py`**는 화면과 서버가 주고받는 데이터의 형식을 정의한다. 예를 들어 요청에 이름, 이메일, 나이가 들어왔는지 확인하는 기준이 된다.
