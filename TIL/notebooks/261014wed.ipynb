##### 새롭게 이해한 것
* HTML 요소는 기본적으로 박스로 배치되며, 안쪽부터 content → padding → border → margin으로 구성된다.
* padding은 내용과 테두리 사이의 안쪽 공간, margin은 테두리 바깥의 외부 공간이다.
* 몇 px의 차이도 화면 배치에 영향을 준다. CSS에서는 px외에 %, rem 등의 단위도 사용한다.
* div는 기본적으로 block이라 새 줄을 차지하고, span은 inline이라 글자처럼 같은 줄에 이어진다. 이 배치는 CSS의 display로 바꿀 수 있다.
