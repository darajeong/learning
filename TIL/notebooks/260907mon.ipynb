##### 새롭게 이해한 것
- **SQL 집계와 필터링**에서 `COUNT(*)`는 전체 행 수를 세고, `AS`는 결과 열에 별칭을 붙이며, `WHERE`는 조건에 맞는 행을 고르고, `IN`은 여러 값 중 일치하는 값을, `LIKE`는 문자열 패턴을 검색하며, `IS NULL`·`IS NOT NULL`은 값의 미입력 여부를 확인하고, `NULLS LAST`는 정렬할 때 `NULL`을 마지막에 배치합니다.
- **AI**는 인공지능을 넓게 부르는 말이고, **AGI**는 다양한 분야에 지식을 응용하는 범용 인공지능이며, **멀티모달**은 글·이미지·음성 등 여러 형태의 정보를 다루는 능력이므로 멀티모달 AI가 반드시 AGI인 것은 아닙니다.
- **우분투 서버**는 웹사이트·파일 공유·데이터베이스 같은 서비스를 실행하는 데 사용하는 리눅스 운영체제로, 일반적으로 명령어나 SSH 원격 접속을 통해 관리합니다.
- **`INSERT ... RETURNING`**은 새 데이터를 추가하고 그 결과를 바로 보여주는 반면, **`SELECT ... WHERE`**는 데이터를 변경하지 않고 조건에 맞는 기존 데이터를 조회합니다.
