##### 오늘 한 일
* NAVER 검색 트렌드 API와 공공데이터포털의 관광 API로 데이터를 수집했다. 검색 관심도를 분석하고, AI 요약과 그래프를 담은 PDF 보고서를 자동으로 생성하는 과정을 학습했다.

##### 새롭게 이해한 것
* 정기적으로 생성되는 보고서를 통해 관심 주제의 변화를 살펴보고, 콘텐츠 기획과 마케팅 시점을 판단하는 데 도움을 받을 수 있다는 것을 이해했다.
* API를 호출할 때는 요청 주소인 URL과 HTTP 메서드인 GET과 POST 등을 함께 확인해야 한다. GET은 주로 데이터를 조회할 때 사용하고, 이번 NAVER 검색 트렌드 API는 POST 방식으로 조회 조건을 사용한다. 
* URI는 자원을 식별하는 개념이며, URL은 자원의 위치와 접근 방법을 나타내는 URI의 한 종류이다. 

##### 오늘 느낀 점
* 이 방법을 회사에 다닐 때 알았더라면 광고비·클릭·구매 데이터를 연결해 광고 성과를 분석하고, 캠페인을 더 효율적으로 운영할 수 있지 않았을까 하는 아쉬움이 들었다. 앞으로 배운 내용을 실제 업무에 어떻게 활용할 수 있을지 생각하며 공부하고 싶다.
