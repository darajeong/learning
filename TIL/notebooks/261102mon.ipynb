- 회원 정보를 DB에 저장하고, 성공하면 변경을 확정(commit), 실패하면 취소(rollback)하는 API를 만들었다.
- PATCH로 회원의 나이만 수정하고, Swagger 응답·DB 조회·GET 재조회 결과가 일치하는지 확인했다.

##### 새롭게 이해한 것
- `insert_member()` 함수를 Swagger에서 실행하려면, 해당 함수를 호출하는 API 엔드포인트를 `main.py`에 연결해야 한다.
- GET은 기존 데이터를 조회할 때, POST는 새 회원을 등록할 때 사용한다.
- Uvicorn은 FastAPI 앱을 실행하는 서버이고, Swagger UI는 API 요청을 보내고 응답을 확인하는 도구이다. 실제 회원 등록과 DB 저장은 백엔드 코드가 처리한다.
- 엔드포인트는 서버의 특정 기능을 요청하는 진입점이다. 예를 들어 `GET /members`는 회원 목록 조회, `POST /members`는 회원 등록을 위한 엔드포인트이다.
- 상태 코드 `200`은 요청 처리 성공을, `201`은 새 데이터 생성 성공을 의미한다.
- Pydantic은 DB 저장 코드가 실행되기 전에 입력값의 형식과 조건을 검사한다.
- PostgreSQL의 자동 증가 ID는 발급된 뒤 rollback되더라도 되돌아가지 않아 번호가 건너뛸 수 있다.
- PUT은 수정 가능한 전체 항목을 보내고, PATCH는 바꿀 항목만 보낸다. PATCH에서 생략한 항목은 기존 값을 유지한다.
