##### 새롭게 이해한 것
- 여러 CSV DataFrame은 순서로 구분하는 리스트보다 `"customers"`처럼 이름으로 꺼낼 수 있는 딕셔너리로 관리하면 더 읽기 쉽고 실수를 줄일 수 있다.
- `isna().sum().sum()`은 모든 열의 결측값을 합쳐 전체 결측값 수를 구하고, `duplicated().sum()`은 중복 여부 결과를 한 번 더해 중복 행 수를 구한다.
- `UNION`은 같은 구조의 데이터를 위아래로 붙이고, `JOIN`은 공통 ID를 기준으로 서로 다른 테이블의 정보를 옆으로 연결한다.
- `seaborn`은 데이터 분석용 그래프를 쉽게 만들도록 돕는 도구이고, `matplotlib`는 그래프 제목·축·크기처럼 세부 모양을 조절하는 데 사용한다.
