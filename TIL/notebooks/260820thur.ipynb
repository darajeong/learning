#### 8월 21일 TIL — 변수, 데이터형, 연산자와 문자열

##### 오늘 한 일

- Python의 기본 데이터형과 형 변환 방법을 학습했다.
- 산술 연산자와 복합 대입 연산자를 실습했다.
- 문자열의 인덱싱, 슬라이싱, 메서드를 사용해 봤다.
- f-string과 이스케이프 문자의 사용법을 알아봤다.

##### 새롭게 이해한 것

- 따옴표로 감싼 숫자는 문자열이므로 계산 전에 형 변환이 필요하다.
- `=`는 값을 저장하고 `==`는 두 값을 비교한다.
- 문자열의 인덱스는 `0`부터 시작하며, 음수 인덱스는 뒤에서부터 센다.
- 문자열 메서드는 기존 문자열이 아니라 새로운 문자열을 반환한다.
- 문자열과 변수를 함께 출력할 때는 f-string을 사용하면 편리하다.
