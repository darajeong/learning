##### 새롭게 이해한 것
* `greet()`처럼 같은 함수를 여러 번 호출하면 함수 안에 작성된 코드도 호출한 횟수만큼 반복해서 실행된다.
* 함수의 파라미터는 함수를 호출할 때 전달된 실제 값을 받아 함수 안에서 사용할 수 있도록 저장하는 변수이다.
* `.venv`는 작성한 코드를 저장하는 폴더가 아니라 프로젝트 전용 파이썬 실행 환경과 설치한 외부 패키지를 관리하는 공간이다.
* 터미널의 `PS`는 PowerShell, 경로는 현재 폴더, `(.venv)`는 가상환경이 활성화된 상태를 뜻하며 동그라미나 별표 모양은 터미널 디자인이므로 코드 실행에 영향을 주지 않는다.
* Gemini, GPT, Claude는 각각 별도의 API 키가 필요하며, 대시보드에 연결할 AI는 비용·기능·응답 품질·프로젝트 목적에 따라 선택할 수 있다.
