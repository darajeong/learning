##### 새롭게 이해한 것
- PostgreSQL과 MySQL은 데이터를 실제로 저장하고 관리하는 DBMS이고, DBeaver는 이 DBMS에 접속해 SQL을 실행하고 데이터를 편리하게 관리하는 도구입니다.
- SQL은 `SELECT`부터 작성하지만, 데이터베이스는 먼저 `FROM`으로 대상 테이블을 정한 뒤 조건 적용과 그룹화 등을 거쳐 최종적으로 보여줄 데이터를 선택합니다.
- 기준 데이터는 업무 판단의 근거가 되는 원래 기록이고, 파생 데이터는 기준 데이터를 계산하거나 가공해서 만든 결과입니다.
- Figma Design은 화면을 설계하고 Figma Make는 작동 가능한 프로토타입과 코드를 만들며, 공개 링크만 필요하면 Figma Make에서 발행하고 실제 서비스에 적용하려면 코드를 개발 도구로 가져와 프로젝트에 맞게 통합해야 합니다.
