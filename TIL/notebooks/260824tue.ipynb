##### 새롭게 이해한 것
* `set`의 `remove()`로 존재하지 않는 값을 삭제하면, 집합이 값을 `key`처럼 관리하기 때문에 `KeyError`가 발생한다. 
* `discard()`는 값이 없어도 오류가 발생하지 않으며, `remove()`는 값이 반드시 있어야 하는 상황에 사용한다. 
* `total = 0`, `count = 0`은 반복문에서 합계나 개수를 누적하기 위한 시작값으로 자주 나온다.
