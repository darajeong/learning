##### 새롭게 이해한 것
* Flexbox: 메뉴나 버튼을 한 방향 (가로 또는 세로)으로 정렬할 떄 사용한다. 부모 요소의 `display: flex;`(가로),`flex-direction: column;` (세로) 로 적용한다.
* Grid: 카드 등을 행과 열에 맞춰 배치할 때 사용한다. 부모 요소의 `display: grid;`로 적용한다. (grid-template-columns로 열 수를 정함)
* Mobile First: 작은 화면에서 사용하기 편하도록 기본 CSS를 먼저 작성하고, 넓은 화면용 규칙을 추가하는 방식이다. 카드 1열→2열→3열은 그 예시이다.
* Breakpoint: 화면 배치를 변경하는 기준 너비이다. ex. 768px에서 2열, 1024px에서 3열로 바꾼다면 두 값이 Breakpoint이다.
