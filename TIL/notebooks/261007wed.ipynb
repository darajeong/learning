##### 오늘 한 일
- 4번째 프로젝트를 기획했다. 가상의 카드사인 AURA CARD의 포인트를 활용할 수 있는 리워드몰을 기획했고, 일반 포인트몰과 차별화하기 위해 향 관련 상품과 커스텀 요소를 결합했다.
- 카드 사용으로 쌓인 포인트를 활용해 향 관련 리워드 상품을 교환할 수 있도록 서비스 흐름을 정리했다.
- 카드 등급별 적립 혜택과 리워드 사용 방식을 구분해 카드사만의 차별성을 주는 방향으로 기획했다.
- Next.js 프로젝트 초기 환경을 구성하고 GitHub 저장소와 연결했다.
- 팀 프로젝트 역할 분담과 서비스 콘셉트, 메인 화면 구조를 정리했다.

##### 새롭게 이해한 것
- 단순히 화면을 예쁘게 만드는 것보다 서비스의 목적과 사용자 흐름을 먼저 정리하는 것이 중요하다는 점을 알게 되었다.
- 카드 혜택, 포인트 적립, 포인트 사용, 리워드 교환처럼 서로 연결되는 기능은 기획 단계에서 관계를 명확하게 정해야 이후 개발할 때 로직을 구현하기 쉽다는 것을 이해했다.
- 재사용 가능한 컴포넌트와 공통 레이아웃을 미리 구분해 두면 여러 페이지를 개발할 때 중복 작업을 줄일 수 있다는 점을 배웠다.

##### 오늘 느낀 점
- 생각보다 페이지마다 신경 써야 할 부분이 많았다. 단순히 화면을 나누는 것뿐만 아니라 각 페이지가 어떤 역할을 하는지, 서로 어떻게 연결되는지도 함께 고민해야 했다.
- 그래도 기획 단계에서 서비스 구조와 기능을 구체적으로 정리해두는 것이 이후 개발을 훨씬 수월하게 만든다는 생각이 들었다.
