##### 새롭게 이해한 것

* 딕셔너리는 데이터를 **`키(key): 값(value)` 형태로 연결하여 저장**하며, 리스트나 튜플과 달리 위치가 아닌 키를 이용해 원하는 값을 찾는다.
* 딕셔너리의 키는 중복될 수 없으며, 같은 키에 새로운 값을 저장하면 기존 값이 새로운 값으로 덮어써진다.
* 메서드는 특정 객체가 사용할 수 있는 기능으로 `변수이름.메서드()` 형태로 작성하며, `split()`은 문자열을 나누어 리스트로 반환한다.
* `word_count[word]`는 현재 단어의 등장 횟수를 가져오며, `word_count[word] = word_count[word] + 1`은 기존 횟수에 `1`을 더해 다시 저장한다.
