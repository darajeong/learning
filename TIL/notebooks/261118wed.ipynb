##### 새롭게 이해한 것

* pgvector: PostgreSQL에서 벡터를 저장하고 검색할 수 있게 해 주는 확장 도구.
* `cosine=1.000`은 유사도가 1이라는 뜻으로, 학습용 벡터에서 질문과 문단의 방향이 동일하다는 의미다.
* Fixture는 미리 정한 규칙으로 컴퓨터 안에서 계산한다. 이번 실습의 Fixture 테이블에는 숫자 3개로 이루어진 벡터(`[1, 0, 0]`)를 저장한다. Live는 실습 코드에서 `openai` 라이브러리를 사용해 OpenAI 서버에 임베딩을 요청하며, Live 테이블에는 숫자 1,536개로 이루어진 벡터를 저장한다.
* Fixture와 Live는 서로 다른 기준으로 만든 숫자를 섞어 비교하지 않도록 테이블을 나누어 저장한다.
* ChromaDB: AI가 문서를 검색할 때 쓰는 벡터를 저장하고 찾아주는 DB.
* 코사인 유사도 = 1 − 코사인 거리.
* 키워드 검색 점수가 높을수록 질문과 문단에 겹치는 키워드가 많거나, 코드가 중요하게 보는 단어가 일치했다는 뜻이다.
