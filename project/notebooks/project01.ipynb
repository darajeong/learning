##### 새롭게 배운 내용

`src/charts.py`는 차트를 만드는 기능을 담당하고, `app.py`는 만들어진 차트나 표를 실제 화면에 보여 주는 역할을 한다.

`st.stop()`이 실행되면 그 아래에 있는 지표, 차트, 표 관련 코드는 실행되지 않는다. 데이터가 하나도 없는 상태에서 차트를 만들려고 하면 오류가 발생할 수 있으므로 필요한 기능이다.

필터는 드롭다운 형식만 있는 것으로 알고 있었지만, 다른 형태의 위젯도 있다는 것을 배웠다.

- `st.multiselect()`: 여러 개의 옵션을 선택할 수 있는 위젯
- `st.date_input()`: 날짜를 선택할 수 있는 위젯

Streamlit을 다시 실행할 때는 다음 명령어를 입력한다.

```bash
streamlit run app.py
```

새로고침을 해도 수정 내용이 적용되지 않으면 `Ctrl + C`를 눌러 실행을 종료한 뒤, 다음 명령어로 다시 실행한다.
