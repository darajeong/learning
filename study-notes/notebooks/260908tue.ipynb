````markdown
##### 1. 추가 학습 내용 정리

- 파이썬으로 데이터 폴더를 찾는 방법을 배우고, 작성한 코드를 함수와 모듈로 만들어 여러 곳에서 재사용합니다.
- `line_total`처럼 기존 데이터로부터 계산해 만든 컬럼을 파생 컬럼이라고 하며, 수량과 단가를 곱하면 주문상세별 금액을 만들 수 있습니다.
- `PROJECT_ROOT = Path.cwd()`는 실행 중인 파일의 위치가 아니라 파이썬의 현재 작업 폴더를 가져오며, 그 위치가 반드시 프로젝트 루트인 것은 아닙니다.
- `current = Path.cwd().resolve()`는 현재 작업 폴더를 가져와 경로를 정리한 절대 경로를 `current`에 저장합니다.
- `current / "data"`는 현재 경로 뒤에 `data`를 연결할 뿐, 실제 폴더를 만들거나 작업 폴더를 이동시키지는 않습니다.
- `current.parent`는 부모 폴더의 경로이고, `current.name`은 경로의 마지막 파일명이나 폴더명입니다.
- `current.exists()`는 해당 경로에 파일이나 폴더가 있는지 확인하고, `current.is_dir()`는 해당 경로가 실제 폴더인지 확인합니다.
- `for`는 주어진 항목을 하나씩 꺼내 반복하고, `while`은 조건이 참인 동안 반복합니다.
- 데이터 폴더 탐색은 `for`로 부모 경로들을 차례로 확인하거나, `while`로 탐색 위치를 부모 경로로 바꾸며 반복하도록 작성할 수 있습니다.
- 모듈은 `pathlib`이나 `math`처럼 관련 코드를 묶은 단위이며, 사용하기 전에 `import`로 필요한 코드를 가져옵니다.
- 클래스는 객체를 만드는 설계도이고, 객체는 그 설계도로 만든 개별 값이며, 변수는 값을 가리키는 이름입니다.
- 함수는 호출해서 실행하는 기능이고, 메서드는 객체나 클래스에 속한 함수입니다.
- 점(`.`)은 모듈·클래스·객체의 속성에 접근할 때 사용하며, 속성에는 값뿐 아니라 메서드도 포함됩니다.
- 함수나 메서드 이름 뒤의 `()`는 호출을 뜻하며, 괄호 안에는 실행에 필요한 입력값인 인수를 넣습니다.
- 빈 괄호 `()`는 아무것도 하지 않는다는 뜻이 아니라 별도 인수를 전달하지 않고 호출한다는 뜻입니다.
- `current.exists`는 메서드 자체를 가리키고, `current.exists()`는 그 메서드를 실행해 결과를 받습니다.
- `current.name`은 문자열이므로, `current.name()`처럼 함수로 호출하려고 하면 오류가 발생합니다.
- `Path("C:/dev")`는 경로 객체를 만들고, `list()`는 빈 리스트를 만듭니다.
- `text.replace("민수", "지수")`는 원래 문자열을 직접 수정하지 않고 해당 내용을 바꾼 새 문자열을 반환합니다.
- `Path.cwd().resolve()`는 먼저 현재 작업 폴더를 구한 뒤, 반환된 경로 객체의 `resolve()`를 실행합니다.
- `__init__.py`는 일반적인 파이썬 패키지를 구성하는 파일로 패키지를 처음 가져올 때 코드가 실행되며, 초기화할 내용이 없으면 비워 두어도 됩니다.
- 클래스 안의 `__init__()`는 새 객체의 이름이나 나이 같은 초기 상태를 설정하는 메서드로, 패키지의 `__init__.py`와는 역할이 다릅니다.
- `self`는 인스턴스 메서드에서 해당 객체 자신을 가리키는 매개변수에 관례적으로 붙이는 이름입니다.
- SQL의 `CHECK (age >= 0)`는 음수 나이가 저장되는 것을 막으며, 나이를 비워 두는 `NULL`도 막으려면 `NOT NULL`을 함께 지정해야 합니다.


##### 2. 파이썬 기초와 프로젝트 루트 탐색

**질문**

프로젝트 루트를 찾는 예제를 통해 `for`와 `while`의 차이, 조건문과 들여쓰기, 함수와 `break`·`return`의 역할을 설명해 주세요.

**답변**

프로젝트 루트는 **프로젝트 전체를 담는 기준 폴더**입니다. 현재 작업 폴더와 다를 수 있으므로, 이 예제에서는 현재 위치부터 부모 폴더를 차례로 확인하여 `data` 폴더가 있는 위치를 찾습니다.

- `Path.cwd()`는 파이썬의 현재 작업 폴더를 가져옵니다.
- `.resolve()`는 경로를 절대 경로로 정리합니다.
- `current / "data"`는 현재 경로 뒤에 `data`를 연결합니다. 실제 폴더를 생성하지는 않습니다.
- `.is_dir()`는 해당 경로가 실제 폴더인지 확인합니다.
- `current.parent`는 부모 폴더의 경로입니다.
- `current = current.parent`는 탐색용 변수의 값을 바꿉니다. 파이썬의 실제 작업 폴더를 이동시키지는 않습니다.

반복문과 함수의 역할은 다음과 같습니다.

| 문법 | 역할 |
|---|---|
| `for` | 여러 항목을 하나씩 꺼내 반복 |
| `while` | 조건이 참인 동안 반복 |
| `if` | 조건이 참일 때만 안쪽 코드 실행 |
| `def` | 재사용할 작업을 함수로 정의 |
| `break` | 가장 안쪽 반복문 종료 |
| `return 값` | 현재 함수 종료 및 결과값 전달 |
| `raise` | 오류 상황을 발생시켜 알림 |

**찾은 순간 함수의 일이 끝나고 경로를 전달해야 하므로 `return`을 사용합니다.** 함수 안에서도 `break`를 사용할 수 있지만, 반복문 밖에서 결과를 반환하는 코드가 따로 필요합니다.

파이썬은 들여쓰기로 코드의 소속을 구분합니다. `def`, `while`, `if` 아래의 코드는 각각 한 단계씩, 보통 4칸 들여씁니다.

**예시 코드**

```python
from pathlib import Path


def get_project_root():
    current = Path.cwd().resolve()

    while True:
        if (current / "data").is_dir():
            return current

        if current.parent == current:
            raise FileNotFoundError("data 폴더를 찾지 못했습니다.")

        current = current.parent


project_root = get_project_root()
data_dir = project_root / "data" / "raw"

print("프로젝트 루트:", project_root)
print("데이터 경로:", data_dir)
```

**실행 결과**

현재 작업 폴더가 `C:/dev/llm-data-analysis-course/notebooks/ch04`이고 프로젝트 루트 바로 아래에 `data` 폴더가 있다면 다음처럼 출력됩니다.

```text
프로젝트 루트: C:\dev\llm-data-analysis-course
데이터 경로: C:\dev\llm-data-analysis-course\data\raw
```

이 코드는 현재 위치와 부모 폴더만 탐색합니다. 그중 가장 먼저 발견한 `data` 폴더의 부모를 프로젝트 루트로 선택합니다.


##### 3. 파이썬 모듈 가져오기와 오류 해결

**질문**

하위 폴더의 노트북에서 `course_utils`를 찾지 못하고 `pd`가 정의되지 않았다는 오류가 발생하는 이유와 해결 방법은 무엇인가요?

**답변**

`ModuleNotFoundError: No module named 'course_utils'`는 **파이썬이 모듈 검색 경로에서 `course_utils`를 찾지 못했다**는 뜻입니다.

폴더가 탐색기에 보이더라도 파이썬이 그 위치를 자동으로 검색하는 것은 아닙니다.

다음 코드에서는 첫 번째 줄에서 오류가 나면 두 번째 줄이 실행되지 않습니다.

```python
from course_utils.paths import get_data_dir
import pandas as pd
```

그 상태에서 `pd.read_csv()`를 실행하면 `pd`가 정의되지 않았다는 `NameError`가 발생합니다.

- 먼저 `course_utils`를 가져오는 오류를 해결해야 합니다.
- 그다음 `import pandas as pd`를 실행해야 합니다.
- 다른 노트북에서 실행한 설정이나 변수는 자동으로 공유되지 않습니다.
- 같은 코드라도 현재 작업 폴더, 설치 상태, 노트북 커널이 다르면 실행 결과가 달라질 수 있습니다.

이번 실습에서는 `import sys` 대신 **Jupyter의 `%cd` 명령으로 현재 작업 폴더를 프로젝트 루트로 변경**할 수 있습니다. 노트북 파일 자체를 옮기는 것은 아닙니다.

**예시 코드**

실제 프로젝트 경로가 아래와 같고, 선택한 파이썬 환경에 `pandas`가 설치되어 있다는 기준입니다. 노트북 셀에서 실행하세요.

```python
%cd C:/dev/llm-data-analysis-course

from course_utils.paths import get_data_dir
import pandas as pd

print("불러오기 성공!")
print("데이터 폴더:", get_data_dir())
```

**실행 결과**

`get_data_dir()`가 프로젝트의 `data` 경로를 반환하도록 작성되어 있다면 다음 메시지가 표시됩니다. `%cd`의 경로 출력이 앞에 추가될 수 있습니다.

```text
불러오기 성공!
데이터 폴더: C:\dev\llm-data-analysis-course\data
```

`%cd`는 일반 파이썬 문법이 아니라 Jupyter에서 제공하는 명령입니다. 커널을 재시작한 뒤에는 준비 셀을 다시 실행하세요.


##### 4. PostgreSQL과 DBeaver의 역할

**질문**

PostgreSQL과 DBeaver는 각각 어떤 역할을 하며, 두 프로그램의 차이는 무엇인가요?

**답변**

**PostgreSQL은 데이터를 저장하고 SQL 명령을 처리하는 데이터베이스 관리 시스템(DBMS)**입니다.

**DBeaver는 데이터베이스에 연결하여 SQL을 실행하고 결과를 확인하는 관리 도구**입니다.

| 구분 | PostgreSQL | DBeaver |
|---|---|---|
| 주요 역할 | 데이터 저장 및 SQL 처리 | 데이터베이스 연결 및 조작 화면 제공 |
| 테이블 데이터 저장 | 담당함 | 연결된 PostgreSQL에 저장하도록 요청함 |
| SQL 사용 과정 | 명령을 받아 처리함 | 사용자가 작성한 명령을 서버로 전달함 |
| 결과 처리 | 조회 결과를 반환함 | 반환된 결과를 표 등으로 표시함 |

DBeaver의 SQL 편집기에서 명령을 실행하면 다음 순서로 동작합니다.

1. 사용자가 DBeaver에 SQL을 작성합니다.
2. DBeaver가 PostgreSQL로 SQL을 보냅니다.
3. PostgreSQL이 명령을 처리합니다.
4. DBeaver가 처리 결과를 화면에 보여줍니다.

**예시 코드**

```sql
SELECT 1 + 1 AS result;
```

- `SELECT`: 계산한 값을 보여 달라는 명령입니다.
- `1 + 1`: 계산할 식입니다.
- `AS result`: 결과 열의 이름을 `result`로 지정합니다.
- `;`: SQL 명령의 끝을 표시합니다.

**실행 결과**

| result |
|---|
| 2 |


##### 5. 주문과 주문상세의 관계 이해

**질문**

“주문 한 건에 메뉴 세 종류와 각각의 수량을 넣으라”는 실습은 어떤 뜻이며, 주문 건수·주문상세 건수·총수량은 어떻게 구분하나요?

**답변**

**주문 한 건은 고객이 한 번에 주문한 전체 거래**입니다. **주문상세는 그 주문에 포함된 개별 메뉴와 수량**입니다.

예를 들어 한 고객이 아메리카노 2잔, 카페라떼 1잔, 치즈케이크 3개를 한 번에 주문했다면 다음처럼 저장합니다.

- `orders`: 주문한 고객과 주문 상태 등을 **1행** 저장합니다.
- `order_items`: 세 종류의 메뉴와 각각의 수량을 **3행** 저장합니다.
- 주문상세 3행의 `order_id`는 모두 같아야 합니다.
- 총수량은 `2 + 1 + 3 = 6`개입니다.

| 구분 | 값 |
|---|---|
| 주문 건수 | 1건 |
| 주문상세 건수 | 3건 |
| 총수량 | 6개 |

주문상세의 `order_id`는 주문 테이블을 참조하는 **외래키(FK)**입니다. 존재하지 않는 주문 번호를 넣으면 오류가 발생합니다.

**예시 코드**

고객 번호 `1`이 존재한다고 가정하고, 먼저 주문을 생성합니다.

```sql
INSERT INTO cafe_practice.orders (customer_id, status)
VALUES (1, '접수')
RETURNING order_id;
```

반환된 주문 번호가 `7`이라고 가정합니다. 메뉴 번호 `2`, `3`, `19`가 존재하고 판매 당시 단가가 아래와 같다면 다음을 실행합니다.

**실제 주문 번호가 다르면 세 행의 첫 번째 숫자 `7`을 모두 바꿔야 합니다.**

```sql
INSERT INTO cafe_practice.order_items
    (order_id, menu_id, quantity, unit_price)
VALUES
    (7, 2, 2, 4000),
    (7, 3, 1, 4500),
    (7, 19, 3, 6500);
```

해당 주문의 상세 건수, 총수량, 총금액을 확인합니다.

```sql
SELECT
    order_id,
    COUNT(*) AS detail_count,
    SUM(quantity) AS total_quantity,
    SUM(quantity * unit_price) AS total_amount
FROM cafe_practice.order_items
WHERE order_id = 7
GROUP BY order_id;
```

**실행 결과**

해당 주문에 위의 상세 3행만 입력했다면 다음과 같습니다.

| order_id | detail_count | total_quantity | total_amount |
|---|---|---|---|
| 7 | 3 | 6 | 32000 |

`unit_price`는 메뉴 한 개의 판매 당시 가격입니다. 상세별 금액은 `quantity × unit_price`로 계산합니다.

`INSERT`를 다시 실행하면 데이터가 추가되므로, 입력 후 확인할 때는 `SELECT`만 실행하세요.
````
