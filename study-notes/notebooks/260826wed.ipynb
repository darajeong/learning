##### 1. 함수·모듈·패키지와 가상환경에서 새롭게 이해한 내용

- `python -m venv .venv`는 현재 프로젝트에 `.venv`라는 가상환경을 만들고, `.venv\Scripts\activate`는 그 가상환경을 활성화하는 명령어이다.
- `greet()`처럼 함수 안에서 `print()`만 실행하고 `return`을 작성하지 않은 함수도 실제로는 `None`을 반환한다.
- 함수 정의문의 끝에는 콜론(`:`)을 반드시 붙이며, 그 아래 들여쓰기된 코드가 함수의 실행 영역이 된다.
- 함수에서 `return`이 실행되면 결과를 함수 밖으로 반환하면서 함수가 즉시 종료되므로, 같은 함수 안에서 `return` 아래에 있는 코드는 실행되지 않는다.
- 함수 안에서 사용하는 변수는 **지역변수(Local Variable)**이고 함수 밖에서 만든 변수는 **전역변수(Global Variable)**이며, 혼동을 줄이기 위해 서로 구분되는 이름을 사용하는 것이 좋다.
- `first_sum()` 함수 안의 `result`는 지역변수이고 함수 밖의 `ret`은 반환값을 저장하는 별도의 변수이다.
- `return total, difference`처럼 여러 값을 쉼표로 구분해 반환하면 파이썬이 자동으로 하나의 튜플로 묶어 반환한다.
- **모듈(Module)**은 공통 함수 등을 작성한 하나의 `.py` 파일이며, 다른 파일에서 `import`하여 재사용할 수 있다.
- `import`는 모듈 전체를 가져오고 `from 모듈 import 함수`는 모듈에서 필요한 함수만 가져오는 방식이며, 표준·외부·직접 만든 모듈에 모두 사용할 수 있다.
- `from math import sqrt`에서 `math`는 **모듈**이고 `sqrt`는 그 모듈 안에 정의된 **함수**이며 메서드가 아니다.
- `import price`처럼 사용하면 `import` 뒤의 `price`는 일반적으로 불러올 모듈, 즉 `price.py`의 파일명을 의미한다.
- `__pycache__`는 파이썬이 모듈을 빠르게 불러오기 위해 자동으로 만드는 캐시 폴더이며, 일반적으로 `.gitignore`에 추가해 GitHub에 올리지 않는다.
- **패키지(Package)**는 관련된 여러 모듈을 하나의 폴더로 묶어 관리하는 구조이며, `shop` 폴더는 여러 쇼핑 관련 모듈을 포함하는 패키지가 될 수 있다.
- `__init__.py`는 해당 폴더가 파이썬 패키지임을 명확하게 나타내고 패키지를 초기화할 때 사용하는 파일이다.
- 모듈 안에는 여러 함수가 들어갈 수 있으며, 모듈 안에 있다고 해서 메서드가 되는 것은 아니고 클래스에 소속된 함수를 메서드라고 한다.
- `from shop.price import calculate_total, apply_discount`에서 `shop`은 패키지, `price`는 모듈, `calculate_total`과 `apply_discount`는 모듈 안의 함수이다.

##### 2. `def` 함수란 무엇인가요?

- **질문:** 파이썬에서 `def`는 무엇을 의미하나요?
- **답변:** `def`는 새로운 함수를 정의할 때 사용하는 파이썬 키워드로, 여러 줄의 코드를 하나로 묶고 이름을 붙여 필요할 때 반복해서 실행할 수 있게 한다.

```python
def greet():
    print("안녕하세요!")

greet()
```

**실행 결과**

```text
안녕하세요!
```

- `def`는 함수를 만들겠다는 의미이다.
- `greet`은 사용자가 정한 함수 이름이다.
- `greet()`은 만들어 놓은 함수를 실행하는 함수 호출이다.

##### 3. `greet()` 함수를 두 번 호출하면 어떻게 실행되나요?

- **질문:** `greet()`을 두 번 작성하면 어떤 결과가 나오며 어떤 순서로 실행되나요?
- **답변:** 함수 안의 코드는 호출할 때마다 실행되므로, `greet()`을 두 번 호출하면 함수 안의 두 `print()`도 각각 두 번 실행된다.

```python
def greet():
    print("안녕하세요!")
    print("파이썬 공부를 시작합니다.")

greet()
greet()
```

**실행 결과**

```text
안녕하세요!
파이썬 공부를 시작합니다.
안녕하세요!
파이썬 공부를 시작합니다.
```

**실행 순서**

1. `def greet():`로 함수를 정의한다.
2. 첫 번째 `greet()`이 함수 안의 코드를 실행한다.
3. 두 번째 `greet()`이 함수 안의 코드를 다시 실행한다.

##### 4. 이름과 공부 과목을 각각 다르게 출력하려면 어떻게 하나요?

- **질문:** 홍길동은 파이썬, 홍길래는 SQL을 공부한다고 각각 다르게 출력하려면 어떻게 해야 하나요?
- **답변:** 함수에 `name`과 `subject`라는 두 개의 파라미터를 만들고, 함수를 호출할 때 이름과 과목을 전달하면 된다.

```python
def greet(name, subject):
    print(f"안녕하세요! {name}님")
    print(f"{subject} 공부를 시작합니다.")

greet("홍길동", "파이썬")
greet("홍길래", "SQL")
```

**실행 결과**

```text
안녕하세요! 홍길동님
파이썬 공부를 시작합니다.
안녕하세요! 홍길래님
SQL 공부를 시작합니다.
```

**값이 전달되는 구조**

```text
greet("홍길동", "파이썬")
       ↓          ↓
     name       subject
```

##### 5. 함수가 값을 파라미터로 받는다는 것은 무슨 뜻인가요?

- **질문:** 함수가 두 수 `a`와 `b`를 파라미터로 받는다는 것은 무슨 의미인가요?
- **답변:** 함수를 호출할 때 전달한 값을 함수 안에서 사용할 수 있도록 변수에 담아 받는다는 뜻이다.

```python
def calc(a, b):
    print(a + b)

calc(10, 5)
```

**값이 전달되는 구조**

```text
calc(10, 5)
      ↓   ↓
      a   b
```

함수 안에서는 다음과 같은 상태가 된다.

```text
a = 10
b = 5
```

**실행 결과**

```text
15
```

##### 6. `parameter`와 `argument`의 차이는 무엇인가요?

- **질문:** `parameter`와 `argument`는 각각 무엇이며 어떤 차이가 있나요?
- **답변:** `parameter`는 함수를 정의할 때 값을 받기 위해 만든 변수이고, `argument`는 함수를 호출할 때 전달하는 실제 값이다.

```python
def calc(a, b):
    return a + b

result = calc(10, 5)
```

- `a`, `b` → **Parameter(매개변수)**
- `10`, `5` → **Argument(인수 또는 인자)**

| 구분 | 작성 위치 | 예시 |
|---|---|---|
| Parameter | 함수를 정의할 때 | `a`, `b` |
| Argument | 함수를 호출할 때 | `10`, `5` |

> 함수 정의에서 값을 받을 이름은 `parameter`, 함수 호출에서 전달하는 실제 값은 `argument`이다.

##### 7. 튜플이란 무엇인가요?

- **질문:** 파이썬의 튜플은 어떤 자료형인가요?
- **답변:** 튜플은 여러 값을 순서대로 하나에 묶어 저장하는 자료형이며, 한 번 만들어진 튜플의 값은 일반적으로 수정할 수 없다.

```python
numbers = (10, 20, 30)

print(numbers)
print(numbers[0])
```

**실행 결과**

```text
(10, 20, 30)
10
```

| 구분 | 리스트 | 튜플 |
|---|---|---|
| 기호 | `[ ]` | `( )` |
| 값 수정 | 가능 | 불가능 |
| 예시 | `[10, 20]` | `(10, 20)` |

함수에서 여러 값을 쉼표로 구분하여 반환하면 파이썬이 자동으로 튜플로 묶어 준다.

```python
def calc(a, b):
    return a + b, a - b, a * b, a / b

result = calc(10, 5)
print(result)
```

**실행 결과**

```text
(15, 5, 50, 2.0)
```

##### 9. `.venv`는 어떤 것을 저장하는 공간인가요?

- **질문:** `.venv`는 코드를 실행하는 데 필요한 것들을 저장하는 공간인가요?
- **답변:** `.venv`는 작성한 코드를 저장하는 곳이 아니라, 코드를 실행하는 데 필요한 프로젝트 전용 파이썬과 외부 패키지 등을 저장하는 폴더이다.

```text
python-basics
├── .venv
│   ├── Scripts
│   ├── Lib
│   │   └── site-packages
│   └── pyvenv.cfg
└── chapter15
    └── study.py
```

- `Scripts` → Python, `pip`, 가상환경 활성화 파일 등이 들어 있다.
- `Lib/site-packages` → `pip`으로 설치한 외부 패키지가 들어 있다.
- `pyvenv.cfg` → 가상환경의 설정 정보가 들어 있다.
- `study.py` → 사용자가 직접 작성한 코드 파일이다.

> `.venv`는 코드 저장소가 아니라 코드를 실행하기 위한 프로젝트 전용 도구 상자이다.

##### 10. `.venv\Scripts\activate`는 무슨 명령어인가요?

- **질문:** PowerShell에서 `.venv\Scripts\activate`를 입력하는 이유는 무엇인가요?
- **답변:** 현재 프로젝트에 만들어 놓은 파이썬 가상환경을 활성화하는 명령어이다.

```powershell
.venv\Scripts\activate
```

명령어의 구조는 다음과 같다.

```text
.venv       → 가상환경 폴더
Scripts     → 실행 관련 파일이 들어 있는 폴더
activate    → 가상환경을 활성화하는 파일
```

**활성화 전**

```text
PS C:\dev\python-basics>
```

**활성화 후**

```text
(.venv) PS C:\dev\python-basics>
```

앞에 `(.venv)`가 나타나면 가상환경이 정상적으로 활성화된 것이다.

가상환경을 종료할 때는 다음 명령어를 사용한다.

```powershell
deactivate
```

##### 11. 터미널의 동그라미와 별표 모양은 어떤 차이가 있나요?

- **질문:** VS Code 터미널 앞에 나타나는 동그라미와 별표처럼 생긴 표시는 무엇인가요?
- **답변:** 해당 표시는 파이썬 문법이 아니라 VS Code 터미널 테마가 명령어 입력 줄을 구분하기 위해 보여 주는 디자인이다.

```text
● PS C:\dev\python-basics>
❖ (.venv) PS C:\dev\python-basics>
```

- `●` → 이전에 명령어를 입력했던 줄을 나타낼 수 있다.
- `❖` → 현재 명령어를 입력할 줄을 나타낼 수 있다.
- `PS` → PowerShell 터미널을 의미한다.
- `C:\dev\python-basics` → 터미널의 현재 폴더 위치이다.
- `(.venv)` → 파이썬 가상환경이 활성화되었다는 의미이다.

동그라미와 별표의 모양은 터미널 설정이나 테마에 따라 달라질 수 있으며 코드 실행에는 영향을 주지 않는다.

##### 12. 대시보드에 AI를 연결할 때 API 키가 필요한가요?

- **질문:** 대시보드에서 외부 AI를 사용하려면 API 키가 필요한가요?
- **답변:** 대시보드에서 Gemini, GPT, Claude와 같은 외부 AI 서비스를 호출하려면 일반적으로 해당 서비스에서 발급받은 API 키가 필요하다.

```text
사용자 입력
    ↓
대시보드 화면
    ↓
백엔드 서버
    ↓ API 키를 이용한 인증
AI 서비스
    ↓
AI 답변을 대시보드에 표시
```

API 키는 코드에 직접 작성하면 안 된다.

```python
# 사용하면 안 되는 방식
api_key = "실제_API_키"
```

대신 `.env` 파일에 보관한다.

```text
OPENAI_API_KEY=발급받은_API_키
```

파이썬에서는 환경변수로 불러온다.

```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")
```

API 키가 GitHub에 올라가지 않도록 `.gitignore`에 다음 내용을 추가한다.

```gitignore
.env
```

##### 13. Gemini, GPT, Claude 모두 API 키가 있나요?

- **질문:** Gemini, GPT, Claude 모두 API 키가 있으며, 대시보드에는 보통 Gemini를 연결하나요?
- **답변:** 세 서비스 모두 API와 API 키를 제공하며, 사용하려는 AI 서비스에 맞는 키를 각각 발급받아야 한다.

| AI 모델 | 제공 회사 | 환경변수 이름 예시 |
|---|---|---|
| Gemini | Google | `GEMINI_API_KEY` |
| GPT | OpenAI | `OPENAI_API_KEY` |
| Claude | Anthropic | `ANTHROPIC_API_KEY` |

하나의 API 키로 세 가지 AI를 모두 사용할 수 있는 것은 아니다.

```text
Gemini API 키 → Gemini 호출
OpenAI API 키 → GPT 호출
Anthropic API 키 → Claude 호출
```

대시보드가 반드시 Gemini를 사용해야 하는 것은 아니며, 대시보드는 사용자에게 정보를 보여 주는 화면이고 어떤 AI를 연결할지는 개발자가 선택할 수 있다.

초급 수업이나 팀 프로젝트에서는 다음과 같은 이유로 Gemini를 사용하는 경우가 있다.

- Google AI Studio에서 API 키를 비교적 쉽게 발급할 수 있다.
- 간단한 코드로 AI 기능을 연결할 수 있다.
- 일부 모델은 제한된 무료 사용량을 제공할 수 있다.
- 프로토타입이나 수업용 프로젝트에 접근하기 편리하다.

GPT와 Claude도 대시보드에 같은 방식으로 연결할 수 있으며, 비용·기능·응답 품질·프로젝트 목적을 고려하여 선택해야 한다.
