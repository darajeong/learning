##### 1. 딕셔너리란 무엇인가?

딕셔너리(Dictionary)는 데이터를 **`키(key): 값(value)` 형태로 연결하여 저장하는 자료형**이다.

예를 들어 다음과 같은 관계를 저장할 수 있다.

* 학생: 수업을 통해 공부하는 사람
* 선생님: 다른 사람을 가르치는 사람
* 파이썬: 프로그래밍 언어

이를 파이썬 딕셔너리로 표현하면 다음과 같다.

```python
words = {
    "학생": "수업을 통해 공부하는 사람",
    "선생님": "다른 사람을 가르치는 사람",
    "파이썬": "프로그래밍 언어"
}
```

* 왼쪽에 있는 `"학생"`, `"선생님"`, `"파이썬"`은 **키(key)**이다.
* 오른쪽에 있는 설명은 **값(value)**이다.
* 딕셔너리는 중괄호 `{}`를 사용하여 만든다.

##### 2. 키와 값은 무엇인가?

딕셔너리는 다음과 같은 `키: 값`의 구조로 이루어진다.

```python
student = {
    "name": "지현",
    "age": 24,
    "score": 90
}
```

* 키: `"name"`, `"age"`, `"score"`
* 값: `"지현"`, `24`, `90`

키는 해당 값을 찾기 위한 이름표 역할을 한다.

```python
print(student["name"])
```

```text
사과
```

딕셔너리는 위치 번호가 아니라 사용자가 정한 **키**로 값을 찾는다.

```python
student = {
    "name": "지현",
    "score": 90
}

print(student["name"])
```

출력 결과:

```text
지현
```

| 자료형 | 값을 찾는 기준 | 예시 |
|---|---|---|
| 리스트 | 위치 번호 | `fruits[0]` |
| 튜플 | 위치 번호 | `fruits[0]` |
| 딕셔너리 | 키 | `student["name"]` |

##### 6. 리스트와 튜플의 차이는 무엇인가?

리스트와 튜플은 모두 여러 값을 순서대로 저장하는 자료형이다.

가장 큰 차이는 **저장된 값을 변경할 수 있는지**이다.

| 구분 | 리스트 | 튜플 |
|---|---|---|
| 기호 | `[]` | `()` |
| 값 수정 | 가능 | 불가능 |
| 값 추가·삭제 | 가능 | 불가능 |
| 값 접근 방법 | 인덱스 | 인덱스 |

리스트는 값을 수정할 수 있다.

```python
fruits = ["사과", "바나나"]
fruits[1] = "딸기"
```

튜플은 만든 후 값을 수정할 수 없다.

```python
fruits = ("사과", "바나나")
```

다음과 같이 튜플의 값을 수정하려고 하면 오류가 발생한다.

```python
fruits[1] = "딸기"
```

##### 7. 메서드란 무엇인가?

```python
변수이름.메서드()
```

예를 들어 문자열의 `split()` 메서드는 문자열을 쪼개는 기능을 한다.

* `split()`: 문자열을 쪼개는 메서드

메서드 앞에는 일반적으로 해당 기능을 사용할 **객체나 변수 이름**이 나온다.

##### 8. 딕셔너리의 `items()` 메서드는 무엇인가?

`items()`는 딕셔너리의 키와 값을 함께 가져오는 메서드이다.

```python
student = {
    "name": "지현",
    "age": 24
}

print(student.items())
```

각 항목은 다음과 같이 `(키, 값)` 형태의 튜플로 나온다.

```python
("name", "지현")
("age", 24)
```

반복문에서는 키와 값을 각각 다른 변수로 받을 수 있다.

```python
for key, value in student.items():
    print(key, value)
```

출력 결과:

```text
name 지현
age 24
```

딕셔너리와 튜플은 서로 다른 자료형이다.

하지만 딕셔너리의 `items()`를 사용하면 키와 값이 다음처럼 튜플로 묶여서 나온다.

반복문에서는 이 튜플을 두 변수로 나누어 받을 수 있다.

이처럼 튜플에 들어 있는 값을 여러 변수로 나누어 받는 것을 **튜플 언패킹**이라고 한다.

##### 10. `student["name"]`과 `student.get("name")`의 차이는 무엇인가?

두 방법 모두 딕셔너리에서 키에 해당하는 값을 가져온다.

```python
student = {
    "name": "지현",
    "score": 90
}
```

대괄호를 사용하는 방법:

```python
print(student["name"])
```

`get()`을 사용하는 방법:

```python
print(student.get("score"))
```

차이는 존재하지 않는 키를 찾았을 때 나타난다.

| 방법 | 키가 있을 때 | 키가 없을 때 |
|---|---|---|
| `student["name"]` | 해당 값 반환 | `KeyError` 발생 |
| `student.get("name")` | 해당 값 반환 | `None` 반환 |
| `student.get("name", "정보 없음")` | 해당 값 반환 | 지정한 기본값 반환 |

```python
print(student.get("age", "정보 없음"))
```

출력 결과:

```text
정보 없음
```

* 반드시 존재해야 하는 키라면 `[]`를 사용할 수 있다.
* 키가 없을 가능성이 있다면 `get()`을 사용하는 것이 안전하다.

##### 14. `if word in word_count`는 무슨 의미인가?

현재 단어가 딕셔너리의 **키로 이미 저장되어 있는지 확인하는 조건**이다.

```python
word_count = {
    "사과": 1
}
```

현재 `word`가 `"사과"`라면:

```python
if word in word_count:
```

실제로 다음 조건을 확인하는 것과 같다.

```python
if "사과" in word_count:
```

* 단어가 이미 있으면 기존 값에 `1`을 더한다.
* 처음 나온 단어라면 값을 `1`로 저장한다.

##### 15. `word_count[word]`는 무슨 의미인가?

`word_count[word]`는 딕셔너리에서 **현재 단어에 연결된 값**, 즉 등장 횟수를 가져오는 코드이다.

```python
word_count = {
    "사과": 3,
    "바나나": 2
}
```

현재 `word`가 `"사과"`라면:

```python
word_count[word]
```

실제로는 다음과 같다.

```python
word_count["사과"]
```

결과:

```text
3
```

정리하면 다음과 같다.

```text
word_count       → 딕셔너리 전체
word              → 현재 확인하는 단어
word_count[word] → 현재 단어의 등장 횟수
```

`word_count[word]`가 직접 개수를 세는 것은 아니다. 딕셔너리에 저장된 현재 단어의 개수를 가져오는 역할을 한다.

개수를 실제로 증가시키는 코드는 다음과 같다.

```python
word_count[word] = word_count[word] + 1
```

##### 16. 대괄호 `[]`는 딕셔너리에서만 사용하는가?

대괄호는 딕셔너리뿐만 아니라 리스트, 튜플, 문자열에서도 값을 가져올 때 사용한다.

| 자료형 | 대괄호 안에 넣는 것 | 예시 |
|---|---|---|
| 리스트 | 위치 번호 | `fruits[0]` |
| 튜플 | 위치 번호 | `colors[1]` |
| 문자열 | 위치 번호 | `text[0]` |
| 딕셔너리 | 키 | `word_count["사과"]` |

리스트에서는 위치 번호를 사용한다.

```python
fruits = ["사과", "바나나"]

print(fruits[0])
```

딕셔너리에서는 키를 사용한다.

```python
word_count = {"사과": 3}

print(word_count["사과"])
```

즉, 대괄호 `[]`는 여러 자료형에서 값을 가져올 때 사용하지만, 딕셔너리에서는 위치 번호가 아닌 **키**를 넣는다.

##### 18. 디버깅이란 무엇인가?

디버깅(Debugging)은 프로그램이 제대로 작동하지 않을 때 **오류의 원인을 찾아 수정하는 과정**이다.

디버깅할 때는 다음 내용을 확인한다.

* 코드에 오타가 있는지 확인한다.
* 콜론 `:`이 빠지지 않았는지 확인한다.
* 들여쓰기가 올바른지 확인한다.
* 괄호나 따옴표가 빠지지 않았는지 확인한다.
* 변수에 예상한 값이 들어 있는지 확인한다.
* 파일 이름과 경로가 정확한지 확인한다.
* 오류 메시지가 무엇을 알려 주는지 확인한다.

코드가 실행되지 않는 경우뿐만 아니라, 코드가 실행되더라도 결과가 예상과 다를 때 원인을 찾는 것도 디버깅이다.

##### 19. VS Code 파일 옆의 `U`와 `M`은 무엇인가?

VS Code에서 파일 오른쪽에 표시되는 `U`와 `M`은 Git의 파일 상태를 의미한다.

| 표시 | 의미 | 설명 |
|---|---|---|
| `U` | Untracked | 새로 만들었지만 아직 Git에 등록하지 않은 파일 |
| `M` | Modified | Git에 등록된 기존 파일의 내용을 수정한 상태 |

* `U`는 새로 만든 파일을 의미한다.
* `M`은 기존 파일을 수정했다는 의미이다.
* 이는 오류 표시가 아니며 파이썬 코드 실행에는 영향을 주지 않는다.
