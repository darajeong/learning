##### 1. 변수와 기본 데이터형

`int`는 소수점이 없는 정수를 나타낸다.

```python
10
-3
2026
```

`float`는 소수점이 있는 실수를 나타낸다.

```python
3.14
-0.5
180.0
```

`str`은 따옴표로 감싼 문자 데이터를 나타낸다. 숫자가 들어 있어도 따옴표로 감싸면 문자열이다.

```python
"Python"
"20"
"서울"
```

`bool`은 참과 거짓을 나타내는 논리값이다.

```python
True
False
```

변수에 여러 데이터형의 값을 저장했다.

```python
age = 20
height = 175.5
name = "문길동"
is_student = True
```

`print()`와 `type()`은 Python에서 기본적으로 제공하는 내장 함수이다. `type()`을 사용하면 변수에 저장된 값의 데이터형을 확인할 수 있다.

```python
print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

실행 결과:

```text
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

변수 이름이 여러 단어로 구성된 경우 단어 사이에 언더바(`_`)를 넣어 작성할 수 있다.

```python
is_student = True
```

##### 2. 대입과 비교

`=`는 같다는 의미가 아니라 오른쪽 값을 왼쪽 변수에 저장하는 **대입 연산자**이다.

```python
age = 20
```

두 값이 같은지 비교할 때는 `==` 비교 연산자를 사용한다.

```python
age == 20
```

Python 코드는 위에서 아래로 한 줄씩 실행된다. 같은 변수에 새로운 값을 여러 번 대입하면 기존 값이 변경된다. 따라서 마지막으로 `100`을 대입한 뒤 출력하면 `100`이 나온다.

##### 3. 데이터형 확인과 형 변환

데이터형을 확인할 때는 `type()` 내장 함수를 사용한다.

```python
print(type(name))
```

`int()`를 사용하면 정수로 바꿀 수 있는 값을 `int` 데이터형으로 변환할 수 있다.

```python
value1 = int("30")
```

`int("30")`은 문자열 `"30"`을 정수 `30`으로 바꾸겠다는 의미이다.

##### 4. 산술 연산자

`//`는 나눗셈의 몫을 구한다.

```python
17 // 5
```

실행 결과:

```text
3
```

`%`는 나눗셈의 나머지를 구한다.

```python
17 % 5
```

실행 결과:

```text
2
```

`**`는 거듭제곱을 계산한다.

```python
2 ** 3
```

실행 결과:

```text
8
```

몫과 나머지를 이용해 125분을 시간과 분으로 변환했다.

```python
total_minutes = 125

hours = total_minutes // 60
minutes = total_minutes % 60

print(hours, "시간")
print(minutes, "분")
```

실행 결과:

```text
2 시간
5 분
```

##### 5. 복합 대입 연산자

`score += 5`는 기존 `score`에 `5`를 더한 결과를 다시 `score`에 저장한다는 의미이다.

```python
score += 5
```

다음 코드와 같은 의미이다.

```python
score = score + 5
```

복합 대입 연산자를 직접 사용해 봤다.

```python
count = 10

count += 2
print(count)

count -= 5
print(count)
```

실행 결과:

```text
12
7
```

##### 6. 문자열과 따옴표

큰따옴표로 만든 문자열 안에는 작은따옴표를 사용할 수 있고, 작은따옴표로 만든 문자열 안에는 큰따옴표를 사용할 수 있다.

```python
"문길동은 'Python'을 공부한다."
'문길동은 "Python"을 공부한다.'
```

문자열을 감싼 따옴표와 같은 종류의 따옴표를 문자열 안에 그대로 사용하면 `SyntaxError`가 발생할 수 있다.

이때는 바깥쪽에 다른 따옴표를 사용하거나 이스케이프 문자를 사용해야 한다.

`len()`은 문자열 등의 길이를 확인할 때 사용하는 내장 함수이다.

##### 7. 문자열 인덱싱

문자열 `"Python"`의 각 문자에는 `0`부터 시작하는 인덱스가 있다.

```text
문자:    P  y  t  h  o  n
인덱스:  0  1  2  3  4  5
```

인덱스를 사용해 특정 위치의 문자를 가져올 수 있다.

```python
word = "Python"

print(word[0])
print(word[1])
print(word[5])
```

실행 결과:

```text
P
y
n
```

음수 인덱스는 문자열의 뒤에서부터 센다.

```python
print(word[-1])
print(word[-2])
```

실행 결과:

```text
n
o
```

##### 8. 문자열 슬라이싱

슬라이싱에서는 시작 위치를 포함하지만 끝 위치는 포함하지 않는다.

```python
word = "Python"

print(word[0:2])
print(word[2:6])
print(word[:3])
print(word[3:])
```

실행 결과:

```text
Py
thon
Pyt
hon
```

##### 9. 문자열 메서드

`upper()`는 영문자를 대문자로 바꾼 새로운 문자열을 만든다.

```python
text = "Python"
print(text.upper())
```

`lower()`는 영문자를 소문자로 바꾼 새로운 문자열을 만든다.

```python
text = "PYTHON"
print(text.lower())
```

`strip()`은 문자열 양쪽 끝의 공백을 제거한다.

```python
name = "  문길동  "
print(name.strip())
```

`replace()`는 문자열의 일부를 바꾼 새로운 문자열을 만든다.

```python
message = "I like Java"
new_message = message.replace("Java", "Python")

print(message)
print(new_message)
```

실행 결과:

```text
I like Java
I like Python
```

`replace()`를 실행해도 기존 `message`는 그대로 유지되고, 변경된 결과가 `new_message`에 저장된다.

##### 10. 문자열 검색

`find()`는 찾는 문자열이 처음 나타나는 위치를 반환한다.

`count()`는 해당 문자열이 몇 번 나타나는지 반환한다.

```python
text = "banana"

print(text.find("na"))
print(text.count("a"))
print(text.find("tan"))
```

실행 결과:

```text
2
3
-1
```

`find()`의 결과가 `-1`이면 찾는 내용이 문자열에 없다는 의미이다.

##### 11. 문자열과 숫자 연결

문자열과 숫자는 데이터형이 다르기 때문에 `+`로 바로 연결할 수 없다. 숫자를 `str()`로 변환한 뒤 연결해야 한다.

```python
name = "문길동"
age = 20

print("이름: " + name + ", 나이: " + str(age))
```

매번 `str()`을 사용하는 것이 번거롭다면 f-string을 사용할 수 있다.

문자열 앞에 `f`를 붙이고 변수 이름을 `{}` 안에 넣으면 숫자를 따로 문자열로 변환하지 않아도 된다.

```python
print(f"이름: {name}, 나이: {age}")
```

##### 12. 이스케이프 문자

이스케이프 문자를 사용하면 문자열 안에서 줄바꿈이나 탭, 따옴표 등을 표현할 수 있다.

```text
\n  → 줄바꿈
\t  → 탭
\"  → 큰따옴표
\\  → 역슬래시 자체
```

사용 예시:

```python
print("이름: 문길동\n도시: 서울")
print("Python\tAI")
```

실행 결과:

```text
이름: 문길동
도시: 서울
Python  AI
```
