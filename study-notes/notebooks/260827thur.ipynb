##### 1. 파일·CSV 데이터와 오류·예외 처리

**새롭게 이해한 내용**

###### Chapter 16. 파일과 CSV 데이터 다루기

- CSV는 여러 값을 쉼표로 구분하여 행과 열 형태의 데이터를 저장하는 텍스트 파일 형식이며, 이름은 `Comma-Separated Values`의 약자이다.
- CSV는 판다스 전용 파일이 아니며, 기본 파일 기능이나 `csv` 모듈로도 처리할 수 있지만 판다스를 사용하면 표 형태의 데이터를 편리하게 분석할 수 있다.
- `read()`, `readline()`처럼 객체 뒤에 점(`.`)을 붙여 호출하는 함수는 메서드라고 한다.
- `read()`는 파일 전체를 한 번에 읽고, `readline()`은 호출할 때마다 한 줄씩 읽으며, `readlines()`는 모든 줄을 리스트로 읽는다.
- `with open()`을 사용하면 작업이 끝난 뒤 파일을 자동으로 닫아 주므로 `open()`과 `close()`를 따로 사용하는 것보다 안전하고 편리하다.
- `encoding="utf-8"`은 한국어·영어·중국어·일본어 등 다양한 문자를 표현할 수 있지만, 파일을 저장할 때 사용한 인코딩과 읽을 때 지정한 인코딩이 일치해야 글자가 깨지지 않는다.
- 글자가 이상하게 표시되는 현상은 보통 “한글이 깨졌다”보다 “파일의 인코딩이 맞지 않는다”라고 표현하는 것이 정확하다.
- 백슬래시(`\`)는 이스케이프 문자의 시작을 나타내므로 `\n`은 줄바꿈, `\t`는 탭, `\\`는 백슬래시 자체를 의미한다.
- 파이썬에서 Windows 파일 경로는 `C:\\folder\\file.txt`, `r"C:\folder\file.txt"` 또는 `C:/folder/file.txt`처럼 작성할 수 있다.
- `split()`은 문자열을 공백 기준으로 나누고, `split(",")`은 쉼표를 기준으로 나눈다.
- 판다스 데이터프레임의 `info()`는 열, 자료형, 결측값 등의 정보를 출력하는 메서드이므로 괄호를 사용한다.
- 판다스 데이터프레임의 `shape`는 행과 열의 개수를 `(행, 열)`로 알려 주는 속성이므로 괄호를 사용하지 않는다.
- 가비지 컬렉터는 더 이상 사용하지 않는 객체의 메모리를 자동으로 정리하는 기능으로 Python·Java·C# 등에 있으며, C에서는 개발자가 메모리를 직접 해제해야 하는 경우가 많다.

###### Chapter 17. 오류와 예외 처리

- `try-except`는 오류 자체를 무조건 없애는 기능이 아니라, 실행 중 예상한 예외가 발생했을 때 프로그램이 갑자기 중단되지 않도록 대응하는 기능이다.
- `try`에는 예외가 발생할 가능성이 있는 코드를 작성하고, `except`에는 해당 예외가 발생했을 때 실행할 대응 코드를 작성한다.
- `try` 안에는 여러 코드를 작성할 수 있지만, 어떤 코드에서 예외가 발생했는지 쉽게 파악할 수 있도록 필요한 범위만 넣는 것이 좋다.
- `ValueError`는 값을 원하는 자료형으로 변환할 수 없을 때 발생하므로 입력값의 형식을 먼저 확인한다.
- `ZeroDivisionError`는 숫자를 0으로 나눌 때 발생하므로 나누는 값이 0인지 먼저 확인한다.
- `FileNotFoundError`는 열려는 파일을 찾을 수 없을 때 발생하므로 파일 이름과 경로를 확인한다.
- `KeyError`는 딕셔너리에 존재하지 않는 키를 사용할 때 발생하므로 키의 이름과 존재 여부를 확인한다.
- `IndexError`는 리스트 등의 범위를 벗어난 인덱스를 사용할 때 발생하므로 데이터의 길이와 인덱스를 확인한다.
- `while True`는 조건이 항상 참이므로 계속 반복되며, `break`, `return` 또는 예외 등 반복을 끝내는 코드가 실행되어야 종료된다.

##### 2. `try`, `except`, `else`, `finally`의 역할

**질문**

`try·except·else·finally` 코드는 무엇을 실행하며, 각 부분은 어떤 의미인가요?

**답변**

- `try`: 오류가 발생할 가능성이 있는 코드를 실행합니다.
- `except`: `try`에서 특정 오류가 발생하면 실행됩니다.
- `else`: 오류가 발생하지 않았을 때 실행됩니다.
- `finally`: 오류 발생 여부와 관계없이 마지막에 항상 실행됩니다.

**예시 코드**

```python
try:
    number = int(input("정수: "))
except ValueError:
    print("정수가 아닙니다.")
else:
    print(f"입력 성공: {number}")
finally:
    print("입력 처리를 마쳤습니다.")
```

**실행 결과**

정수 `10`을 입력한 경우:

```text
정수: 10
입력 성공: 10
입력 처리를 마쳤습니다.
```

문자열 `안녕`을 입력한 경우:

```text
정수: 안녕
정수가 아닙니다.
입력 처리를 마쳤습니다.
```

##### 3. `while True` 반복 횟수와 종료 조건

**질문**

`while True`로 점수를 입력받는 코드는 몇 번 반복되며, 왜 그렇게 반복되나요?

**답변**

`while True`는 기본적으로 무한 반복하지만, `0~100` 사이의 정수가 입력되면 `break`가 실행되어 종료됩니다.

따라서 반복 횟수는 정해져 있지 않고, **올바른 값을 입력할 때까지 반복**됩니다.

- `continue`: 현재 반복을 중단하고 처음으로 돌아갑니다.
- `break`: 반복문을 완전히 종료합니다.

**예시 코드**

```python
while True:
    try:
        score = int(input("점수(0~100): "))
    except ValueError:
        print("숫자로 입력해 주세요.")
        continue

    if 0 <= score <= 100:
        break

    print("0부터 100 사이의 값을 입력해 주세요.")

print(f"입력된 점수: {score}")
```

**실행 결과**

`안녕`, `150`, `80`을 차례로 입력하면 반복문은 총 3번 실행됩니다.

```text
점수(0~100): 안녕
숫자로 입력해 주세요.
점수(0~100): 150
0부터 100 사이의 값을 입력해 주세요.
점수(0~100): 80
입력된 점수: 80
```

##### 4. 연산자에 따라 계산하는 함수

**질문**

연산자에 따라 사칙연산을 수행하는 `calculate()` 함수는 어떤 코드인가요?

**답변**

`calculate()`는 숫자 두 개와 연산자 하나를 전달받아 덧셈, 뺄셈, 곱셈 또는 나눗셈 결과를 반환하는 함수입니다.

- `a`: 첫 번째 숫자
- `b`: 두 번째 숫자
- `operator`: 실행할 연산자
- `return`: 계산 결과를 함수를 호출한 곳으로 돌려줍니다.

**예시 코드**

```python
def calculate(a, b, operator):
    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        return a / b

result = calculate(10, 2, "*")
print(result)
```

**실행 결과**

```text
20
```

##### 5. `global` 키워드의 역할

**질문**

파이썬의 `global` 키워드는 어떤 역할을 하나요?

**답변**

`global`은 함수 안에서 **함수 밖에 있는 전역 변수를 수정할 때** 사용하는 키워드입니다.

함수 안에서 `global score`라고 작성하면, 함수 안의 `score`가 새로 만든 지역 변수가 아니라 함수 밖의 `score`를 가리키게 됩니다.

**예시 코드**

```python
score = 10

def change_score():
    global score
    score = 20

change_score()
print(score)
```

**실행 결과**

```text
20
```

`global`을 사용하지 않으면 함수 안에서 만든 변수는 함수 안에서만 사용하는 지역 변수가 됩니다.

```python
score = 10

def change_score():
    score = 20

change_score()
print(score)
```

```text
10
```

예를 들어 쇼핑몰에서 구매 버튼을 누르면 다음과 같이 작동합니다.

1. `HTML`이 구매 버튼을 화면에 보여줍니다.

모든 사이트가 모든 기술을 사용하는 것은 아닙니다.

- 간단한 소개 사이트: `HTML + CSS + JavaScript`
- 기능이 많은 웹서비스: `React + 백엔드 언어 + SQL`

##### 8. React의 개념과 역할

**질문**

React는 무엇인가요?

**답변**

React는 **JavaScript를 이용하여 웹사이트 화면을 편리하게 만드는 라이브러리**입니다.

React는 화면을 `컴포넌트`라는 작은 부품으로 나누어 만들고, 데이터가 변경되면 관련 화면도 자동으로 변경되도록 도와줍니다.

React 자체는 JavaScript와 같은 별도의 프로그래밍 언어가 아닙니다.

**예시 코드**

```jsx
function SaveButton() {
    return (
        <button onClick={() => alert("저장되었습니다.")}>
            저장
        </button>
    );
}
```

**실행 결과**

화면에 `저장` 버튼이 나타나고, 버튼을 클릭하면 다음 안내창이 표시됩니다.

```text
저장되었습니다.
```

**실행 결과**

##### 10. Figma·React·Bootstrap과 HTML·CSS·JavaScript의 관계

**질문**

Figma·React·Bootstrap의 관계는 HTML·CSS·JavaScript의 관계와 비슷한가요?

**답변**

뼈대에 살을 붙이는 관계로 어느 정도 이해할 수 있지만, 정확히는 각각 담당하는 역할이 다릅니다.

| 기술 | 역할 | 비유 |
|---|---|---|
| Figma | 화면을 미리 디자인함 | 설계도 |
| HTML | 화면의 기본 구조를 만듦 | 뼈대 |
| CSS | 화면을 꾸밈 | 옷과 피부 |
| JavaScript | 화면에 동작을 추가함 | 근육과 신경 |
| React | 화면을 부품으로 나누고 변경을 관리함 | 조립·관리 시스템 |
| Bootstrap | 미리 만들어진 디자인을 제공함 | 기성복 |

기본적인 웹사이트는 다음과 같이 이해할 수 있습니다.

`HTML → CSS → JavaScript`

- HTML로 구조를 만들고
- CSS로 꾸미고
- JavaScript로 움직이게 합니다.

Figma, React, Bootstrap은 단순히 순서대로 사용하는 기술이 아니라 필요에 따라 조합하여 사용합니다.

`Figma로 설계 → React로 화면과 기능 구현 → Bootstrap으로 빠르게 꾸미기`

##### 11. 바이브 코딩과 React의 종류

**질문**

바이브 코딩은 무엇이며, React는 HTML·CSS·Java와 같은 종류인가요?

**답변**

**바이브 코딩**은 코드를 한 줄씩 직접 작성하기보다, AI에게 원하는 기능을 말로 설명하여 프로그램을 만드는 방식입니다.

일반적인 진행 과정은 다음과 같습니다.

1. AI에게 원하는 기능을 설명합니다.
2. AI가 코드를 작성합니다.
3. 작성된 코드를 실행합니다.
4. 결과를 확인합니다.
5. 오류나 부족한 점을 AI에게 다시 설명하여 수정합니다.

React는 HTML·CSS·JavaScript와 정확히 같은 종류가 아닙니다.

| 기술 | 종류 |
|---|---|
| HTML | 웹페이지 구조를 만드는 마크업 언어 |
| CSS | 웹페이지를 꾸미는 스타일 언어 |
| JavaScript | 웹페이지에 기능을 추가하는 프로그래밍 언어 |
| React | JavaScript로 화면을 편리하게 만드는 라이브러리 |
| Java | JavaScript와 다른 별도의 프로그래밍 언어 |

AI가 코드를 작성해 주더라도 결과가 올바른지 확인하고 오류를 수정하려면 기본적인 코딩 지식이 필요합니다.

##### 12. 바이브 코딩에서 React가 필요한 이유

**질문**

바이브 코딩에서 React가 없으면 어렵다는 말은 무슨 의미인가요?

**답변**

React가 없으면 웹사이트를 만들 수 없다는 뜻은 아닙니다.

간단한 사이트는 `HTML`, `CSS`, `JavaScript`만으로도 만들 수 있습니다. 하지만 화면의 값이 계속 변하는 복잡한 웹서비스는 React를 사용하면 더 편리하게 관리할 수 있습니다.

예를 들어 다음과 같은 기능은 React와 잘 맞습니다.

- 사용자가 질문에 답변하면 화면에 바로 반영되는 기능
- 답변에 따라 마인드맵이 변경되는 기능
- PRD 진행률이 자동으로 올라가는 기능
- 항목을 추가·수정·삭제하는 기능
- 보류함으로 항목을 이동하는 기능

React에서는 데이터를 변경하면 그 데이터를 사용하는 화면도 함께 변경됩니다.

**예시 코드**

```jsx
const [progress, setProgress] = useState(0);

return (
    <div>
        <p>PRD 진행률: {progress}%</p>
        <button onClick={() => setProgress(progress + 10)}>
            진행률 올리기
        </button>
    </div>
);
```

**실행 결과**

처음 화면:

```text
PRD 진행률: 0%
[ 진행률 올리기 ]
```

버튼을 한 번 클릭한 화면:

```text
PRD 진행률: 10%
[ 진행률 올리기 ]
```

##### 13. AI가 없던 시기의 웹사이트 개발과 협업

**질문**

AI가 없던 과거에는 한 사람이 모든 웹 기술을 알아야 했나요, 아니면 여러 개발자가 전문 영역을 나누어 협업했나요?

**답변**

두 가지 방식이 모두 사용되었습니다.

간단한 사이트나 작은 프로젝트에서는 한 사람이 프론트엔드, 백엔드, 데이터베이스를 함께 담당하기도 했습니다. 이런 개발자를 **풀스택 개발자**라고 합니다.

규모가 큰 서비스에서는 여러 사람이 전문 영역을 나누어 협업했습니다.

| 역할 | 주요 업무 |
|---|---|
| 기획자·PM | 필요한 기능과 일정 정리 |
| 웹 디자이너 | 화면 디자인과 사용자 흐름 설계 |
| 프론트엔드 개발자 | 사용자가 보는 화면과 동작 구현 |
| 백엔드 개발자 | 로그인, 주문, 데이터 처리 등 서버 기능 구현 |
| 데이터베이스 담당자 | 데이터 구조 설계와 관리 |
| 인프라 담당자 | 서버 배포, 운영, 보안 관리 |

개발자들이 모든 프로그래밍 언어를 알아야 했던 것은 아닙니다. 프로젝트에 필요한 기술 조합을 선택하여 사용했습니다.

예를 들면 다음과 같습니다.

- `React + Python + SQL`
- `React + Java + SQL`

AI가 없던 시기에도 개발자들은 모든 코드를 외워서 작성한 것이 아니라 공식 문서, 검색 결과, 개발 커뮤니티의 예시 등을 참고하면서 개발했습니다.
