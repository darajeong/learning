### 261002 학습내용

#### 상세 학습 내용

1. 개인정보는 필요한 만큼만 수집하고, 접근을 제한한다.

서비스 목적에 필요한 최소한의 개인정보만 수집해야 한다. 화면·로그·캡처에서는 업무에 필요하지 않은 개인정보를 마스킹하고, 원본에는 필요한 사람만 접근하도록 관리해야 한다. 마스킹만으로 원본 데이터가 보호되는 것은 아니다. 개인정보 보호법은 최소 수집과 안전조치 의무를 규정한다.

“해킹되면 무조건 개인정보법 위반”이라고 단정하기보다는, 유출 사고가 발생하면 안전조치 이행 여부와 신고·통지 등 관련 의무를 검토해야 한다고 이해하는 것이 정확하다.

2. REVOKE는 권한 회수, DROP은 객체 삭제이다.

- GRANT: 권한 부여 — 테이블 조회 허용
- REVOKE: 부여한 권한 회수 — 테이블 조회 권한 제거
- DROP: 객체 자체 삭제 — 테이블·Role·DB 삭제

다만 직접 부여한 권한을 회수해도 다른 역할이나 PUBLIC을 통해 권한이 남을 수 있으므로 최종적으로 가능한 작업까지 확인해야 한다.

3. 퇴사자 접근 차단과 인증정보의 수명 관리가 중요하다.

퇴사 시 계정·권한을 회수하고, 관련 세션·토큰·키의 무효화나 교체도 검토해야 한다. 인증서에 만료일을 두는 것은 도움이 되지만, 인증서 기반이 모든 상황에서 가장 좋다거나 만료만으로 충분하다고 할 수는 없다. 유출된 키의 폐기·교체와 접근 통제가 함께 필요하다.

쿠팡 사건은 회사 공시에서 전 직원의 고객 계정 무단 접근 사건으로 설명한다. 다만 이를 “퇴사 후 기존 개발자 권한이 그대로 남아서 발생했다”는 한 문장으로 단정하면 인증정보 악용 등 세부 원인을 놓칠 수 있다.

4. HTML은 Live Server로 열어 웹 환경에서 실습한다.

HTML을 더블클릭하면 보통 file://로 열린다. 이때 JavaScript로 다른 파일을 읽는 작업 등이 브라우저 보안 제한에 막힐 수 있다.

Live Server로 열면 http://localhost:5500/...처럼 HTTP를 통해 파일을 제공받는다. 다만 다른 API 서버에 대한 CORS 제한까지 자동으로 없어지는 것은 아니다.

5. API·REST API·CRUD·React는 서로 다른 개념이다.

- API: 다른 프로그램의 기능·데이터를 이용하는 정해진 인터페이스이다.
- REST API: 자원과 HTTP의 의미 등을 활용하는 REST 원칙에 따른 API이다.
- CRUD: 생성·조회·수정·삭제라는 네 가지 데이터 작업이다.
- React: 사용자 인터페이스를 만드는 JavaScript 라이브러리이다.

회원 API에서는 보통 다음처럼 연결한다.

- 생성(Create) → POST /members
- 조회(Read) → GET /members
- 수정(Update) → PATCH 또는 PUT /members/1
- 삭제(Delete) → DELETE /members/1

REST는 API를 설계하는 방식 중 하나이며, 모든 API가 REST일 필요는 없다.

6. 수업에서 말하는 fetch()는 서버에 요청하는 JavaScript 함수이다.

“DB에서 데이터를 한 개씩 가져오는 기능”은 이 수업의 fetch() 설명으로는 틀린 설명이다.

브라우저의 fetch() → API 서버에 HTTP 요청 → 백엔드가 DB 조회 → JSON 응답 → 브라우저가 화면 갱신

fetch()는 한 건이나 여러 건의 데이터뿐 아니라 파일도 요청할 수 있다. DB 라이브러리의 fetchone()·fetchall()과는 구분해야 한다.

7. HTTP의 P는 Protocol을 뜻한다.

HTTP는 Hypertext Transfer Protocol의 약자이고, 웹에서 요청·응답을 주고받는 규칙이다. “P로 끝나면 모두 프로토콜”이라는 일반 규칙은 아니다.

8. DNMS가 아니라 DNS이며, 도메인 이름과 IP 주소를 연결한다.

www.daum.net은 도메인 이름이고 IP 주소 자체가 아니다. DNS는 이 이름에 대응하는 IP 주소 등을 찾는 시스템이다.

www.daum.net 입력 → 캐시에 주소 정보가 없으면 DNS 조회 → 얻은 IP 주소를 이용해 서버에 연결 → 웹페이지 요청·응답

DNS 서버가 웹페이지 전체를 전달하는 중간 통로인 것은 아니다.

또한 모든 기기가 독립된 공인 IP를 가져야 하는 것은 아니다. 집 안의 PC와 스마트폰처럼 여러 기기가 사설 IP를 사용하고, 공유기의 NAT를 통해 하나의 공인 IP를 공유할 수 있다.

“응답 기다리는 중”, “몇 KB 받는 중”은 통신 진행 상황을 나타내며, 그 자체가 DNS 처리만을 의미하지는 않는다.

9. AI 개발에서는 결과를 이해하고 검증하는 능력이 중요하다.

요구사항을 읽고, AI 코드의 동작을 설명하며, 화면·Console·Network·테스트 결과로 정상 여부를 확인하는 태도는 이 수업의 핵심이다.

다만 “근거와 출처를 가지는 사람이 AX 개발자이다”는 표준 기술 정의라기보다 수업에서 강조하는 개발자상으로 이해하는 것이 적절하다. 독해력은 요구사항과 오류 메시지를 정확하게 이해하는 데 도움이 된다.

10. HTTP 상태 코드 200은 성공 신호이지만, 기능 전체의 성공을 보장하지 않는다.

- 200: 요청 성공
- 403: 서버가 요청 처리를 거부함
- 404: 요청한 자원을 찾지 못함
- 500: 서버 내부 오류

200을 받았더라도 JSON 내용이 예상과 다르거나 화면 표시가 실패할 수 있으므로 응답 내용과 화면까지 확인해야 한다.

11. Django에서는 Template·View·Model의 역할을 구분한다.

- Template: 데이터를 넣어 HTML 화면을 구성한다.
- View: 요청을 처리하고 필요한 데이터를 가져와 응답한다.
- Model: 데이터 구조와 DB 조회·저장을 담당한다.

Template은 보통 서버에서 처리되어 HTML이 된 뒤 브라우저로 전달된다. 프론트엔드에는 CSS와 브라우저에서 실행되는 JavaScript도 포함된다.

12. 정적·동적 웹은 DB 연결 여부만으로 구분하지 않는다.

- 정적 제공: 서버가 미리 만들어진 HTML·CSS·JavaScript 등의 파일을 전달한다.
- 동적 생성: 서버가 요청이나 상황에 따라 내용을 만들어 응답한다.

DB 없이도 서버가 현재 시각을 계산해 응답하면 동적일 수 있다. 반대로 정적으로 제공받은 HTML에서도 JavaScript가 화면을 바꾸거나 API 데이터를 표시할 수 있다.

따라서 “정적 파일 제공”, “서버의 동적 응답”, “브라우저의 동적 화면”을 구분하면 정확하다.
