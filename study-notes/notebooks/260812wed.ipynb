가상환경을 사용하지 않으면 애플리케이션 안에 불필요한 오버헤드가 생길 수 있다.

프로젝트마다 별도의 가상환경을 만들어 필요한 라이브러리가 서로 충돌하지 않도록 한다.

##### 4. 데이터 조회

CRUD에서 데이터를 확인하는 것은 `Read`에 해당한다. SQL에서는 `SELECT`를 사용해 데이터를 조회한다.

```sql
SELECT *
FROM practice.members
ORDER BY member_id;
```

위 SQL은 `practice` 스키마에 있는 `members` 테이블의 데이터를 `member_id` 순서로 조회한다.

---

##### 5. 스키마 설정

`practice`는 스키마이다.

왼쪽에서 `practice`를 활성 스키마로 설정할 수 있다. 활성 스키마로 설정하지 않았다면 다음과 같이 스키마 이름과 테이블 이름을 함께 작성한다.

```sql
FROM practice.members
```

`practice`를 활성 스키마로 설정하면 테이블을 조회할 때 `practice`를 생략할 수 있다.

##### 6. Git 저장소 설정

현재 프로젝트는 Git에 없으며 Git의 추적을 받지 않는다. Git과 관계없이 만든 프로젝트이다.

다음 명령어를 사용해 Git 저장소를 초기화한다.

```bash
git init
```

저장소를 만든 뒤 `git clone`을 사용하는 방법도 학습했다.

현재 위치를 의미하는 `.`을 붙이면 현재 상태 또는 현재 위치에서 열라는 의미인지 추가로 확인한다.

---

##### 7. Python 가상환경 만들기

현재 프로젝트에는 가상환경이 아직 없다.

다음 명령어를 터미널에 입력해 가상환경을 만든다.

```bash
python -m venv .venv
```

명령어 마지막의 `.venv`는 가상환경의 이름이다. 다른 이름을 사용할 수도 있지만 관례상 `.venv`를 많이 사용한다.



명령어를 실행한 뒤 왼쪽 파일 목록에 `.venv`가 나타나는지 확인한다.
