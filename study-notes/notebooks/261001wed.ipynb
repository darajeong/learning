### 261001 학습내용

#### 상세 학습 내용
- HTML(HyperText Markup Language)은 제목, 문단, 버튼, 입력창처럼 웹 화면의  내용과 기본 구조 를 만드는 마크업 언어입니다.
- HTML은 이름 끝에 `Language`가 들어가지만, 조건문이나 반복문으로 동작을 처리하는 프로그래밍 언어가 아니라 문서 구조를 표현하는 마크업 언어 입니다.
- CSS(Cascading Style Sheets)는 색상, 글꼴, 크기, 간격, 배치처럼 HTML 요소의 디자인과 화면 표현을 담당하는 스타일시트 언어입니다.
- CSS도 넓은 의미에서는 컴퓨터 언어이지만, 일반적으로 프로그래밍 언어가 아니라 스타일시트 언어로 분류합니다.
- JavaScript는 버튼 클릭, 알림창, 입력값 검증, 메뉴 열기, 화면 내용 변경처럼 웹페이지의 동작과 상호작용을 담당하는 프로그래밍 언어입니다.
- JavaScript의 입력값 검증은 사용자 편의를 위한 것이므로, 보안에 필요한 검증은 반드시 백엔드에서도 다시 수행해야 합니다.
- 프런트엔드는 사용자가 직접 보고 조작하는 화면 영역으로, 주로 HTML·CSS·JavaScript를 사용해 UI와 UX를 구현합니다.
- 백엔드는 사용자에게 직접 보이지 않지만 요청 처리, 로그인, 권한 확인, 업무 규칙 실행, 데이터베이스 연결처럼 서비스의 핵심 기능을 담당합니다.
- 프런트엔드가 강조되는 이유는 백엔드가 당연해서가 아니라, 사용자가 서비스의 편리함과 품질을 화면과 상호작용을 통해 직접 경험하기 때문입니다.
- HTML과 CSS는 한 파일에 함께 작성할 수도 있지만, 규모가 커지면 HTML은 구조를 담당하고 CSS는 디자인을 담당하도록 별도 파일로 분리하는 것이 관리와 재사용에 유리합니다.
- 웹사이트의 역할을 간단히 정리하면 HTML은 골격, CSS는 장식과 배치, JavaScript는 동작, 백엔드는 데이터와 핵심 처리를 담당합니다.
