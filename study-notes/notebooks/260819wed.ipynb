##### 1. VS Code의 코드 색상

VS Code에서는 코드의 구성 요소를 쉽게 구분할 수 있도록 서로 다른 색상으로 표시한다.

- 빨간색: 문법
- 검정색: 직접 작성하거나 바꿀 수 있는 내용
- 보라색: 메서드처럼 정해진 기능

메서드는 Python이나 라이브러리에서 제공하는 기능이므로 이름을 임의로 변경하면 해당 기능을 사용할 수 없다.

다만 코드에 표시되는 색상은 사용하는 VS Code 테마와 설정에 따라 달라질 수 있다. 따라서 색상 자체를 외우기보다 각 코드가 어떤 역할을 하는지 이해하는 것이 중요하다.

---

##### 2. print()에서 쉼표 사용하기

`print()`에서 여러 값을 쉼표로 구분하면 출력되는 값 사이에 보통 한 칸의 공백이 자동으로 들어간다.

```python
name = "문길동"
age = 20

print(name, age)
```

실행 결과:

```text
문길동 20
```

---

##### 3. 주석 작성하기

Python에서 `#`은 주석을 작성할 때 사용한다.

```python
# 이름을 변수에 저장한다.
name = "문길동"
```

주석은 코드의 모든 내용을 그대로 설명하기보다 꼭 필요한 부분만 간결하게 작성하는 것이 좋다.

VS Code에서 여러 줄을 한꺼번에 주석 처리하려면 원하는 코드의 여러 줄을 선택하고 `Ctrl + /`를 누른다. 같은 단축키를 다시 누르면 주석을 해제할 수 있다.

---

##### 4. 들여쓰기 오류

`IndentationError`는 Python 코드의 들여쓰기가 잘못되었을 때 발생하는 오류이다.

Python에서는 들여쓰기가 단순히 코드를 보기 좋게 정리하는 용도가 아니라 문법의 일부로 사용된다.

```python
if age >= 20:
    print("성인입니다.")
```

조건에 포함되는 `print()` 앞에 들여쓰기가 없거나 들여쓰기 위치가 잘못되면 오류가 발생할 수 있다.

---
