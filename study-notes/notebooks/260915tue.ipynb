##### 1. `value_counts()`의 결과가 `Series`인 이유

**질문**

pandas의 `value_counts()` 결과가 왜 `Series` 자료형으로 나오는가?

**답변**

`value_counts()`는 각 값이 몇 번 나왔는지를 세어 **값의 이름과 개수로 이루어진 한 줄짜리 목록**을 만듭니다.

pandas에서 한 개의 열처럼 생긴 자료는 `Series`로 표현합니다.

- 숫자 하나: `int` 또는 `float`
- 한 개의 행이나 열: `Series`
- 여러 개의 열로 이루어진 표: `DataFrame`

**예시 코드**

```python
product_count = products["category"].value_counts(dropna=False)

print(product_count)
print(type(product_count))
```

**실행 결과**

```text
category
전자제품    3
의류        2
식품        1
Name: count, dtype: int64

<class 'pandas.core.series.Series'>
```

결과의 왼쪽에는 카테고리 이름이 있고, 오른쪽에는 각 카테고리의 개수가 있습니다. 이처럼 한 개의 열 모양이기 때문에 `Series`입니다.

##### 2. 시그모이드, 인코딩, 크로스 엔트로피가 필요한 이유

**질문**

시그모이드, 원핫 인코딩, BCE, CCE, Sparse CCE 같은 기능은 왜 필요하며 각각 어떤 상황에서 사용하는가?

**답변**

컴퓨터는 `스팸`, `고양이`, `주말` 같은 말의 의미를 사람처럼 바로 이해하지 못합니다. 따라서 현실의 정보를 숫자로 바꾸고, 모델의 답을 확률로 표현하고, 얼마나 틀렸는지 숫자로 채점해야 합니다.

전체 과정은 다음과 같습니다.

> 현실의 정보 → 숫자로 변환 → 모델이 계산 → 확률 출력 → 틀린 정도 계산 → 모델 수정

각 기능의 역할은 다음과 같습니다.

- **원핫 인코딩:** 순서가 없는 범주를 가짜 서열 없이 숫자로 표현합니다.
- **시그모이드:** 모델이 계산한 점수를 `0~1` 사이의 확률 하나로 바꿉니다.
- **BCE:** 두 가지 중 하나를 분류할 때 예측 확률을 채점합니다.
- **소프트맥스:** 여러 범주의 점수를 합이 1인 확률들로 바꿉니다.
- **CCE:** 정답이 원핫 인코딩일 때 여러 범주의 예측을 채점합니다.
- **Sparse CCE:** 정답이 정수형 레이블일 때 여러 범주의 예측을 채점합니다.

**예시 코드**

```python
# 범주가 3개일 때 정답 표현

# 원핫 인코딩
dog_one_hot = [0, 1, 0]

# 정수형 레이블
dog_integer = 1
```

**실행 결과**

```text
고양이 = 0
강아지 = 1
토끼 = 2

강아지의 원핫 정답 = [0, 1, 0]
강아지의 정수형 정답 = 1
```

CCE와 Sparse CCE의 목적은 거의 같고, 정답의 표현 방식이 다릅니다.

- `CCE`의 정답: `[0, 1, 0]`
- `Sparse CCE`의 정답: `1`

이진 분류에서는 보통 다음 조합을 사용합니다.

> **시그모이드 + BCE**

다중 분류에서는 보통 다음 조합을 사용합니다.

> **소프트맥스 + CCE 또는 Sparse CCE**

##### 3. 엔터티와 속성

**질문**

데이터베이스에서 엔터티란 무엇이며, 열을 속성이라고 하는 이유는 무엇인가?

**답변**

엔터티는 데이터베이스에서 **정보를 저장하고 관리하려는 대상**입니다.

예를 들어 학교 데이터베이스에서는 다음과 같은 대상이 엔터티가 될 수 있습니다.

- 학생
- 강사
- 강의
- 수강신청

관계형 데이터베이스에서는 보통 엔터티를 테이블로 만듭니다.

열은 엔터티의 특징이나 정보의 종류를 나타내기 때문에 **속성**이라고 합니다.

| 학생번호 | 이름 | 학년 |
|---:|---|---:|
| 101 | 철수 | 3 |
| 102 | 영희 | 4 |

이 표를 기준으로 구분하면 다음과 같습니다.

- **엔터티:** 학생
- **테이블:** 학생 정보를 저장한 표
- **속성 또는 열:** 학생번호, 이름, 학년
- **행:** 철수 한 명 또는 영희 한 명의 기록
- **인스턴스:** 엔터티의 실제 사례인 철수 또는 영희

쉽게 기억하면 다음과 같습니다.

> **독립적으로 관리할 대상이나 사건은 테이블이 되고, 그 대상을 설명하는 정보는 열이 된다.**

##### 4. 정규화 위배와 1NF·2NF·3NF

**질문**

정규화 위배란 무엇이며 1NF·2NF·3NF는 각각 어떤 규칙인가?

**답변**

정규화는 뒤죽박죽 섞인 정보를 **각 정보의 주인에 맞게 정리하는 과정**입니다.

정규화 위배는 정보가 정규화 규칙에 맞게 정리되지 않아 중복, 수정 오류, 삭제 오류 등이 생길 수 있는 상태를 뜻합니다.

정규화에서 가장 중요한 질문은 다음과 같습니다.

> **이 정보의 진짜 주인은 누구인가?**

**1NF는 한 칸에 값 하나를 넣는 규칙입니다.**

잘못된 예:

| 학생번호 | 좋아하는 과목 |
|---:|---|
| 101 | 수학, 과학 |

한 칸에 독립적인 과목 두 개가 들어 있습니다.

정리한 예:

| 학생번호 | 좋아하는 과목 |
|---:|---|
| 101 | 수학 |
| 101 | 과학 |

쉽게 기억하면 다음과 같습니다.

> **1NF = 한 칸에 하나**

**2NF는 복합키의 일부에만 속한 정보를 분리하는 규칙입니다.**

다음 수강 테이블의 한 행을 찾으려면 `학생번호 + 과목번호`가 필요하다고 가정합니다.

| 학생번호 | 과목번호 | 학생이름 | 과목이름 | 성적 |
|---:|---:|---|---|---|
| 1 | 10 | 철수 | 수학 | A |

각 정보가 결정되는 방식은 다음과 같습니다.

- 학생이름은 학생번호만 알면 결정됩니다.
- 과목이름은 과목번호만 알면 결정됩니다.
- 성적은 학생번호와 과목번호를 모두 알아야 결정됩니다.

따라서 학생이름과 과목이름을 각각의 테이블로 분리합니다.

- `students(학생번호, 학생이름)`
- `courses(과목번호, 과목이름)`
- `enrollments(학생번호, 과목번호, 성적)`

쉽게 기억하면 다음과 같습니다.

> **2NF = 두 개짜리 열쇠의 한쪽에만 딸린 정보는 자기 테이블로 보낸다.**

**3NF는 기본키가 아닌 다른 일반 열에 속한 정보를 분리하는 규칙입니다.**

예를 들어 다음 관계가 있다고 가정합니다.

> 학생번호 → 우편번호 → 도시

도시는 학생번호보다 우편번호에 직접 속한 정보입니다. 따라서 다음처럼 분리할 수 있습니다.

- `students(학생번호, 이름, 우편번호)`
- `zip_codes(우편번호, 도시)`

쉽게 기억하면 다음과 같습니다.

> **3NF = 키가 아닌 다른 열에 딸린 정보는 자기 테이블로 보낸다.**

전체 내용을 짧게 정리하면 다음과 같습니다.

- **1NF:** 한 칸에 값 하나
- **2NF:** 복합키 일부에만 의존하는 속성 분리
- **3NF:** 기본키가 아닌 다른 속성에 의존하는 속성 분리

##### 5. 회원 고아 참조

**질문**

회원 고아 참조란 무엇이며, 여기서 ‘고아’는 어떤 상태를 뜻하는가?

**답변**

회원 고아 참조는 대여 기록에는 회원번호가 있지만, 회원 테이블에는 해당 회원이 존재하지 않는 상태입니다.

예를 들어 회원 테이블에 `101번`과 `102번` 회원만 있는데, 대여 테이블의 어떤 행이 `999번` 회원을 가리킨다면 문제가 됩니다.

| 대여번호 | 회원번호 | 도서번호 |
|---:|---:|---:|
| 1001 | 101 | 201 |
| 1002 | 999 | 202 |

`999번` 회원은 회원 테이블에 없기 때문에, `1002번` 대여 기록은 연결되어야 할 부모 회원이 없습니다.

부모와 연결되지 못한 상태에 빗대어 **고아 참조**라고 부릅니다.

- **부모 테이블:** `members_nf`
- **자식 테이블:** `loans_nf`
- **연결 열:** `member_id`
- **고아 참조:** 자식이 가리키는 부모가 존재하지 않는 상태

이를 막아주는 데이터베이스 규칙이 **외래 키(Foreign Key)**입니다.

**예시 코드**

```sql
SELECT COUNT(*) AS member_orphan_count
FROM loans_nf l
LEFT JOIN members_nf m
    ON l.member_id = m.id
WHERE m.id IS NULL;
```

**실행 결과**

```text
member_orphan_count
-------------------
0
```

결과가 `0`이면 존재하지 않는 회원을 가리키는 대여 기록이 없다는 뜻입니다.

결과가 `1`이면 회원 테이블에 없는 회원번호를 가리키는 대여 기록이 한 건 있다는 뜻입니다.
