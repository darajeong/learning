##### 1. SQL 집계·필터링·문자열 검색과 NULL 정렬

- `SELECT COUNT(*) AS student_count FROM public.students;`는 전체 학생 수를 세고 결과 열 이름을 `student_count`로 표시하며, PostgreSQL에서는 큰따옴표 없는 별칭을 소문자로 처리하므로 대문자로 표시하려면 `AS "STUDENT_COUNT"`라고 씁니다.
- `WHERE`는 데이터를 조회할 때 조건에 맞는 행만 골라내는 데 사용합니다.
- `IN`은 같은 열이 여러 값 중 하나와 일치하는지 확인하며, `AND (grade = 2 OR grade = 3)`는 `AND grade IN (2, 3)`와 같은 조건입니다.
- `LIKE`는 문자열의 패턴을 검색하며, `%`는 0개 이상의 문자를, `_`는 정확히 한 문자를 뜻합니다.
- `NULL`인 행은 `IS NULL`로, `NULL`이 아닌 행은 `IS NOT NULL`로 찾으며, `= NULL`로는 찾을 수 없습니다.
- PostgreSQL의 `ORDER BY grade DESC`는 기본적으로 `NULL`을 먼저 표시하므로, 학년을 내림차순으로 정렬하면서 미입력 값을 마지막에 두려면 `ORDER BY grade DESC NULLS LAST`라고 씁니다.

##### 2. AI·AGI·멀티모달의 개념과 차이

**질문**

AI와 AGI는 어떻게 다르며, 멀티모달과 modal은 무슨 뜻이고 AGI와 어떤 관계가 있나요?

**답변**

- **AI(인공지능)**: 언어 이해나 이미지 인식처럼 지능이 필요한 작업을 수행하는 기술을 넓게 부르는 말입니다.
- **AGI(인공 일반 지능)**: 다양한 분야를 배우고 새로운 문제에도 지식을 폭넓게 응용할 수 있는 AI를 뜻합니다. 어느 수준부터 AGI라고 할지는 합의된 기준이 없습니다.
- **멀티모달(Multimodal)**: 글·이미지·음성·영상 등 여러 형태의 정보를 이해하거나 생성하는 능력입니다.
- **modal**: 여기서는 정보의 방식이나 형태와 관련된 말입니다. 정보를 표현하거나 받아들이는 형태를 **modality(모달리티)**라고 하며, 텍스트·이미지·음성이 각각 하나의 모달리티입니다.

**AGI는 AI에 포함되는 개념**이고, 멀티모달은 다루는 정보의 종류에 관한 개념입니다. 따라서 멀티모달 AI라고 해서 반드시 AGI인 것은 아닙니다.

예를 들어 사진을 보고 음성으로 설명하는 AI는 멀티모달이지만, 그 능력만으로 AGI라고 판단할 수는 없습니다.

##### 3. 우분투 서버의 기본 개념

**질문**

우분투 서버는 무엇인가요?

**답변**

**우분투 서버(Ubuntu Server)는 서버 운영에 사용하는 리눅스 운영체제**입니다. 윈도우처럼 컴퓨터의 기본 동작을 관리하면서 웹사이트나 데이터베이스 같은 서비스를 실행하는 기반이 됩니다.

주로 다음과 같은 용도로 사용합니다.

- **웹 서버**: 웹사이트와 앱 서비스를 제공합니다.
- **파일 서버**: 파일을 저장하고 공유합니다.
- **데이터베이스 서버**: PostgreSQL 같은 DBMS를 실행합니다.
- **개발 서버**: 프로그램을 실행하고 테스트합니다.

일반적으로 기본 설치에서는 그래픽 화면 대신 명령어로 관리하며, 다른 컴퓨터에서 **SSH**로 원격 접속해 사용합니다.

우분투 서버는 기계 이름이 아닙니다. 개인 PC나 클라우드 컴퓨터에 설치해 사용하는 운영체제입니다.

##### 4. SQL 조회문의 작성 순서와 결과 개수 제한

**질문**

`SELECT`, `FROM`, `WHERE`, `ORDER BY` 등은 어떤 순서로 작성하며, `LIMIT`은 어떤 역할을 하나요?

**답변**

일반적인 조회문의 작성 순서는 다음과 같습니다. 필요한 항목만 사용하면 됩니다.

1. **`WITH`**: 조회에서 사용할 임시 결과를 정의합니다.
2. **`SELECT`**: 보여줄 열이나 계산식을 지정합니다.
3. **`FROM`**: 데이터를 가져올 테이블을 지정합니다.
4. **`JOIN ... ON`**: 다른 테이블을 조건에 맞춰 연결합니다.
5. **`WHERE`**: 조건에 맞는 행을 고릅니다.
6. **`GROUP BY`**: 행을 특정 기준으로 묶습니다.
7. **`HAVING`**: 묶인 그룹에 조건을 적용합니다.
8. **`ORDER BY`**: 결과를 정렬합니다.
9. **`LIMIT`**: 결과를 최대 몇 행까지 보여줄지 지정합니다.
10. **`OFFSET`**: 정렬된 결과의 앞부분에서 몇 행을 건너뛸지 지정합니다.

이는 **작성 순서**이며 데이터베이스 내부의 실제 처리 순서와는 다릅니다. PostgreSQL에서는 `LIMIT`과 `OFFSET`의 작성 순서를 바꿀 수도 있습니다.

추가로 알아둘 구문은 다음과 같습니다.

- **`DISTINCT`**: `SELECT` 바로 뒤에 쓰며 중복된 결과 행을 제거합니다.
- **`ASC` / `DESC`**: `ORDER BY`에서 오름차순 / 내림차순을 지정합니다.
- **`UNION`·`INTERSECT`·`EXCEPT`**: 여러 조회 결과를 결합하거나 비교합니다.

`LIMIT 3`은 **최대 3행만 표시**한다는 뜻입니다. 나머지 데이터를 삭제하지 않으며, 조건에 맞는 행이 2개뿐이면 2개만 보여줍니다.

`OFFSET`은 결과를 건너뛰는 기능입니다. 예를 들어 `LIMIT 3 OFFSET 2`는 앞의 2행을 건너뛰고 최대 3행을 보여줍니다.

**예시 코드**

컴퓨터공학 전공의 2·3학년 학생을 id가 작은 순서대로 최대 3명 조회합니다.

```sql
SELECT id, name, grade
FROM public.students
WHERE major = '컴퓨터공학'
  AND grade IN (2, 3)
ORDER BY id ASC
LIMIT 3;
```

**실행 결과**

조건에 맞는 데이터가 있다고 가정한 예시이며, 실제 결과는 저장된 데이터에 따라 달라집니다.

| id | name | grade |
|---|---|---|
| 1 | 김민지 | 2 |
| 4 | 이준호 | 3 |
| 7 | 박서연 | 2 |

##### 5. RETURNING의 역할과 SELECT와의 차이

**질문**

`INSERT ... RETURNING`은 어떤 기능을 하며, 기존 데이터를 조회하는 `SELECT ... WHERE`와 어떻게 다른가요?

**답변**

**`RETURNING`은 추가·수정·삭제한 행의 값을 바로 반환하는 구문**입니다.

`INSERT ... RETURNING`은 학생을 새로 등록하면서 접수증을 받는 것과 같습니다. 반면 `SELECT`는 이미 등록된 학생의 접수 내역을 찾아보는 것과 같습니다.

| 구문 | 역할 | 데이터 변경 여부 |
|---|---|---|
| `INSERT ... RETURNING` | 새 행을 추가하고 그 값을 보여줌 | 변경함 |
| `SELECT ... WHERE` | 조건에 맞는 기존 행을 보여줌 | 변경하지 않음 |

- `RETURNING id, name, created_at`은 지정한 세 열을 보여줍니다.
- `RETURNING *`는 작업한 행의 모든 열을 보여줍니다.
- `SELECT *`는 조회 조건에 맞는 행의 모든 열을 보여줍니다.
- `RETURNING`은 독립적인 조회문이 아니라 `INSERT` 등의 문장에 붙여 사용합니다.

**예시 코드**

아래 이메일이 아직 등록되지 않았다는 가정으로 학생을 추가합니다. 세미콜론은 `RETURNING` 뒤, 문장 맨 끝에 붙입니다.

```sql
INSERT INTO public.students (name, email, major, grade)
VALUES ('이서준', 'seojun_new@example.com', '컴퓨터공학', 2)
RETURNING id, name, created_at;
```

**실행 결과**

학생이 추가되면서 아래와 같은 결과가 표시됩니다. id와 등록 시각은 예시입니다.

| id | name | created_at |
|---|---|---|
| 10 | 이서준 | 2026-09-07 15:30:00+09 |

같은 학생을 나중에 다시 확인할 때는 다음 조회문을 사용합니다.

**예시 코드**

```sql
SELECT id, name, created_at
FROM public.students
WHERE email = 'seojun_new@example.com';
```

**실행 결과**

데이터가 변경되지 않았다면 앞서 등록한 학생의 같은 값이 표시됩니다.

| id | name | created_at |
|---|---|---|
| 10 | 이서준 | 2026-09-07 15:30:00+09 |

이 `SELECT`는 여러 번 실행해도 학생을 추가하지 않습니다. 반면 위 `INSERT`를 다시 실행하면 이메일의 `UNIQUE` 제약조건 때문에 중복 오류가 발생합니다.
