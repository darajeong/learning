##### 1. 데이터베이스 핵심 개념

- 데이터베이스는 데이터를 저장하고 관리하는 논리적 공간이며, `ai_database_book`처럼 이름을 붙여 구분한다.
- 스키마는 테이블, 열, 관계, 제약조건 등 데이터베이스 내부의 구조를 정의한 설계도이지만, MySQL처럼 데이터베이스와 스키마를 같은 의미로 사용하는 DBMS도 있다.
- DBMS는 데이터베이스를 생성·관리하고 SQL을 실행하는 소프트웨어이며, PostgreSQL이 대표적인 예이다.
- 기본 키(Primary Key)는 테이블의 각 행을 고유하게 식별하도록 선택된 열 또는 열의 조합이다.
- 시스템이 임의로 생성하는 내부 식별자는 대리 키(Surrogate Key)라고 하며, 기본 키로 자주 사용된다.
- 학번이나 주민등록번호 같은 업무 식별자를 기본 키로 사용할 수도 있지만, 값이 변경되거나 정책이 바뀔 가능성을 고려해 대리 키와 분리하는 것이 일반적으로 관리하기 쉽다.
- 기본 키가 아닌 업무 식별자도 값의 중복을 허용하지 않아야 한다면 `UNIQUE` 제약조건을 설정한다.
- 기본 키에는 `UNIQUE`와 `NOT NULL`의 성격이 있으므로 값이 중복되거나 `NULL`일 수 없다.
- 모든 테이블에 기본 키가 문법적으로 반드시 필요한 것은 아니지만, 데이터를 안정적으로 식별·수정·참조하려면 기본 키를 두는 것이 권장된다.
- 카디널리티(Cardinality)는 문맥에 따라 테이블의 행 개수 또는 테이블 간 관계의 일대일(`1:1`), 일대다(`1:N`), 다대다(`N:M`) 연결 수를 의미한다.
- 일대일 관계는 하나의 테이블로 합칠 수도 있지만, 데이터의 보안 수준·사용 빈도·생명주기가 다르거나 독립적으로 확장될 가능성이 있다면 테이블을 분리할 수 있다.
- 다대다 관계는 두 테이블을 직접 연결하지 않고 양쪽의 기본 키를 외래 키로 갖는 연결 테이블을 사용해 두 개의 일대다 관계로 변환한다.
- 학생과 강의의 다대다 관계는 `students 1 ── N enrollments N ── 1 courses`처럼 수강 테이블을 사이에 두어 표현할 수 있다.
- 참조는 한 테이블이 다른 테이블의 데이터를 연결해 사용하는 관계이며, 외래 키(Foreign Key)는 다른 테이블의 기본 키나 고유 키를 가리켜 그 관계를 구현하는 열이다.
- `VARCHAR(50)`은 DBMS와 문자 인코딩에 따라 길이 계산 방식이 달라질 수 있으므로 사용 중인 DBMS의 규칙을 확인해야 한다.
- SQL Server에서 유니코드 문자를 안정적으로 저장하려면 `NVARCHAR(50)`을 사용하며, 이는 한글도 최대 50자까지 저장할 수 있다.
- 한 테이블에 서로 다른 주제의 데이터가 섞여 중복이나 수정 오류가 발생한다면 정규화를 통해 역할에 맞는 여러 테이블로 분리한다.

##### 2. 데이터베이스와 스키마의 관계

**질문**

데이터베이스와 스키마는 각각 무엇이며, 서로 어떤 관계가 있는가?

**답변**

데이터베이스는 데이터를 저장하고 관리하는 논리적 공간이며, 스키마는 데이터베이스 안에서 테이블, 열, 관계, 제약조건 등의 구조를 정의한 설계도이다.

##### 3. 카디널리티

**질문**

데이터베이스에서 카디널리티란 무엇인가?

**답변**

카디널리티(Cardinality)는 일반적으로 테이블에 저장된 행의 개수를 의미하며, 테이블 간의 관계에서는 일대일(`1:1`), 일대다(`1:N`), 다대다(`N:M`)와 같이 데이터가 연결되는 수를 의미하기도 한다.

**예시 코드**

~~~sql
SELECT COUNT(*) AS cardinality
FROM student;
~~~

**실행 결과**

~~~text
cardinality
-----------
3
~~~

`student` 테이블에 행이 3개 있으므로 이 테이블의 카디널리티는 `3`이다.

##### 4. 참조와 외래 키

**질문**

참조와 외래 키는 각각 무엇이며, 두 개념의 관계를 예시로 어떻게 설명할 수 있는가?

**답변**

참조는 한 테이블이 다른 테이블의 데이터를 연결하여 사용하는 관계이고, 외래 키는 다른 테이블의 기본 키를 가리킴으로써 그 참조 관계를 실제로 구현하는 열이다.

**예시 코드**

~~~sql
CREATE TABLE department (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(50)
);

CREATE TABLE student (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50),
    department_id INT,
    FOREIGN KEY (department_id)
        REFERENCES department(department_id)
);

INSERT INTO department
    (department_id, department_name)
VALUES
    (101, '컴퓨터공학과'),
    (102, '경영학과');

INSERT INTO student
    (student_id, student_name, department_id)
VALUES
    (1, '민수', 101),
    (2, '지수', 102);

SELECT
    s.student_name,
    d.department_name
FROM student AS s
JOIN department AS d
    ON s.department_id = d.department_id;
~~~

**실행 결과**

~~~text
student_name | department_name
-------------|----------------
민수         | 컴퓨터공학과
지수         | 경영학과
~~~

`student` 테이블의 `department_id`는 외래 키이며, `department` 테이블의 기본 키인 `department_id`를 참조한다.
