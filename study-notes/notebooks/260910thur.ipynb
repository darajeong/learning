##### 1. 추가 학습 내용 정리

- `RETURNING`은 `SELECT`와 함께 쓰는 문법이 아니라 `INSERT`, `UPDATE`, `DELETE`처럼 데이터를 변경하는 명령 뒤에 붙여 변경된 행을 바로 확인하는 문법입니다.
- 데이터 모델을 설계할 때는 현재 상태만 보여주는 스냅숏보다 데이터가 장기간 어떻게 생성·변경·연결될지를 함께 고려해야 합니다.
- 엔터티(Entity)는 회원이나 도서처럼 독립적으로 관리할 대상이며, 관계형 데이터베이스에서는 일반적으로 테이블로 표현합니다.
- 속성(Attribute)은 엔터티가 가진 이메일이나 ISBN 같은 특징이며, 테이블에서는 열(Column)로 표현합니다.
- 인스턴스(Instance) 또는 오커런스(Occurrence)는 엔터티의 개별 사례이며, 테이블에서는 하나의 행(Row)으로 표현합니다.
- `members` 테이블의 한 행은 회원 한 명을, `books` 테이블의 한 행은 도서 한 권을 나타냅니다.
- `loans` 테이블의 한 행은 특정 회원이 특정 도서를 한 번 빌린 대출 사건 또는 대출 기록을 나타냅니다.
- 대출 기록은 회원과 도서의 관계에 대출일·반납예정일 같은 추가 정보를 저장하므로 연결 엔터티(Associative Entity)로 모델링할 수 있습니다.
- 회원, 도서, 대출 기록은 엔터티이고 이메일, ISBN, 반납예정일은 각 엔터티를 설명하는 속성입니다.
- 시드 데이터(Seed Data)는 애플리케이션의 초기 실행이나 실습에 필요한 기본 데이터를 미리 `INSERT`해 두는 데이터입니다.
- CSV는 데이터를 저장하는 파일 형식이고, DataFrame은 pandas가 Python 안에서 데이터를 표 형태로 다루는 자료구조입니다.
- 처음 수집한 데이터에는 결측값·중복·오류·이상치 등이 포함될 수 있으며, 이를 흔히 지저분한 데이터(Dirty Data)라고 합니다.
- 전처리는 pandas 등의 도구를 사용해 결측값, 중복, 잘못된 형식 등을 분석 가능한 상태로 정리하는 과정입니다.
- EDA(탐색적 데이터 분석)는 통계값과 데이터 분포·관계를 살펴보는 과정이며 pandas와 matplotlib·seaborn 등을 사용할 수 있습니다.
- 머신러닝은 정리된 과거 데이터에서 패턴을 학습해 새로운 데이터의 결과를 추정하는 방법입니다.
- 회귀는 시험 점수처럼 연속적인 숫자를 예측하고, 분류는 생존 여부나 VIP 여부처럼 정해진 범주를 예측합니다.
- scikit-learn은 데이터 분리, 전처리, 머신러닝 모델 학습, 예측 및 평가 기능을 제공하는 대표적인 Python 라이브러리입니다.
- 전체 학습 흐름은 `과거 데이터 수집 → DataFrame으로 불러오기 → 전처리와 EDA → 모델 학습과 평가 → 새로운 데이터 예측`으로 정리할 수 있습니다.

##### 2. Python의 줄바꿈 문자 `\n`

**질문**

Python에서 `\n`은 무슨 뜻이며 어떻게 읽나요?

**답변**

`\n`은 문자열 안에서 줄을 바꾸는 **줄바꿈 문자**입니다. 영어로는 **뉴라인(newline)** 또는 **라인 피드(line feed)**라고 읽습니다.

- `\`: 백슬래시
- `n`: 엔
- `\n`: 뉴라인 또는 줄바꿈 문자

**예시 코드**

~~~python
print("안녕하세요\n반갑습니다")
~~~

**실행 결과**

~~~text
안녕하세요
반갑습니다
~~~

##### 3. 데이터 분석과 머신러닝의 전체 흐름

**질문**

CSV·DataFrame·pandas·EDA·scikit-learn을 사용하는 데이터 분석과 머신러닝의 전체 흐름을 쉬운 문장으로 설명하면 어떻게 되나요?

**답변**

먼저 과거에 수집한 데이터를 CSV 파일 등에 저장합니다. Python에서는 `pandas`를 이용해 CSV 파일을 `DataFrame`이라는 표 형태로 불러옵니다.

처음 수집한 데이터에는 빈 값, 중복된 값, 잘못된 값 등이 포함될 수 있습니다. 이러한 데이터를 **Dirty Data**라고 하며, 분석하기 좋은 상태로 정리하는 작업을 **전처리**라고 합니다.

데이터를 정리한 뒤에는 평균, 분포, 데이터 사이의 관계 등을 확인합니다. 필요하면 `matplotlib` 등을 이용해 그래프도 그립니다. 이렇게 데이터의 특징을 탐색하는 과정을 **EDA(탐색적 데이터 분석)**라고 합니다.

이후 `scikit-learn`을 이용해 과거 데이터의 패턴을 머신러닝 모델에 학습시킵니다. 학습된 모델을 사용하면 시험 점수, 생존 여부, VIP 고객 여부처럼 새로운 데이터의 결과를 예측할 수 있습니다.

전체 흐름은 다음과 같습니다.

1. 과거 데이터 수집
2. CSV 파일로 저장
3. pandas로 DataFrame 생성
4. Dirty Data 전처리
5. EDA 및 시각화
6. scikit-learn으로 머신러닝 모델 학습
7. 새로운 데이터의 결과 예측

**예시 코드**

~~~python
import pandas as pd
from sklearn.linear_model import LinearRegression

# CSV 파일을 DataFrame으로 불러오기
df = pd.read_csv("scores.csv")

# 결측값과 중복 데이터 제거
df = df.dropna()
df = df.drop_duplicates()

# 공부 시간과 시험 점수 준비
X = df[["study_hours"]]
y = df["score"]

# 머신러닝 모델 생성 및 학습
model = LinearRegression()
model.fit(X, y)

# 하루 7시간 공부했을 때의 점수 예측
prediction = model.predict([[7]])

print(prediction)
~~~

**실행 결과**

데이터에 따라 다음과 같이 예상 점수가 출력될 수 있습니다.

~~~text
[91.5]
~~~

##### 4. pandas의 `dtype`

**질문**

pandas에서 `dtype`으로 데이터를 확인한다는 것은 무슨 뜻인가요?

**답변**

`dtype`은 **Data Type**, 즉 자료형을 뜻합니다. pandas에서는 각 열에 숫자, 문자열, 날짜 등 어떤 종류의 데이터가 들어 있는지 확인할 때 사용합니다.

대표적인 자료형은 다음과 같습니다.

- `int64`: 정수
- `float64`: 소수
- `object`: 주로 문자열 또는 여러 자료형이 섞인 데이터
- `bool`: 참과 거짓
- `datetime64`: 날짜와 시간

`df.dtypes`를 사용하면 모든 열의 자료형을 확인할 수 있고, `df["열 이름"].dtype`을 사용하면 특정 열의 자료형만 확인할 수 있습니다.

**예시 코드**

~~~python
import pandas as pd

df = pd.DataFrame({
    "name": ["민수", "영희"],
    "age": [20, 22],
    "score": [85.5, 90.0]
})

print(df.dtypes)
print(df["age"].dtype)
~~~

**실행 결과**

~~~text
name      object
age        int64
score    float64
dtype: object

int64
~~~

##### 5. `concat`, `join`, `aggregate`의 차이

**질문**

pandas에서 `concat`, `join`, `aggregate`는 각각 어떤 역할을 하며 차이는 무엇인가요?

**답변**

세 기능은 모두 데이터를 다루지만 목적이 다릅니다.

| 기능 | 의미 | 대표적인 사용 사례 |
|---|---|---|
| `concat` | 표를 위아래 또는 좌우로 이어 붙임 | 1월 데이터와 2월 데이터 합치기 |
| `join`·`merge` | 공통된 값을 기준으로 두 표를 연결함 | 회원 정보와 주문 정보 연결하기 |
| `aggregate` | 여러 값을 하나의 통계값으로 요약함 | 평균, 합계, 최댓값 구하기 |

`concat`은 데이터를 순서대로 붙이는 기능이고, `join`은 두 데이터의 관계를 찾아 연결하는 기능입니다. `aggregate`는 여러 행의 값을 평균이나 합계 등으로 요약하는 기능입니다.

**예시 코드**

~~~python
import pandas as pd

# concat 예시
january = pd.DataFrame({
    "name": ["민수", "영희"],
    "score": [80, 90]
})

february = pd.DataFrame({
    "name": ["철수"],
    "score": [70]
})

all_scores = pd.concat(
    [january, february],
    ignore_index=True
)

print(all_scores)
~~~

**실행 결과**

~~~text
  name  score
0   민수     80
1   영희     90
2   철수     70
~~~

**예시 코드**

~~~python
# join 또는 merge 예시
members = pd.DataFrame({
    "member_id": [1, 2],
    "name": ["민수", "영희"]
})

scores = pd.DataFrame({
    "member_id": [1, 2],
    "score": [80, 90]
})

result = pd.merge(
    members,
    scores,
    on="member_id"
)

print(result)
~~~

**실행 결과**

~~~text
   member_id name  score
0          1   민수     80
1          2   영희     90
~~~

**예시 코드**

~~~python
# aggregate 예시
result = scores["score"].aggregate(
    ["mean", "sum", "max"]
)

print(result)
~~~

**실행 결과**

~~~text
mean     85.0
sum     170.0
max      90.0
Name: score, dtype: float64
~~~

##### 6. SQL의 `public.member`와 `member`

**질문**

SQL에서 `FROM public.member`와 `FROM member`의 차이는 무엇인가요?

**답변**

`public.member`는 **`public` 스키마의 `member` 테이블**을 정확하게 지정한 이름입니다.

반면 `member`는 스키마 이름을 생략한 테이블 이름입니다. PostgreSQL은 `search_path`에 등록된 스키마를 순서대로 검색해서 `member` 테이블을 찾습니다.

현재 기본 스키마가 `public`이고 다른 스키마에 같은 이름의 테이블이 없다면 두 표현은 같은 테이블을 조회할 수 있습니다.

- `public.member`: 스키마와 테이블을 모두 지정
- `member`: 스키마를 생략하고 검색 경로에 따라 조회

단, 실제 테이블 이름이 `members`라면 `member`가 아니라 `members`라고 정확히 작성해야 합니다.

**예시 코드**

~~~sql
-- public 스키마를 직접 지정
SELECT *
FROM public.members;
~~~

~~~sql
-- 스키마를 생략
SELECT *
FROM members;
~~~

~~~sql
-- 현재 스키마 검색 경로 확인
SHOW search_path;
~~~

**실행 결과**

검색 경로가 다음과 같을 수 있습니다.

~~~text
"$user", public
~~~

이 경우 PostgreSQL은 스키마를 생략한 `members`를 먼저 사용자 이름과 같은 스키마에서 찾고, 그다음 `public` 스키마에서 찾습니다.

##### 7. JPG와 SVG의 차이

**질문**

JPG와 SVG는 어떤 차이가 있으며 용도별로 어떻게 선택해야 하나요?

**답변**

JPG는 이미지를 작은 점인 픽셀로 표현하는 **래스터 이미지 형식**입니다. 사진처럼 색상과 명암이 복잡한 이미지에 적합하지만, 크게 확대하면 흐려지거나 깨질 수 있습니다.

SVG는 선, 도형, 좌표 등의 정보로 이미지를 표현하는 **벡터 이미지 형식**입니다. 확대하거나 축소해도 선명함이 유지되므로 로고, 아이콘, 도표 등에 적합합니다.

| 구분 | JPG | SVG |
|---|---|---|
| 표현 방식 | 픽셀 | 선·도형·좌표 |
| 확대 시 화질 | 깨지거나 흐려질 수 있음 | 선명함 유지 |
| 적합한 용도 | 사진, 복잡한 이미지 | 로고, 아이콘, 도표 |
| 투명 배경 | 지원하지 않음 | 지원 |
| 편집 방법 | 이미지 편집 프로그램 | 디자인 프로그램 또는 코드 |
| 주요 장점 | 사진을 비교적 작은 용량으로 저장 | 크기를 변경해도 화질 유지 |
| 주요 단점 | 압축하거나 확대하면 화질 저하 | 사진처럼 복잡한 이미지에는 비효율적 |

용도에 따라 다음과 같이 선택할 수 있습니다.

- 사진을 저장할 때: JPG
- 로고나 아이콘을 저장할 때: SVG
- 크기를 자주 변경해야 하는 그래픽: SVG
- 투명 배경이 필요한 사진형 이미지: PNG 또는 WebP

파일 이름의 확장자만 `.jpg`에서 `.svg`로 변경한다고 실제 SVG 파일이 되는 것은 아닙니다. SVG로 사용하려면 파일 내부 데이터도 벡터 형식으로 변환해야 합니다.
