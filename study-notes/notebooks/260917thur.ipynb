##### 2. INNER JOIN과 LEFT JOIN

**질문:** `INNER JOIN`과 `LEFT JOIN`은 왜 그런 이름이며 결과가 어떻게 다른가?

**답변:** `INNER JOIN`은 두 테이블에서 **연결 조건이 일치하는 행만** 보여준다. `LEFT JOIN`은 SQL에서 **왼쪽에 적은 테이블의 모든 행**을 남기고, 연결되는 오른쪽 데이터가 없으면 그 부분을 `NULL`로 표시한다.

**예시 코드:**

```sql
-- 수강 신청 기록이 있는 학생만 조회
SELECT s.name, e.course_id
FROM students AS s
INNER JOIN enrollments AS e
    ON s.id = e.student_id;

-- 수강 신청 기록이 없는 학생도 조회
SELECT s.name, e.course_id
FROM students AS s
LEFT JOIN enrollments AS e
    ON s.id = e.student_id;
```

**실행 결과:** 민지가 수강 신청을 했고 준호는 하지 않았다고 가정하면, `INNER JOIN`에는 민지만 나온다. `LEFT JOIN`에는 민지와 준호가 모두 나오며 준호의 `course_id`는 `NULL`이다. 한 학생의 신청이 여러 건이면 결과에도 여러 줄이 나올 수 있다.

##### 3. GROUP BY status

**질문:** 상태별 집계 결과를 보려면 왜 `GROUP BY status`가 필요한가?

**답변:** `GROUP BY status`는 같은 상태의 행을 한 그룹으로 묶는다. `COUNT(*)` 같은 집계 함수를 사용하면서 **상태마다 한 줄씩** 결과를 보려면 필요하다.

**예시 코드:**

```sql
SELECT status, COUNT(*) AS enrollment_count
FROM course_project.enrollments
GROUP BY status;
```

**실행 결과:** 데이터에 따라 `신청 3건`, `수강중 5건`, `취소 2건`처럼 상태별 건수가 각각 한 줄로 나온다. `GROUP BY` 없이 `COUNT(*)`만 조회하면 전체 행의 건수 한 줄이 나온다.

##### 4. 그래프 종류 선택

**질문:** 히스토그램과 막대그래프는 어떻게 다르며, 여러 그래프는 각각 언제 사용하는가?

**답변:** **히스토그램**은 가격이나 나이처럼 연속된 숫자를 구간으로 나눠 분포를 본다. **막대그래프**는 주문 상태나 상품 카테고리처럼 서로 다른 항목의 크기를 비교한다. 시간에 따른 변화는 **선그래프**, 두 숫자형 변수의 관계는 **산점도**, 상위 항목의 순위 비교는 **가로 막대그래프**가 적합하다.

**예시 코드:**

```python
import matplotlib.pyplot as plt

# 가격대별 상품 수: 히스토그램
plt.hist(
    [9000, 12000, 15000, 22000, 25000],
    bins=[0, 10000, 20000, 30000],
)
plt.xlabel("상품 가격")
plt.ylabel("상품 수")
plt.show()

# 주문 상태별 건수: 막대그래프
plt.bar(["신청", "수강중", "취소"], [3, 5, 2])
plt.xlabel("주문 상태")
plt.ylabel("건수")
plt.show()
```

**실행 결과:** 첫 그래프는 각 **가격 구간에 상품이 몇 개 있는지**, 두 번째 그래프는 **상태별 건수가 얼마나 다른지** 보여준다.

##### 5. 대시보드의 row span

**질문:** 대시보드에서 말하는 “로스펜(row span)”은 무엇인가?

**답변:** 대시보드를 격자로 배치할 때 카드나 차트가 **세로 방향으로 몇 칸을 차지하는지**를 뜻한다. 예를 들어 `row span 2`는 한 요소가 세로 두 칸의 공간을 차지한다는 의미다.

**예시 코드:**

```css
.chart-card {
    grid-row: span 2;
}
```

**실행 결과:** 해당 차트 카드가 대시보드 격자에서 세로 두 칸 높이로 배치된다.
