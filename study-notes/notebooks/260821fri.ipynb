- `input()`으로 입력받은 값은 항상 문자열이므로, 숫자로 사용하려면 목적에 맞게 형 변환해야 한다.
- `if`는 조건이 참일 때 코드를 실행하고, `else`는 조건이 거짓일 때 코드를 실행한다.

#### 상세 학습 내용

##### 1. input()으로 데이터 입력받기

`input()`은 사용자에게 값을 입력받을 때 사용하는 함수이다.

```python
name = input()
```

위 코드는 별도의 안내 문구 없이 사용자의 입력을 기다린다. 사용자가 입력한 값은 `name` 변수에 저장된다.

`input()` 안에 안내 문구를 작성할 수도 있다.

```python
name = input("이름: ")
```

화면에 `"이름: "`이라는 문구를 먼저 보여 준 뒤 사용자의 입력을 받는다. 입력한 값은 동일하게 `name` 변수에 저장된다.

두 코드 모두 같은 방식으로 값을 입력받지만, 안내 문구를 넣으면 사용자가 무엇을 입력해야 하는지 알 수 있어 이해하기 쉽다.

---

##### 2. 입력값의 데이터형

`input()`으로 입력받은 결과는 항상 문자열이다.

따라서 숫자를 입력받더라도 바로 계산에 사용하지 않고 목적에 맞는 데이터형으로 변환해야 한다.

정수를 입력받을 때는 `int()`를 사용할 수 있다.

```python
age = int(input("나이: "))
```

실수를 입력받을 때는 `float()`를 사용할 수 있다.

```python
height = float(input("키: "))
```

---

##### 3. if 조건문

`if`는 조건에 따라 코드를 실행할 때 사용한다.

기본 구조는 다음과 같다.

```python
if 조건식:
    참일 때 실행할 코드
```

`if` 뒤에는 참이나 거짓으로 판단할 수 있는 조건식이 들어간다.

`if`문 끝에는 문법에 따라 반드시 콜론(`:`)을 작성해야 한다. 콜론 뒤에서 줄을 바꾸면 일반적으로 공백 4칸이 들여쓰기된다.

```python
age = 20

if age >= 18:
    print("성인입니다.")
```

들여쓰기된 코드는 `if`문의 실행 영역이다. 들여쓰기가 없다면 해당 코드는 `if`문에 포함되지 않는다.

---

##### 4. if와 else

`else`는 `if`의 조건이 거짓일 때 실행된다.

```python
if 조건식:
    참일 때 실행할 코드
else:
    거짓일 때 실행할 코드
```

`else` 뒤에도 반드시 콜론(`:`)을 작성해야 한다.

```python
age = 17

if age >= 18:
    print("성인입니다.")
else:
    print("미성년자입니다.")
```

---

##### 5. elif로 여러 조건 처리하기

학점을 구하는 것처럼 조건이 여러 개일 때는 `elif`를 사용한다.

예를 들어 95점 이상은 `A+`, 88점 이상은 `A`로 처리하는 것처럼 여러 점수 조건을 순서대로 검사할 수 있다.

```python
if score >= 95:
    print("A+")
elif score >= 88:
    print("A")
else:
    print("나머지 학점")
```

`elif`는 위 조건이 거짓일 때 다음 조건을 검사한다. 따라서 위 조건과 다음 조건 사이에 해당하는 범위를 처리할 수 있다.

`else`는 앞에 작성한 모든 조건에 해당하지 않는 나머지 경우를 처리한다.

---

##### 6. 논리 연산자

논리 연산자를 사용하면 여러 조건이나 논리값을 조합할 수 있다.

```python
age = 22
has_ticket = False

print(age >= 18 and has_ticket)
print(age >= 18 or has_ticket)
print(not has_ticket)
```

실행 결과:

```text
False
True
True
```

`age >= 18`은 `22 >= 18`이므로 `True`이다.

##### and

`and`는 양쪽 조건이 모두 참일 때만 `True`가 된다.

```python
age >= 18 and has_ticket
```

- `age >= 18`은 `True`이다.
- `has_ticket`은 `False`이다.
- 두 조건이 모두 참은 아니므로 결과는 `False`이다.

##### or

`or`는 양쪽 조건 중 하나만 참이어도 `True`가 된다.

```python
age >= 18 or has_ticket
```

- `age >= 18`이 `True`이므로 결과는 `True`이다.

##### not

`not`은 참과 거짓을 반대로 바꾼다.

```python
not has_ticket
```

- `has_ticket`이 `False`이므로 `not has_ticket`은 `True`이다.

##### 1. 카페 주문 계산기

##### 문제

사용자에게 음료 이름, 가격, 수량을 입력받아 주문한 상품과 총금액을 출력한다.

입력받을 내용은 다음과 같다.

- 음료 이름
- 음료 한 잔의 가격
- 주문 수량

출력할 내용은 다음과 같다.

- 주문 상품
- 총금액

##### 정답

```python
drink_name = input("음료 이름: ")
price = int(input("가격: "))
quantity = int(input("수량: "))

total_price = price * quantity

print(f"주문 상품: {drink_name}")
print(f"총 금액: {total_price}원")
```

##### 설명

`input()`으로 음료 이름, 가격, 수량을 입력받는다. 음료 이름은 문자열로 사용하지만, 가격과 수량은 곱셈에 사용해야 하므로 `int()`로 정수로 변환한다.

음료 한 잔의 가격과 수량을 곱해 `total_price`에 저장하고, f-string을 사용해 주문 상품과 총금액을 출력한다.

#### 2. 운동 시간 계산기

##### 문제

사용자에게 이름, 하루 운동 시간, 운동 일수를 입력받아 총 운동 시간을 출력한다.

하루 운동 시간과 운동 일수를 곱해 총 운동 시간을 계산한 뒤 f-string으로 출력한다.

#### 4. 연령 안내 프로그램

##### 문제

사용자에게 나이를 입력받고 다음 기준에 맞는 연령 구분을 출력한다.

- 0~7세: 미취학
- 8~13세: 초등학생 연령
- 14~16세: 중학생 연령
- 17~19세: 고등학생 연령
- 20세 이상: 성인

다음 조건을 지켜서 작성한다.

- `int(input(...))`을 사용한다.
- `if-elif-else`를 사용한다.
- 경계값을 최소 5개 이상 테스트한다.

##### 정답

```python
age = int(input("나이: "))

if age <= 7:
    print("미취학")
elif age <= 13:
    print("초등학생 연령")
elif age <= 16:
    print("중학생 연령")
elif age <= 19:
    print("고등학생 연령")
else:
    print("성인")
```

##### 설명

`input()`으로 입력받은 나이는 문자열이므로 `int()`를 사용해 정수로 변환한다.

조건은 위에서부터 차례대로 검사한다. 첫 번째 조건에서 7세 이하를 검사했기 때문에 다음 `elif`의 `age <= 13`은 자연스럽게 8세부터 13세까지를 의미한다. 같은 방식으로 각 연령 범위를 차례대로 구분한다.

##### 경계값 테스트

다음과 같이 각 구간이 시작하거나 끝나는 값을 중심으로 테스트할 수 있다.

```text
0세  → 미취학
7세  → 미취학
8세  → 초등학생 연령
13세 → 초등학생 연령
14세 → 중학생 연령
16세 → 중학생 연령
17세 → 고등학생 연령
19세 → 고등학생 연령
20세 → 성인
```

#### 5. 쇼핑 할인 프로그램

##### 문제

사용자에게 상품 금액을 입력받아 다음 기준에 따라 할인 내용을 출력한다.

- 100,000원 이상: 10% 할인
- 50,000원 이상: 5% 할인
- 그 외: 할인 없음

할인 금액 계산이 어렵다면 할인율을 안내하는 메시지만 출력해도 된다.

추가 도전으로 산술 연산자를 이용해 실제 할인 금액과 최종 금액까지 계산한다.

##### 기본 정답

```python
price = int(input("상품 금액: "))

if price >= 100000:
    print("10% 할인")
elif price >= 50000:
    print("5% 할인")
else:
    print("할인 없음")
```

##### 기본 정답 설명

상품 금액은 계산과 비교에 사용해야 하므로 `int()`로 변환한다.

100,000원 이상인지를 먼저 확인하고, 해당하지 않으면 50,000원 이상인지를 확인한다. 두 조건에 모두 해당하지 않으면 `else`에서 할인 없음 메시지를 출력한다.

##### 추가 도전 정답

```python
price = int(input("상품 금액: "))

if price >= 100000:
    discount_rate = 0.1
elif price >= 50000:
    discount_rate = 0.05
else:
    discount_rate = 0

discount_amount = int(price * discount_rate)
final_price = price - discount_amount

print(f"할인 금액: {discount_amount}원")
print(f"최종 금액: {final_price}원")
```

##### 추가 도전 설명

상품 금액에 따라 할인율을 `discount_rate`에 저장한다. 상품 금액과 할인율을 곱해 할인 금액을 구하고, 원래 상품 금액에서 할인 금액을 빼서 최종 금액을 계산한다.

#### 6. 간단한 로그인 판정

##### 문제

다음과 같이 연습용 아이디와 비밀번호를 미리 정한다.

```python
saved_id = "python"
saved_password = "1234"
```

사용자에게 아이디와 비밀번호를 입력받는다.

두 값이 모두 맞으면 다음 내용을 출력한다.

```text
로그인 성공
```

두 값 중 하나라도 맞지 않으면 다음 내용을 출력한다.

```text
아이디 또는 비밀번호를 확인하세요.
```

다음 조건을 지켜서 작성한다.

- `and`를 사용한다.
- 문자열 비교 연산자 `==`를 사용한다.
- 비밀번호 같은 실제 개인정보는 사용하지 않고 연습용 값만 사용한다.

##### 정답

```python
saved_id = "python"
saved_password = "1234"

input_id = input("아이디: ")
input_password = input("비밀번호: ")

if input_id == saved_id and input_password == saved_password:
    print("로그인 성공")
else:
    print("아이디 또는 비밀번호를 확인하세요.")
```

##### 설명

사용자가 입력한 아이디와 저장된 아이디를 `==`로 비교하고, 입력한 비밀번호와 저장된 비밀번호도 `==`로 비교한다.

두 비교 결과가 모두 참이어야 로그인에 성공하므로 두 조건 사이에 `and`를 사용한다. 아이디와 비밀번호 중 하나라도 일치하지 않으면 `else`가 실행된다.

실습에서는 실제 아이디나 비밀번호와 같은 개인정보를 사용하지 않고 연습용 값을 사용한다.
