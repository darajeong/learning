##### 1. 파일 경로 구분자 이해

- **질문:** 슬래시(`/`)와 백슬래시(`\`)의 차이는 무엇인가?

- **답변:** 백슬래시(`\`)는 주로 Windows 경로에서 사용하고, 슬래시(`/`)는 Linux·macOS·웹 주소에서 사용한다. Python 등 대부분의 개발 도구에서는 Windows에서도 슬래시(`/`)를 지원하므로 코드에서는 `/`를 사용하는 편이 이식성이 좋다.

- **예시 코드:**

```python
# Windows에서도 사용할 수 있는 상대 경로
file_path = "../data/raw/customers.csv"
```

##### 2. 여러 DataFrame 관리: 리스트보다 딕셔너리 사용

- **질문:** 여러 CSV DataFrame을 리스트와 딕셔너리 중 어떤 방식으로 관리하는 것이 좋은가?

- **답변:** 여러 데이터셋은 순서 번호로 접근하는 리스트보다, 데이터 이름을 키로 사용하는 딕셔너리로 관리하는 편이 읽기 쉽고 실수할 가능성이 적다. 리스트는 `data[0]`이 어떤 데이터인지 기억해야 하지만, 딕셔너리는 `data["customers"]`처럼 의미가 바로 드러난다.

- **예시 코드:**

```python
raw_data = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "products": products,
}

print(raw_data["customers"])
```

##### 3. 결측값과 중복값 검사 원리

- **질문:** `isna()`에는 왜 `sum()`을 두 번 사용하고, `duplicated()`에는 한 번만 사용하는가?

- **답변:** `isna()`는 DataFrame의 모든 셀에 대해 결측 여부를 반환하므로, 첫 번째 `sum()`은 열별 결측값 수를 계산하고 두 번째 `sum()`은 그 열별 결과를 모두 더해 전체 결측값 수를 계산한다. 반면 `duplicated()`는 각 행 또는 특정 열 값이 중복인지 1차원 결과로 반환하므로 `sum()` 한 번이면 전체 중복 개수를 구할 수 있다.

- **예시 코드:**

```python
missing_count = frame.isna().sum().sum()
duplicated_count = frame.duplicated().sum()

print("전체 결측값 수:", missing_count)
print("전체 중복 행 수:", duplicated_count)
```

- **실행 결과:**

```text
전체 결측값 수: 3
전체 중복 행 수: 1
```

##### 4. UNION과 JOIN의 차이

- **질문:** SQL의 `UNION`과 `JOIN`은 어떻게 다른가?

- **답변:** `UNION`은 같은 구조의 테이블을 위아래로 붙여 행을 늘리는 기능이고, `JOIN`은 공통 키를 기준으로 서로 다른 테이블 정보를 옆으로 붙여 열을 늘리는 기능이다.

- **예시 코드:**

```sql
-- UNION: 1월 주문과 2월 주문을 아래로 이어 붙임
SELECT order_id, amount FROM january_orders
UNION
SELECT order_id, amount FROM february_orders;

-- JOIN: 주문 정보에 고객 정보를 옆으로 붙임
SELECT *
FROM orders
JOIN customers
    ON orders.customer_id = customers.customer_id;
```

##### 5. JOIN 종류와 미매칭 값 처리

- **질문:** `INNER JOIN`과 `LEFT JOIN`은 무엇이 다르며, LEFT JOIN에서 일치하는 주문이 없으면 어떻게 표시되는가?

- **답변:** `INNER JOIN`은 양쪽 테이블의 ON 조건이 일치하는 행만 남긴다. `LEFT JOIN`은 왼쪽 테이블의 모든 행을 남기며, 오른쪽 테이블에 일치하는 행이 없으면 오른쪽 열 값은 SQL에서 `NULL`로 표시된다. pandas에서는 보통 `NaN`으로 보인다.

- **예시 코드:**

```sql
SELECT *
FROM customers
LEFT JOIN orders
    ON customers.customer_id = orders.customer_id;
```

- **실행 결과:**

```text
customer_id  name  order_id  amount
C001         민지  O001      10000
C002         준호  NULL      NULL
```

##### 6. GROUP BY와 HAVING의 역할

- **질문:** `GROUP BY`와 `HAVING`은 무엇이 다른가?

- **답변:** `GROUP BY`는 같은 기준값을 가진 행들을 하나의 그룹으로 묶는다. `HAVING`은 그룹화가 끝난 뒤 합계, 평균, 개수 같은 집계 결과를 조건으로 그룹을 필터링한다. 반면 `WHERE`는 그룹화 전에 개별 행을 필터링한다.

- **예시 코드:**

```sql
SELECT customer_id, SUM(amount) AS total_amount
FROM orders
GROUP BY customer_id
HAVING SUM(amount) >= 100000;
```

- **실행 결과:**

```text
customer_id  total_amount
C001         110000
```

##### 7. 문자열 포함 검색

- **질문:** SQL에서 상품명에 특정 단어가 포함된 행을 찾으려면 무엇을 사용하는가?

- **답변:** `LIKE` 연산자와 `%` 와일드카드를 사용한다. `%Trek%`는 앞뒤에 어떤 문자열이 있어도 되고, 중간에 `Trek`이 포함된 값을 찾는다는 뜻이다.

- **예시 코드:**

```sql
SELECT *
FROM products
WHERE product_name LIKE '%Trek%';
```

##### 8. matplotlib와 seaborn의 관계

- **질문:** `matplotlib`와 `seaborn`은 어떻게 다르며 seaborn이 더 좋은가?

- **답변:** `matplotlib`는 그래프를 세밀하게 조절할 수 있는 기본 시각화 도구이고, `seaborn`은 matplotlib를 기반으로 DataFrame 분석용 그래프를 더 간단하고 보기 좋게 만들도록 돕는 라이브러리다. seaborn이 항상 더 좋은 것은 아니며, 보통 seaborn으로 그래프를 만들고 matplotlib로 제목·축·크기 등을 조절한다.

- **예시 코드:**

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.barplot(
    data=category_sales,
    x="category",
    y="total_sales",
)

plt.title("카테고리별 Completed 주문 기준 금액")
plt.xlabel("카테고리")
plt.ylabel("금액")
plt.show()
```
