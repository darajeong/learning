##### 1. 오류의 종류와 디버깅

- **질문:** 런타임 오류, 로직 에러, `pass`, 디버깅 전략은 각각 무엇인가?
- **답변:**
  - **런타임 오류(runtime error)**는 코드 문법에는 문제가 없지만 실행 도중 잘못된 값이나 연산 등으로 발생하는 오류입니다.
  - **`pass`**는 아직 코드를 작성하지 않은 함수나 조건문 등의 자리를 비워 두기 위해 아무 작업도 하지 않고 넘어가는 명령입니다.
  - **로직 에러(logic error)**는 프로그램이 멈추지 않고 실행되지만 잘못된 계산이나 조건 때문에 의도와 다른 결과가 나오는 오류입니다.
  - **디버깅 전략**은 오류를 감으로 수정하지 않고 오류 메시지와 변수 값을 확인하면서 원인을 단계적으로 좁혀 가는 방법입니다.
  - 개발에서는 코드를 작성하는 능력뿐 아니라 문제의 원인을 체계적으로 찾고 해결하는 **디버깅 능력도 매우 중요합니다.**

- **예시 코드:**

```python
# 런타임 오류: 실행 도중 ValueError 발생
number = int("abc")

# 로직 에러: 실행은 되지만 계산 방법이 잘못됨
price = 10000
discount = 1000
result = price + discount  # 빼야 하는데 더함

# pass: 나중에 작성할 코드의 빈자리
def calculate_shipping(price):
    pass
```

- **실행 결과:**

```text
ValueError: invalid literal for int() with base 10: 'abc'
```

> 런타임 오류는 실행 도중 프로그램이 멈출 수 있지만, 로직 에러는 실행이 끝나도 결과가 잘못될 수 있으므로 예상 결과와 실제 결과를 비교해야 합니다.

##### 2. `range()`와 `len()` 함께 사용하기

- **질문:** `range()`와 `len()`을 함께 사용하는 이유는 무엇인가?
- **답변:** 리스트의 요소 개수만큼 반복하면서 각 요소의 **인덱스 번호**를 사용하기 위해 함께 사용합니다.

- **예시 코드:**

```python
fruits = ["사과", "바나나", "포도"]

for i in range(len(fruits)):
    print(i, fruits[i])
```

- **실행 결과:**

```text
0 사과
1 바나나
2 포도
```

> 번호와 값이 모두 필요할 때는 `enumerate()`를 사용하는 것이 더 간단합니다.

```python
for i, fruit in enumerate(fruits):
    print(i, fruit)
```

##### 3. 디버깅 전략

- **질문:** 디버깅 전략을 잘 세운다는 것은 무슨 뜻인가?
- **답변:** 오류를 감으로 고치는 것이 아니라, **오류 메시지를 확인하고 원인을 단계적으로 좁혀 가는 것**을 의미합니다.

- **대표적인 순서:**
  1. 트레이스백의 마지막 줄을 확인합니다.
  2. 예외 이름과 오류 메시지를 확인합니다.
  3. 오류가 발생한 코드의 줄을 찾습니다.
  4. 변수의 값과 자료형을 확인합니다.
  5. 코드를 작은 부분으로 나누어 실행합니다.
  6. 한 번에 한 부분만 수정한 후 다시 테스트합니다.

- **예시 코드:**

```python
age = "스무 살"

print(age)
print(type(age))

number = int(age)
```

- **실행 결과:**

```text
스무 살
<class 'str'>
ValueError: invalid literal for int()
```

`age`의 값과 자료형을 확인하면 정수로 변환할 수 없는 문자열이라는 원인을 찾을 수 있습니다.

##### 4. 스모크 테스트

- **질문:** 스모크 테스트란 무엇이며 언제 사용하는가?
- **답변:** 프로그램을 자세히 검사하기 전에 **핵심 기능이 최소한 정상적으로 작동하는지 빠르게 확인하는 기본 테스트**입니다.

- **예시:** 계산기 프로그램이라면 다음 내용을 먼저 확인합니다.
  - 프로그램이 실행되는가?
  - 숫자를 입력받을 수 있는가?
  - 간단한 계산 결과가 올바른가?
  - 실행하자마자 오류가 발생하지 않는가?

- **예시 코드:**

```python
def add(a, b):
    return a + b

assert add(1, 2) == 3
print("기본 기능 정상")
```

- **실행 결과:**

```text
기본 기능 정상
```

##### 5. Django와 React의 역할

- **질문:** Django와 React는 무엇이며 역할이 어떻게 다른가?
- **답변:**
  - **Django:** Python으로 서버와 백엔드를 만드는 웹 프레임워크입니다.
  - **React:** JavaScript로 사용자가 보는 화면을 만드는 라이브러리입니다.

- **역할 예시:**
  - React가 로그인 화면과 버튼을 보여줍니다.
  - 사용자가 로그인 버튼을 누르면 React가 Django에 요청합니다.
  - Django가 아이디와 비밀번호를 검사합니다.
  - Django가 결과를 보내면 React가 성공 또는 실패 화면을 표시합니다.

```text
사용자
  ↓
React: 화면 표시와 사용자 입력
  ↓ API 요청
Django: 요청 처리와 데이터베이스 작업
  ↓ 응답
React: 결과를 화면에 표시
```

##### 8. 엔터티와 테이블

- **질문:** 엔터티란 무엇이며 엔터티와 테이블은 어떤 관계인가?
- **답변:**
  - **엔터티(Entity):** 데이터로 관리하려는 대상의 종류입니다.
  - **테이블(Table):** 엔터티의 데이터를 실제로 저장하는 표입니다.

- **예시:**
  - 회원 → 엔터티
  - 회원 테이블 → 회원 정보를 저장하는 표
  - 상품 → 엔터티
  - 상품 테이블 → 상품 정보를 저장하는 표

```text
회원 테이블

회원번호 | 이름 | 이메일
--------|------|-----------------
1       | 민수 | minsu@email.com
2       | 지수 | jisu@email.com
```

- `회원`은 관리할 대상인 **엔터티**입니다.
- `회원 테이블`은 회원 데이터를 저장하는 **표**입니다.
- `민수`와 `지수`는 테이블에 저장된 실제 회원 데이터입니다.

##### 9. 행과 키를 이용한 테이블 연결

- **질문:** 각 테이블의 한 행은 무엇을 의미하며, 테이블끼리는 어떤 키로 연결하는가?
- **답변:** 한 행은 실제 데이터 하나를 의미하고, 테이블은 주로 **기본 키(PK)**와 **외래 키(FK)**로 연결합니다.

- **회원 테이블:**

| 회원번호(PK) | 이름 |
|---:|---|
| 1 | 민수 |
| 2 | 지수 |

회원 테이블의 한 행은 **회원 한 명**을 의미합니다.

- **주문 테이블:**

| 주문번호(PK) | 회원번호(FK) | 상품 |
|---:|---:|---|
| 101 | 1 | 운동화 |
| 102 | 2 | 가방 |

주문 테이블의 한 행은 **주문 한 건**을 의미합니다.

- **키의 역할:**
  - `회원번호(PK)`: 회원 한 명을 고유하게 구별합니다.
  - `주문번호(PK)`: 주문 한 건을 고유하게 구별합니다.
  - `회원번호(FK)`: 주문한 회원이 누구인지 회원 테이블과 연결합니다.

```text
회원번호 1번 민수 ── 주문번호 101번 운동화
회원번호 2번 지수 ── 주문번호 102번 가방
```

##### 10. ERD

- **질문:** 엔터티와 테이블의 구조 및 연결 관계를 도면으로 그린 것이 ERD인가?
- **답변:** 맞습니다. **ERD(Entity Relationship Diagram)**는 테이블의 정보와 테이블 사이의 관계를 그림으로 나타낸 데이터베이스 설계도입니다.

```text
[회원]
- 회원번호(PK)
- 이름
- 이메일
      │
      │ 회원 1명은 여러 주문 가능
      ▼
[주문]
- 주문번호(PK)
- 회원번호(FK)
- 상품
```

- **ERD에서 확인할 수 있는 내용:**
  - 어떤 테이블이 있는가?
  - 각 테이블에는 어떤 열이 있는가?
  - 기본 키와 외래 키는 무엇인가?
  - 테이블끼리 어떻게 연결되는가?
  - 한 회원이 주문을 몇 건 가질 수 있는가?

위 관계는 회원 한 명이 여러 주문을 할 수 있으므로 **1:N 관계**입니다.

##### 11. 데이터베이스 핵심 개념

- **질문:** PostgreSQL, DBeaver, SQL, 데이터 모델링, 정규화, 데이터 무결성, 트랜잭션, 인덱스, 보안, 백업·복구, RDBMS와 NoSQL은 각각 무엇인가?
- **답변:**
  - **PostgreSQL:** 데이터를 표 형태로 저장하고 관리하는 관계형 데이터베이스 프로그램입니다.
  - **DBeaver:** 데이터베이스의 표를 확인하고 SQL을 실행할 수 있는 관리 도구입니다.
  - **SQL:** 데이터베이스에 데이터 조회·추가·수정·삭제 명령을 내리는 언어입니다.
  - **데이터 모델링:** 어떤 테이블과 열을 만들고 어떻게 연결할지 설계하는 과정입니다.
  - **정규화:** 중복 데이터를 줄이기 위해 데이터를 여러 테이블로 적절하게 나누는 작업입니다.
  - **데이터 무결성:** 데이터가 정확하고 일관된 상태를 유지하도록 만드는 규칙입니다.
  - **트랜잭션:** 여러 데이터 작업을 하나의 작업 단위로 묶어 모두 성공하거나 모두 취소되게 하는 기능입니다.
  - **인덱스:** 데이터를 빠르게 검색할 수 있도록 만드는 색인입니다.
  - **보안:** 허가된 사용자만 데이터에 접근하거나 변경하도록 보호하는 것입니다.
  - **백업과 복구:** 데이터를 별도로 저장하고 문제가 발생했을 때 되살리는 작업입니다.
  - **RDBMS:** 표와 테이블 관계를 중심으로 데이터를 저장하는 시스템입니다.
  - **NoSQL:** 고정된 표 구조를 덜 사용하고 다양한 형태의 데이터를 유연하게 저장하는 시스템입니다.

- **SQL 예시 코드:**

```sql
SELECT name, email
FROM member
WHERE member_id = 1;
```

- **실행 결과:**

```text
name | email
-----|-----------------
민수 | minsu@email.com
```

- **RDBMS와 NoSQL의 사용 예시:**
  - 은행 계좌, 쇼핑몰 주문처럼 정확한 관계와 일관성이 중요함 → **RDBMS**
  - 채팅 메시지나 대량의 로그처럼 구조가 다양하고 빠른 처리가 중요함 → **NoSQL**
````
