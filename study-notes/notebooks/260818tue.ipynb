##### 오늘의 핵심 요약

- 프로젝트 경험은 수상 여부뿐만 아니라 참여 목적, 문제 해결 과정, 자신의 역할과 개선 내용을 설명할 수 있어야 한다.
- Python 코드는 읽기 편하도록 작성하고 들여쓰기를 지켜야 한다.
- Python 파일을 실행하기 전에는 수정한 내용을 저장해야 한다.
- 가상환경을 따로 만들지 않으면 기본 Python 환경으로 파일이 실행된다.
- 터미널의 화살표 키를 이용하면 이전에 실행했던 명령을 다시 찾을 수 있다.

##### 1. 프로젝트 경험 정리하기

위비티에서 다음과 관련된 공모전이나 대외 활동을 확인할 수 있다.

- 프로젝트
- 경험
- 팀 협업
- 문제 정의
- 데이터 분석
- AI 활용
- 발표
- 외부 평가
- 수상 가능성
- 포트폴리오

프로젝트나 공모전에서 반드시 수상해야 하는 것은 아니다. 수상 여부보다 해당 경험을 면접에서 설명할 수 있어야 한다.

프로젝트 경험을 정리할 때는 다음 질문을 확인한다.

- 왜 참가했는가?
- 무슨 문제를 해결했는가?
- 내 역할은 무엇이었는가?
- 어떤 데이터를 사용했는가?
- 무엇을 제출했는가?
- 어떤 피드백을 받았는가?
- 무엇을 개선했는가?

##### 2. 읽기 편한 코드 작성하기

보통 변수 선언이 끝나면 한 줄을 띄운다. 코드 사이에 적절한 빈 줄을 넣으면 읽기 편하다.

들여쓰기는 스페이스 4개 또는 탭 하나를 사용할 수 있다.

---

##### 3. Python 파일 실행하기

다음 명령어를 사용해 Python 파일을 실행할 수 있다.

```bash
python 00_01.py
```

이 명령어는 Python 인터프리터인 `python`으로 `00_01.py` 파일을 실행하라는 의미이다.

파일을 수정했다면 실행하기 전에 먼저 저장해야 한다. 수정한 내용이 저장되지 않은 경우 VS Code 오른쪽에 흰색 동그라미가 표시된다.

---

##### 4. 기본 Python 환경에서 실행하기

별도의 가상환경이 없어도 Python 파일을 실행할 수 있다. 이 경우 기본 Python 환경으로 실행된다.

Python의 전체 경로가 설정되어 있으면 별도의 가상환경을 만들지 않아도 여러 위치에서 Python을 실행할 수 있다.

그러나 가상환경을 따로 사용하지 않으면 추후 불필요한 패키지가 적절하지 않은 환경에 설치될 수 있다.
