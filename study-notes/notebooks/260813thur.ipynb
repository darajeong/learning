##### 3. Python 가상환경 만들기

다음 명령어를 사용해 Python 가상환경을 만든다.

```bash
python -m venv .venv
```

이 명령어는 Python에서 `.venv`라는 이름의 가상환경을 만들라는 의미이다. `.venv`는 관례적으로 많이 사용하는 가상환경 이름이다.

명령어를 실행했을 때 오류가 발생하면 생성된 `venv` 파일을 삭제하고 다시 실행한다.

가상환경의 스크립트 폴더에는 다음 파일이 있어야 한다.

```text
activate
deactivate
```

`pip`는 패키지를 설치할 때 사용하는 명령어이며 가상환경에 함께 있어야 한다.

##### 6. Django 실행하기

다음 명령어를 사용해 Django를 실행한다.

```bash
python manage.py runserver
```

Django 실행 후 다음 주소가 열려야 한다.

```text
http://127.0.0.1:8000/
```

`http://127.0.0.1:8000/`은 `localhost:8000`과 같으며 자기 컴퓨터에서 실행 중인 서버를 가리킨다.

`Ctrl + F5`는 강력한 새로고침을 할 때 사용한다.

##### 8. 환경변수 공유하기

`.env.example`은 GitHub에 올라가는 파일이며 환경변수의 형식을 팀원과 공유하는 용도로 사용한다.

팀원들이 동일한 환경변수 이름을 사용할 수 있도록 키 이름을 동일하게 작성해야 한다.

`.env.example`에는 실제 비밀번호를 절대 작성하면 안 된다. 파일이 GitHub에 올라가기 때문이다.

각 팀원은 서로 다른 데이터베이스를 사용할 수 있다.

팀장이 `.env.example`을 커밋하고 Push하면 팀원은 Pull을 통해 파일을 내려받는다. 팀원은 `.env.example`을 복사해 자신의 환경변수 파일을 만든 뒤 각자의 비밀번호를 작성한다.

실제 비밀번호가 작성된 파일은 Git과 서버에 올라가지 않도록 제외해야 한다.

##### 9. CRUD

테이블에서는 다음 네 가지 기본 작업을 수행한다.

```text
C = Create
R = Read
U = Update
D = Delete
```

도서를 등록하는 작업은 `Create`에 해당하며 SQL에서는 `INSERT`를 사용한다.

네이버 블로그 글을 검색하는 작업은 `Read`에 해당하며 SQL에서는 `SELECT`를 사용한다.

스마트폰으로 사용하는 채팅도 대부분 CRUD 구조로 동작한다. CRUD를 기반으로 이후 비즈니스 기능이 확장된다.
