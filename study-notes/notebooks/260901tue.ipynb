##### 1. 파이썬·pandas·데이터 전처리 기초

**질문**  
파이썬, pandas의 메소드·속성·API, `read_csv()`, `__init__()`, `isna()`, PK, Series·DataFrame, 데이터 전처리와 관련된 기본 개념은 어떻게 이해하면 되는가?

**답변**

- 파이썬은 데이터 분석에 많이 사용되는 범용 프로그래밍 언어이며, 데이터 분석을 바탕으로 머신러닝과 딥러닝 같은 인공지능 기술을 구현하는 데도 활용된다.
- `df.head()`처럼 괄호 `()`를 붙여 실행하는 기능은 **메소드(method)**이고, `df.columns`처럼 객체가 가지고 있는 정보를 나타내는 것은 **속성(attribute)**이다.
- pandas가 제공하는 함수·메소드·속성 등을 어떤 방식으로 사용할 수 있는지 정해 놓은 전체 사용 방법을 **pandas API**라고 이해하면 된다.
- `pd.read_csv()`는 CSV 파일을 읽어 DataFrame으로 만드는 함수이며, 괄호 `()` 안에 파일 경로나 URL 등의 **인자(argument)**를 넣을 수 있다.
- 함수나 메소드를 정의할 때 괄호 안에 정해 놓는 입력값의 이름을 **파라미터(parameter)**라고 하고, 실제로 호출할 때 넣는 값을 **인자(argument)**라고 한다.
- 클래스의 `__init__()`은 객체가 만들어질 때 자동으로 실행되는 특수 메소드이며, `self.속성명 = 값`처럼 작성해 객체의 속성을 만들거나 초기화할 수 있다.
- pandas의 `isna()`는 각 값이 결측값인지 확인하여 결측값이면 `True`, 아니면 `False`를 반환한다.
- `True`와 `False`를 정수로 변환하면 각각 `1`과 `0`이 되므로, `isna().sum()`을 사용하면 결측값의 개수를 셀 수 있다.
- `isna()`처럼 `is`로 시작하는 이름은 보통 어떤 상태가 **맞는지 아닌지를 `True` 또는 `False`로 표현하는 기능이나 변수**에 자주 사용되는 관례이다.
- 데이터베이스에서는 각 행을 서로 구별하기 위한 고유한 식별자가 필요하며, 이를 위해 **기본 키(Primary Key, PK)**를 사용할 수 있다.
- DataFrame은 직접 만들 수도 있고 CSV·Excel·데이터베이스 등 외부 데이터를 불러와 만들 수도 있으므로, 반드시 CSV 파일로 만들어야 하는 것은 아니다.
- pandas의 **Series는 1차원 데이터 구조**이고 **DataFrame은 행과 열로 이루어진 2차원 데이터 구조**이므로, 단순히 '컬럼이 한 개냐 여러 개냐'로만 구분하는 것은 정확하지 않다.
- 데이터에 결측값·중복값·잘못된 자료형 등이 있으면 분석 목적에 맞게 처리할 수 있으며, 이렇게 분석 전에 데이터를 정리하고 변환하는 과정을 **데이터 전처리(preprocessing)**라고 한다.


##### 2. 메소드·속성과 API의 관계

**질문**  
파이썬에서 메소드(method)와 속성(attribute)은 API라고 부르는 건가?

**답변**  
메소드나 속성 **하나하나를 보통 API라고 부르지는 않는다.**

API는 쉽게 말하면 **프로그램이 다른 프로그램이나 기능을 사용할 수 있도록 만들어 놓은 사용 방법 전체**라고 생각하면 된다.

예를 들어 `pandas`에는 여러 메소드와 속성이 있다.

- `df.head()` → 앞부분 데이터 보기
- `df.mean()` → 평균 구하기
- `df.columns` → 열 이름 확인하기
- `df.shape` → 행과 열 개수 확인하기

이렇게 외부에서 사용할 수 있도록 제공된 **메소드, 속성, 함수 등의 사용 방법 전체를 넓게 API라고 할 수 있다.**

즉,

> **메소드·속성 = 우리가 실제로 사용하는 각각의 기능**  
> **API = 그런 기능들을 사용할 수 있도록 정해놓은 전체적인 사용 방법**

**예시 코드**

```python
import pandas as pd

df = pd.DataFrame({
    "score": [80, 90, 100]
})

print(df.mean())
print(df.columns)
```

**실행 결과**

```text
score    90.0
dtype: float64

Index(['score'], dtype='object')
```


##### 3. AVERAGE와 MEAN의 차이

**질문**  
데이터 분석에서 `AVERAGE`와 `MEAN`은 같은 의미인가?

**답변**  
초보 단계에서는 둘 다 **평균**이라고 생각해도 된다.

예를 들어 `10, 20, 30`의 평균은 다음과 같다.

`(10 + 20 + 30) ÷ 3 = 20`

다만 정확하게 구분하면:

- **Mean** → 보통 산술평균을 정확하게 의미
- **Average** → 일상적으로 평균을 뜻하는 조금 더 넓은 표현

그래서 Excel에서는 `AVERAGE()`라는 함수를 사용하고, Python의 pandas에서는 보통 `.mean()`을 사용한다.

**예시 코드**

```python
import pandas as pd

scores = pd.Series([10, 20, 30])

print(scores.mean())
```

**실행 결과**

```text
20.0
```


##### 5. GitHub 주소의 `/blob/` 의미

**질문**  
GitHub 주소의 `/blob/`은 무엇을 의미하며 파일이 대용량이라서 붙는 것인가?

**답변**  
`/blob/`은 **파일이 크다는 뜻이 아니다.**

GitHub에서 저장소 안의 **특정 파일을 웹페이지로 보여줄 때 사용하는 주소 형태**이다.

예를 들어 주소가 다음과 같다고 하자.

```text
github.com/사용자/저장소/blob/main/data.csv
```

여기서 `/blob/`은 단순히 **GitHub에서 파일을 보는 페이지**라는 의미다.

따라서 작은 파일에도 `/blob/`이 붙고 큰 파일에도 붙을 수 있다.

반대로 Python에서 CSV의 실제 내용 자체를 직접 읽으려면 보통 `raw` 형태의 주소를 사용한다.

> **blob = GitHub에서 파일을 보는 화면**  
> **raw = 파일 내용 자체**


##### 6. 결측값과 데이터 전처리

**질문**  
데이터에 결측값이 나오면 전처리를 해야 하는가? 그리고 데이터 전처리는 무엇인가?

**답변**  
데이터 전처리는 **본격적으로 데이터를 분석하기 전에 데이터를 사용할 수 있는 상태로 정리하는 과정**이다.

쉽게 비유하면 요리하기 전에 재료를 씻고 손질하는 것과 비슷하다.

예를 들어 데이터에 다음과 같은 문제가 있을 수 있다.

- 값이 비어 있음 → 결측값
- 같은 데이터가 여러 번 있음 → 중복값
- 숫자가 문자로 저장되어 있음 → 자료형 문제
- 지나치게 이상한 값이 있음 → 이상치

결측값이 있다고 해서 무조건 삭제해야 하는 것은 아니다. 분석 목적에 따라 삭제하거나 다른 값으로 채울 수 있다.

**예시 코드**

```python
import pandas as pd

df = pd.DataFrame({
    "score": [80, None, 100]
})

print(df)
```

**실행 결과**

```text
   score
0   80.0
1    NaN
2  100.0
```

결측값을 0으로 채운다면:

```python
df = df.fillna(0)

print(df)
```

**실행 결과**

```text
   score
0   80.0
1    0.0
2  100.0
```


##### 7. 대시보드에 연동하는 AI와 GPT·Gemini·Ollama 비교

**질문**  
보통 대시보드에 AI를 연동할 때 어떤 AI를 사용하며 GPT, Gemini, Ollama는 어떤 차이가 있는가?

**답변**  
웹사이트나 대시보드에 AI를 붙일 때는 보통 **AI API**를 이용한다.

대표적으로 다음과 같은 선택지가 있다.

| 구분 | GPT API | Gemini API | Ollama + 모델 |
|---|---|---|---|
| 제공 | OpenAI | Google | Ollama에서 여러 모델 실행 |
| AI 실행 장소 | OpenAI 서버 | Google 서버 | 내 PC 또는 내가 관리하는 서버 |
| 인터넷 | 필요 | 필요 | 로컬 실행이면 모델 호출 자체에는 불필요 |
| 사용료 | 사용량에 따라 발생 가능 | 무료 한도 또는 사용량에 따라 발생 가능 | 로컬 모델은 호출당 API 비용 없음 |
| 설치 | 모델 설치 불필요 | 모델 설치 불필요 | Ollama와 모델 설치 필요 |
| 관리 난이도 | 비교적 쉬움 | 비교적 쉬움 | 상대적으로 어려움 |
| 여러 사용자 처리 | 제공업체가 처리 | 제공업체가 처리 | 내 서버 성능의 영향을 크게 받음 |

예를 들어 Gemini를 사용하는 대시보드는 다음과 같은 구조가 될 수 있다.

```text
사용자
  ↓
웹 대시보드
  ↓
Django
  ↓
Gemini API
  ↓
Google의 Gemini 서버
```

Ollama를 직접 운영한다면:

```text
사용자
  ↓
웹 대시보드
  ↓
Django
  ↓
Ollama
  ↓
설치된 AI 모델
```

핵심은 **GPT와 Gemini는 외부 회사의 AI 서버를 API로 사용하는 방식이고, Ollama는 여러 AI 모델을 자신의 PC나 서버에서 직접 실행할 수 있게 해주는 도구**라는 점이다.


##### 8. Ollama와 GitHub Copilot의 차이

**질문**  
Ollama도 여러 AI 모델을 사용할 수 있다면 GitHub Copilot과 비슷한 것인가?

**답변**  
여러 모델을 사용할 수 있다는 점은 비슷해 보이지만 **역할 자체가 다르다.**

- **GitHub Copilot** → 개발자가 코딩할 때 사용하는 AI 코딩 도우미
- **Ollama** → 여러 AI 모델을 PC나 서버에서 실행할 수 있게 해주는 도구

식당으로 비유하면:

> **Copilot = 여러 요리를 주문해서 먹을 수 있는 식당**  
> **Ollama = 여러 요리사를 데려와 직접 일하게 할 수 있는 주방**

따라서 Ollama 자체가 Gemini나 GPT 같은 하나의 AI 모델인 것은 아니다.


##### 9. Gemini API의 속도와 사용 횟수 제한

**질문**  
Gemini API가 느리고 약 20번밖에 답변하지 못한다는 말이 맞는가?

**답변**  
Gemini API가 **무조건 20번만 사용할 수 있는 것은 아니다.**

Gemini API에는 모델과 이용 등급 등에 따라 여러 종류의 사용 제한이 존재한다.

대표적으로:

- **RPM (Requests Per Minute)** → 1분 동안 요청할 수 있는 횟수
- **TPM (Tokens Per Minute)** → 1분 동안 처리할 수 있는 토큰 수
- **RPD (Requests Per Day)** → 하루 동안 요청할 수 있는 횟수

예를 들어 어떤 조건에서 RPM이 20이라면:

```text
1분 동안

1번째 요청  → 가능
2번째 요청  → 가능
...
20번째 요청 → 가능
21번째 요청 → 제한에 걸릴 수 있음
```

따라서 누군가 **"20번밖에 안 된다"**고 했다면 먼저 그것이 **분당 20번인지, 하루 20번인지, 특정 모델의 제한인지** 확인해야 한다.

또한 Gemini가 느리다고 느껴지는 이유도 API 자체뿐 아니라 모델 종류, 입력 길이, 출력 길이, 서버 구조 등 여러 원인이 있을 수 있다.


##### 10. Gemini API 키를 여러 개 사용하면 빨라지는지

**질문**  
Gemini API 키를 여러 개 발급받아 사용하면 호출 제한이 늘어나거나 AI가 더 빨라지는가?

**답변**  
같은 Google Cloud 프로젝트에서 API 키를 여러 개 만든다고 해서 일반적으로 **속도나 프로젝트의 사용 한도가 키 개수만큼 증가하지 않는다.**

예를 들어 하나의 프로젝트에 다음과 같이 키가 있다고 하자.

```text
Google Cloud 프로젝트 A

├─ API Key 1
├─ API Key 2
└─ API Key 3
```

프로젝트의 RPM 제한이 20이라고 가정하면:

```text
Key 1 → 10번 요청
Key 2 → 5번 요청
Key 3 → 5번 요청

총 20번
```

이런 식으로 **프로젝트 전체 사용량을 기준으로 제한이 적용될 수 있다.**

즉,

> **API Key = AI를 사용하는 열쇠**  
> **API Key를 여러 개 만든다고 AI 엔진이 여러 개 생기는 것은 아님**

따라서 속도나 사용량 문제가 있다면 키를 여러 개 만드는 것보다 모델 선택, 요청 최적화, 사용 등급, 서버 구조 등을 확인하는 것이 중요하다.


##### 11. Ollama를 공유 대시보드에서 사용하는 방법

**질문**  
Ollama를 사용하려면 모든 학생이 각자 PC에 설치해야 하는가? 공유 대시보드에서는 AI가 어느 컴퓨터에서 실행되는가?

**답변**  
모든 학생이 Ollama를 설치할 필요는 없다.

한 대의 PC나 서버에 Ollama와 AI 모델을 설치하고, 학생들이 모두 그 서버를 사용하도록 만들 수 있다.

예를 들어 한 대의 공용 서버에 Ollama가 있다면:

```text
학생 A ─┐
학생 B ─┼→ PRD 대시보드 → Django → Ollama 서버 → AI 모델
학생 C ─┘
```

학생들은 웹사이트만 이용하고 실제 AI 모델은 **Ollama가 설치되어 있는 컴퓨터에서 실행된다.**

만약 내 개인 PC에 Ollama를 설치하고 모든 학생의 요청이 내 PC로 오도록 만들었다면:

```text
다른 학생
   ↓
PRD 사이트
   ↓
Django
   ↓
내 PC의 Ollama
   ↓
AI 모델
```

이 경우 실제 AI 계산은 **내 PC에서 이루어진다.**

따라서 여러 사람이 사용하는 서비스라면 개인 노트북보다 **항상 실행할 수 있는 공용 서버**를 사용하는 것이 더 적합하다.


##### 12. Ollama의 비용

**질문**  
Ollama는 무료로 사용할 수 있는가? 사용한다면 내가 비용을 내야 하는가?

**답변**  
Ollama를 이용해 **내 PC나 내가 관리하는 서버에서 로컬 모델을 직접 실행하는 경우**, Gemini처럼 질문할 때마다 Ollama에 API 사용료를 내는 구조는 아니다.

하지만 완전히 비용이 없는 것은 아니다.

예를 들어 다음과 같은 비용이 생길 수 있다.

- AI를 실행할 PC 또는 서버
- GPU
- 전기료
- 클라우드 서버를 빌린다면 서버 이용료

즉,

> **Gemini/GPT API → 다른 회사의 컴퓨터를 빌려 쓰고 사용량에 따라 비용을 낼 수 있음**  
> **Ollama 로컬 실행 → 내 컴퓨터로 직접 계산하므로 호출당 API 비용은 없지만 컴퓨터 자원이 필요함**

또한 Ollama의 클라우드 서비스를 이용하는 경우에는 로컬 실행과 비용 구조가 다를 수 있으므로 **"Ollama를 직접 서버에서 돌리는 것인지, Ollama Cloud를 사용하는 것인지"** 구분해야 한다.


##### 13. 개인 Gemini API 키를 프로젝트에 사용할 때 주의사항

**질문**  
내 Gemini API 키를 프로젝트에 사용하면 사용량과 비용이 내 쪽에서 차감되는가? 그리고 내 컴퓨터를 계속 켜둬야 하는가?

**답변**  
내 Google Cloud 프로젝트에서 발급받은 Gemini API 키를 사용한다면 **그 프로젝트의 사용량으로 기록된다.**

무료 사용 범위라면 해당 프로젝트의 무료 한도를 사용하고, 유료 결제가 연결되어 있다면 사용량에 따라 비용이 발생할 수 있다.

하지만 Gemini는 Google 서버에서 실행되기 때문에 **내 개인 컴퓨터를 계속 켜놓을 필요는 없다.**

예를 들어 Django 서버까지 별도의 서버에 배포되어 있다면:

```text
학생
 ↓
대시보드
 ↓
배포된 Django 서버
 ↓
내 Gemini API Key
 ↓
Google Gemini 서버
```

내 PC를 꺼도 위 구조에는 영향을 주지 않는다.

반대로 Django 서버 자체를 내 PC에서 실행하고 있다면 **내 PC가 꺼졌을 때 서비스도 사용할 수 없게 될 수 있다.**

또한 API 키는 반드시 서버에서 관리해야 한다.

```text
React / 브라우저
      ↓
Django 서버
      ↓
환경변수에 저장된 GEMINI_API_KEY
      ↓
Gemini
```

다음과 같이 프론트엔드 코드에 API 키를 직접 넣으면 안 된다.

**예시 코드**

```javascript
// 이렇게 하면 안 됨
const apiKey = "내_Gemini_API_Key";
```

브라우저에서 다른 사람이 키를 확인할 수 있기 때문이다.

따라서 중요한 주의사항은 다음과 같다.

- API 키를 GitHub에 올리지 않기
- React나 JavaScript에 직접 넣지 않기
- Django 서버의 환경변수 또는 Secret으로 관리하기
- 사용량을 주기적으로 확인하기
- 가능하면 개인 프로젝트보다 팀/서비스 전용 프로젝트를 사용하기


##### 14. PRD와 무관한 질문 및 API 악용 방지

**질문**  
PRD용 AI에 `대전 맛집 소개해줘` 같은 PRD와 관계없는 질문도 할 수 있는가? 다른 사용자가 이런 식으로 AI를 계속 사용하면 내 API 사용량이 소모되는가?

**답변**  
별도의 제한을 만들지 않았다면 **PRD와 관계없는 질문도 Gemini에 전달될 수 있다.**

예를 들어 사용자가 다음과 같이 질문할 수 있다.

```text
대전 맛집 알려줘
파이썬 숙제 풀어줘
영어 자기소개서 작성해줘
```

이 요청들이 Gemini API까지 전달되면 **PRD와 관계없는 질문이라도 API 사용량이 발생한다.**

특히 내 API 키를 사용하고 있다면 해당 요청도 **내 키가 속한 프로젝트의 사용량으로 기록된다.**

그래서 PRD 대시보드에서는 다음과 같은 방어 장치를 두는 것이 좋다.

- **로그인한 사용자만 AI 사용 가능**
- **사용자별 일일 사용 횟수 제한**
- **한 번에 입력할 수 있는 글자 수 제한**
- **PRD와 관계없는 질문 차단**
- **사용자별 AI 사용량 로그 저장**
- **전체 서비스의 AI 사용량 제한**

AI에게 다음과 같이 역할을 지정할 수도 있다.

**예시 코드**

```text
너는 PRD 작성 지원 AI다.

PRD 작성, 요구사항 정의, 기능 정의, 사용자 문제,
KPI, 페인포인트 등 프로젝트 기획과 관련된 질문에만 답변한다.

관련 없는 질문에는
"PRD 작성과 관련된 질문만 도와드릴 수 있습니다."
라고 답변한다.
```

그러면 다음과 같이 동작하도록 만들 수 있다.

**실행 결과**

```text
사용자:
대전 맛집 추천해줘.

AI:
PRD 작성과 관련된 질문만 도와드릴 수 있습니다.
```

다만 여기서 중요한 점이 있다.

질문을 **Gemini에게 보낸 다음 Gemini가 거절하게 하면 이미 API를 호출한 것이므로 사용량이 발생한다.**

더 확실하게 사용량을 아끼려면:

```text
사용자 질문
   ↓
Django에서 먼저 검사
   ↓
PRD 관련 질문인가?
   ↓
 ┌───────┴───────┐
아니오            예
 ↓                ↓
바로 차단       Gemini API 호출
 ↓                ↓
API 사용 X       API 사용 O
```

즉, 비용과 악용 방지 측면에서는 **Gemini에 보내기 전에 Django에서 불필요한 요청을 차단하는 구조**가 더 좋다.
