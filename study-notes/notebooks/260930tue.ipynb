##### 1. PHP는 무슨 뜻인가?

**질문**

PHP는 무슨 뜻이며 웹사이트에서 어떤 역할을 하는가?

**답변**

PHP는 **PHP: Hypertext Preprocessor**의 약자로, 웹 서버에서 실행되는 **서버 측 프로그래밍 언어**입니다.

PHP는 다음과 같은 작업을 담당합니다.

- MySQL과 같은 데이터베이스에 접속
- SQL 실행
- 로그인 정보와 사용자 입력 처리
- 조건과 데이터에 따라 HTML 생성
- 생성한 HTML을 브라우저에 전송

PHP 코드는 일반적으로 브라우저에서 직접 실행되지 않습니다. 서버가 PHP 코드를 실행하고, 브라우저에는 실행 결과인 HTML을 전달합니다.

**예시 코드**

```php
<?php
$name = '홍길동';
?>

<h1>안녕하세요, <?= htmlspecialchars($name) ?>님!</h1>
```

**실행 결과**

브라우저가 받는 결과는 다음과 같은 HTML입니다.

```html
<h1>안녕하세요, 홍길동님!</h1>
```

브라우저에는 원래의 PHP 코드가 아니라 PHP가 생성한 결과가 표시됩니다.

##### 2. 정적 페이지와 동적 페이지의 차이는 무엇인가?

**질문**

정적 페이지와 동적 페이지는 무엇이 다른가?

**답변**

**정적 페이지**는 서버에 미리 저장된 HTML 파일을 주로 그대로 브라우저에 전달하는 페이지입니다.

- 모든 사용자에게 주로 같은 내용을 표시합니다.
- 데이터베이스 조회가 필수적이지 않습니다.
- 구조가 단순하고 빠릅니다.
- 내용을 바꾸려면 일반적으로 파일을 수정해야 합니다.

**동적 페이지**는 요청이 들어올 때 서버 프로그램이 사용자 정보, 입력값 또는 데이터베이스 내용에 따라 새로운 결과를 만드는 페이지입니다.

- 사용자나 상황에 따라 다른 내용을 표시할 수 있습니다.
- PHP, Java, Python, JavaScript 등의 서버 프로그램을 사용할 수 있습니다.
- 데이터베이스와 연결할 수 있습니다.
- 로그인, 게시판, 쇼핑몰, 관리자 페이지 등에 적합합니다.

파일 확장자가 `.html`인지 `.php`인지보다 **서버가 요청과 데이터에 따라 결과를 새로 만드는지**가 더 중요한 구분 기준입니다.

**예시 코드**

정적 페이지:

```html
<h1>학생 목록</h1>

<ul>
  <li>김민지</li>
  <li>이서준</li>
</ul>
```

동적 페이지:

```php
<?php
$stmt = $pdo->query(
    'SELECT name FROM students ORDER BY student_id'
);

$students = $stmt->fetchAll();
?>

<h1>학생 목록</h1>

<ul>
<?php foreach ($students as $student): ?>
  <li><?= htmlspecialchars($student['name']) ?></li>
<?php endforeach; ?>
</ul>
```

**실행 결과**

정적 페이지는 HTML 파일에 직접 입력한 학생만 표시합니다.

동적 페이지는 데이터베이스의 현재 학생 목록을 표시하므로 MySQL에 학생이 추가되면 다음 조회부터 새로운 학생도 화면에 나타날 수 있습니다.

##### 3. 브라우저부터 데이터베이스까지 웹은 어떻게 동작하는가?

**질문**

사용자가 주소를 입력한 뒤 데이터베이스의 내용이 브라우저에 표시되기까지 어떤 과정이 진행되는가?

**답변**

일반적인 동작 순서는 다음과 같습니다.

1. 사용자가 브라우저에 웹 주소를 입력합니다.
2. DNS가 도메인 이름을 서버의 IP 주소로 변환합니다.
3. 브라우저가 해당 서버에 HTTP 또는 HTTPS 요청을 보냅니다.
4. Apache와 같은 웹 서버가 요청을 받습니다.
5. 정적 파일 요청이면 Apache가 파일을 직접 반환합니다.
6. PHP 파일 요청이면 Apache가 PHP 실행 환경에 처리를 전달합니다.
7. PHP가 필요한 SQL을 MySQL에 전달합니다.
8. MySQL이 SQL을 실행하고 결과를 PHP에 반환합니다.
9. PHP가 데이터베이스 결과를 이용해 HTML을 생성합니다.
10. Apache가 생성된 HTML을 HTTP 응답으로 브라우저에 보냅니다.
11. 브라우저가 HTML, CSS, JavaScript를 해석해 화면을 표시합니다.

전체 흐름은 다음과 같습니다.

```text
사용자
  ↓
브라우저
  ↓ DNS로 서버 주소 확인
웹 서버의 IP 주소
  ↓ HTTP 또는 HTTPS 요청
Apache
  ↓ PHP 처리 요청
PHP
  ↓ SQL
MySQL
  ↓ 조회 결과
PHP
  ↓ 생성된 HTML
Apache
  ↓ HTTP 응답
브라우저
```

브라우저가 일반적으로 MySQL에 직접 접속하지 않는 이유는 다음과 같습니다.

- 데이터베이스 계정과 비밀번호 보호
- 사용자 인증과 권한 검사
- 입력값 검증
- SQL Injection 방지
- 애플리케이션의 업무 규칙 처리
- 데이터베이스의 외부 노출 방지

##### 4. DNS, Apache, PHP, MySQL, SQL은 각각 어떤 역할을 하는가?

**질문**

DNS, Apache, PHP, MySQL, SQL은 웹 시스템에서 각각 무엇을 담당하는가?

**답변**

- **DNS**는 `example.com`과 같은 도메인 이름을 서버의 IP 주소로 변환합니다.
- **Apache**는 브라우저의 HTTP 요청을 받는 웹 서버입니다.
- **PHP**는 서버에서 요청을 처리하고 필요한 HTML을 생성하는 프로그래밍 언어입니다.
- **MySQL**은 데이터를 테이블 형태로 저장하고 관리하는 관계형 데이터베이스 관리 시스템입니다.
- **SQL**은 관계형 데이터베이스에 조회·등록·수정·삭제 등의 작업을 지시하는 언어입니다.

식당에 비유하면 다음과 같습니다.

- **DNS**: 식당 주소를 알려주는 내비게이션
- **Apache**: 손님의 주문을 받는 종업원
- **PHP**: 요청에 맞게 음식을 만드는 요리사
- **MySQL**: 식재료를 보관하는 창고
- **SQL**: 창고에 보내는 작업 요청서
- **브라우저**: 완성된 결과를 사용자에게 보여주는 화면

`localhost`로 접속하는 로컬 실습에서는 일반적으로 외부 DNS를 조회하지 않고 자기 컴퓨터의 서버에 접속합니다.

SQL은 데이터베이스 제품이 아니라 데이터베이스를 다루는 언어입니다.

**예시 코드**

```sql
SELECT student_id, name, email
FROM students
ORDER BY student_id DESC
LIMIT 10;
```

**실행 결과**

`students` 테이블에서 학생 번호, 이름, 이메일을 조회한 뒤 학생 번호가 큰 순서로 최대 10건을 반환합니다.

##### 5. Modern Full Stack과 LAMP 방식의 차이는 무엇인가?

**질문**

Modern Full Stack 방식과 LAMP 방식은 구조적으로 무엇이 다른가?

**답변**

두 방식 모두 완전한 웹 애플리케이션을 만들 수 있으며, 주된 차이는 **구성 요소를 나누는 방법과 통신 방식**입니다.

**LAMP 방식**

LAMP는 전통적으로 다음 조합을 의미합니다.

- **L — Linux**: 운영체제
- **A — Apache**: 웹 서버
- **M — MySQL**: 데이터베이스
- **P — PHP**: 서버 프로그램

대표적인 흐름은 다음과 같습니다.

```text
브라우저
  ↓ HTTP 요청
Apache
  ↓
PHP
  ↓ SQL
MySQL
  ↓ 조회 결과
PHP가 HTML 생성
  ↓
브라우저
```

PHP가 데이터베이스를 조회하고 완성된 HTML을 만들어 반환하는 **서버 측 렌더링 방식**으로 많이 사용됩니다.

Windows에서 Apache, MySQL, PHP를 사용하면 엄밀히 말해 **WAMP**에 가깝습니다.

**Modern Full Stack 방식**

Modern Full Stack은 하나의 고정된 기술 조합을 뜻하지 않으며, 프런트엔드와 백엔드를 분리하는 구조를 가리키는 경우가 많습니다.

예시는 다음과 같습니다.

```text
React + TypeScript
  ↓ HTTP 요청 및 JSON 응답
FastAPI
  ↓ SQL
MySQL 또는 PostgreSQL
```

각 기술의 역할은 다음과 같습니다.

- **React + TypeScript**: 브라우저 화면 구성
- **FastAPI**: 서버 로직 처리와 API 제공
- **JSON**: 프런트엔드와 백엔드가 주고받는 대표적인 데이터 형식
- **MySQL 또는 PostgreSQL**: 데이터 저장 및 관리

LAMP에서도 API를 만들 수 있고 Modern Full Stack에서도 서버가 HTML을 생성할 수 있으므로 두 방식을 완전히 반대되는 개념으로 볼 필요는 없습니다.

##### 6. Modern Full Stack과 LAMP를 왜 비교하며, 둘은 어떤 관계인가?

**질문**

두 방식을 비교해서 배우는 이유는 무엇이며 서로 어떤 관계가 있는가?

**답변**

두 방식은 기술과 구조가 다르지만 웹의 기본 원리는 같다는 점을 이해하기 위해 비교합니다.

공통 원리는 다음과 같습니다.

```text
사용자 요청
  ↓
서버 애플리케이션
  ↓
데이터베이스 처리
  ↓
서버 응답
  ↓
사용자 화면
```

LAMP는 구성과 흐름이 비교적 단순해서 웹과 데이터베이스가 연결되는 원리를 처음 학습하기 좋습니다.

Modern Full Stack은 프런트엔드와 백엔드를 분리하여 다음과 같은 장점을 얻을 수 있습니다.

- 웹과 모바일 앱에서 같은 API 사용
- 프런트엔드와 백엔드의 독립적인 개발
- 구성 요소별 배포와 확장
- 대규모 팀의 역할 분담
- 풍부한 브라우저 상호작용 구현

LAMP에서 배우는 HTTP, 서버 처리, SQL, 데이터베이스, 인증, 보안 등의 원리는 Modern Full Stack에서도 그대로 사용됩니다.

LAMP는 단순히 오래된 연습용 기술을 뜻하지 않으며 현재도 실제 서비스에서 사용됩니다.

##### 7. 두 방식 모두 웹사이트를 만드는가? 왜 여러 방식이 존재하는가?

**질문**

LAMP와 Modern Full Stack 모두 웹사이트를 만들 수 있다면 왜 하나로 통일하지 않는가?

**답변**

두 방식 모두 다음과 같은 웹사이트와 웹 애플리케이션을 만들 수 있습니다.

- 회사 소개 사이트
- 블로그
- 게시판
- 관리자 시스템
- 쇼핑몰
- 예약 서비스
- 온라인 교육 시스템

여러 방식이 존재하는 이유는 프로젝트마다 요구사항이 다르기 때문입니다.

- 소규모 사이트는 단순한 PHP 구조가 개발과 운영에 유리할 수 있습니다.
- 화면 상호작용이 많은 서비스는 React 같은 프런트엔드 기술이 유리할 수 있습니다.
- 모바일 앱과 웹이 같은 데이터를 사용하면 API 분리 구조가 편리할 수 있습니다.
- 개발자의 경험과 조직의 기술 역량에 따라 선택이 달라집니다.
- 개발 기간, 서버 비용, 성능, 보안, 유지보수 조건이 서로 다릅니다.
- 이미 운영 중인 시스템과 호환해야 할 수도 있습니다.

따라서 최신 기술이라고 항상 더 좋은 것은 아니며, 오래된 기술이라고 항상 나쁜 것도 아닙니다.

프로젝트의 목적과 조건에 가장 잘 맞는 구조를 선택하는 것이 중요합니다.

##### 8. AMPPS는 무엇이며 왜 설치하는가?

**질문**

AMPPS는 무엇이고 PHP·MySQL 웹 개발을 위해 왜 설치하는가?

**답변**

AMPPS는 로컬 컴퓨터에 웹 개발 환경을 한 번에 설치하고 관리할 수 있도록 도와주는 프로그램 묶음입니다.

일반적으로 다음 구성 요소를 포함하거나 관리합니다.

- Apache
- MySQL 또는 MariaDB
- PHP
- phpMyAdmin
- 웹 애플리케이션 설치 및 관리 도구

Apache, PHP, MySQL을 각각 따로 설치하면 프로그램 사이의 연결과 포트를 직접 설정해야 합니다.

AMPPS를 사용하면 필요한 프로그램을 한 번에 구성하고 제어판에서 실행하거나 중지할 수 있어 로컬 실습이 쉬워집니다.

AMPPS, XAMPP, WampServer는 모두 비슷한 목적을 가진 로컬 개발 환경 도구이므로 일반적으로 한 실습에서는 하나만 선택해서 사용합니다.

여러 개발 환경을 동시에 실행하면 다음 포트가 충돌할 수 있습니다.

- Apache HTTP: `80`
- Apache HTTPS: `443`
- MySQL: `3306`

AMPPS를 설치한다고 웹사이트가 자동으로 만들어지는 것은 아니며, 웹사이트를 개발하고 실행할 수 있는 로컬 서버 환경이 준비되는 것입니다.

##### 9. P2P 방식으로 동작한다는 것은 무슨 뜻인가?

**질문**

P2P 방식으로 동작한다는 것은 무엇이며 일반적인 웹 방식과 무엇이 다른가?

**답변**

P2P는 **Peer-to-Peer**의 약자로, 참여하는 컴퓨터들이 중앙 서버에만 의존하지 않고 서로 직접 연결되어 데이터를 주고받는 방식입니다.

각 참여 컴퓨터는 상황에 따라 다음 역할을 모두 할 수 있습니다.

- 데이터를 요청하는 클라이언트
- 다른 참여자에게 데이터를 제공하는 서버

대표적인 예는 다음과 같습니다.

- BitTorrent 파일 공유
- 일부 블록체인 네트워크
- 직접 연결을 사용하는 일부 통신 프로그램

일반적인 웹사이트는 대부분 P2P가 아니라 **클라이언트-서버 방식**으로 동작합니다.

```text
브라우저(클라이언트)
  ↓ 요청
웹 서버
  ↓ 처리
데이터베이스
```

XAMPP나 AMPPS를 설치하여 한 컴퓨터에서 브라우저, Apache, PHP, MySQL을 모두 실행해도 이를 P2P라고 부르지는 않습니다.

같은 컴퓨터에서 실행되더라도 브라우저는 클라이언트 역할을 하고 Apache와 MySQL은 서버 역할을 하므로 논리적으로는 클라이언트-서버 구조입니다.

##### 10. Nginx는 무엇이며 Apache와 어떤 관계인가?

**질문**

Nginx는 무엇이며 Apache와 같은 프로그램인가?

**답변**

Nginx는 `엔진엑스`라고 읽으며, Apache와 마찬가지로 HTTP 요청을 처리하는 **웹 서버**입니다.

두 프로그램은 다음과 같은 역할을 수행할 수 있습니다.

- 정적 파일 제공
- HTTPS 연결 처리
- 다른 애플리케이션 서버로 요청 전달
- 리버스 프록시
- 로드 밸런싱
- 접근 제어
- 접속 기록 관리

Apache와 Nginx는 서로 대안이 될 수 있지만 함께 사용할 수도 있습니다.

가능한 구성은 다음과 같습니다.

- Apache만 사용
- Nginx만 사용
- Nginx가 외부 요청을 받고 내부 Apache로 전달
- Nginx가 PHP-FPM과 연결되어 PHP 처리

일반적인 특징은 다음과 같습니다.

- Apache는 다양한 모듈과 디렉터리별 설정 파일인 `.htaccess`를 활용하는 환경에서 편리합니다.
- Nginx는 이벤트 기반 구조를 사용하며 많은 동시 연결과 정적 파일 처리에 강점이 있습니다.
- 실제 성능은 웹 서버의 이름만이 아니라 설정, 애플리케이션, 하드웨어, 요청 특성에 따라 달라집니다.

PHP 입문 실습에서는 XAMPP나 WampServer에 포함된 Apache를 사용해도 충분합니다.

##### 11. `ALTER USER ... IDENTIFIED BY '1234'`와 `FLUSH PRIVILEGES`는 무엇을 하는가?

**질문**

다음 SQL은 무엇을 하는가?

```sql
ALTER USER 'root'@'localhost' IDENTIFIED BY '1234';
FLUSH PRIVILEGES;
```

**답변**

다음 명령은 `localhost`에서 접속하는 MySQL의 `root` 계정 비밀번호를 `1234`로 변경합니다.

```sql
ALTER USER 'root'@'localhost' IDENTIFIED BY '1234';
```

각 부분의 의미는 다음과 같습니다.

- `ALTER USER`: 기존 사용자 계정의 설정 변경
- `'root'@'localhost'`: 로컬 컴퓨터에서 접속하는 `root` 계정
- `IDENTIFIED BY '1234'`: 계정 비밀번호를 `1234`로 설정

다음 명령은 MySQL이 권한 관련 시스템 테이블의 내용을 다시 읽게 합니다.

```sql
FLUSH PRIVILEGES;
```

`CREATE USER`, `ALTER USER`, `GRANT`, `REVOKE` 같은 정상적인 계정·권한 명령을 사용하면 MySQL이 변경 내용을 즉시 반영하므로 일반적으로 `FLUSH PRIVILEGES`를 추가로 실행할 필요가 없습니다.

권한 테이블을 직접 수정한 특수한 상황에서는 필요할 수 있지만, 권한 테이블을 직접 수정하는 방식은 권장되지 않습니다.

`1234`는 추측하기 쉬운 비밀번호이므로 외부에 공개되지 않은 교육용 로컬 환경에서만 사용해야 합니다.

실제 서비스에서는 다음 원칙을 지켜야 합니다.

- 강력하고 고유한 비밀번호 사용
- 애플리케이션 전용 DB 계정 생성
- 필요한 데이터베이스에만 최소 권한 부여
- 비밀번호를 소스 코드나 Git 저장소에 기록하지 않기

**실행 결과**

명령이 성공하면 보통 다음과 비슷한 결과가 표시됩니다.

```text
Query OK, 0 rows affected
```

이후 MySQL에 접속할 때는 변경된 비밀번호를 사용합니다.

```text
Host: localhost
Port: 3306
User: root
Password: 1234
```

##### 12. 재귀 깊이를 늘린다는 것은 무슨 뜻인가?

**질문**

다음 명령에서 재귀 깊이를 늘린다는 것은 무슨 뜻인가?

```sql
SET SESSION cte_max_recursion_depth = 20000;
```

**답변**

재귀는 어떤 처리 과정이 이전 결과를 이용해 같은 작업을 반복하는 방식입니다.

MySQL의 **재귀 CTE**는 연속된 숫자나 계층 구조를 생성할 때 사용할 수 있습니다.

MySQL은 무한 반복과 과도한 자원 사용을 막기 위해 재귀 CTE의 반복 횟수를 제한합니다.

다음 명령은 현재 MySQL 연결에서 재귀 CTE가 반복할 수 있는 최대 깊이를 `20000`으로 설정합니다.

```sql
SET SESSION cte_max_recursion_depth = 20000;
```

각 부분의 의미는 다음과 같습니다.

- `SESSION`: 현재 데이터베이스 연결에만 설정 적용
- `cte_max_recursion_depth`: 재귀 CTE의 최대 반복 깊이
- `20000`: 허용할 최대 깊이

연결을 종료하면 세션 설정은 원래 값으로 돌아갈 수 있습니다.

재귀 깊이를 늘리는 것은 데이터가 자동으로 생성된다는 뜻이 아니라 이후 실행할 재귀 CTE가 더 많이 반복될 수 있도록 허용한다는 뜻입니다.

재귀 조건을 잘못 작성하면 불필요하게 많은 연산이 발생할 수 있으므로 반드시 종료 조건을 지정해야 합니다.

**예시 코드**

```sql
SET SESSION cte_max_recursion_depth = 20000;

WITH RECURSIVE seq AS (
    SELECT 1 AS n

    UNION ALL

    SELECT n + 1
    FROM seq
    WHERE n < 10000
)
SELECT COUNT(*) AS generated_count
FROM seq;
```

**실행 결과**

```text
generated_count
---------------
10000
```

`WHERE n < 10000`은 숫자가 계속 증가하다가 재귀를 끝내게 하는 종료 조건입니다.

##### 13. XAMPP에서 Apache와 MySQL을 실행하는 목적은 무엇인가?

**질문**

XAMPP 제어판에서 Apache와 MySQL의 `Start` 버튼을 누르는 이유는 무엇인가?

**답변**

프로그램을 설치한 것과 실제 서버 프로세스를 실행하는 것은 다릅니다.

XAMPP 제어판에서 `Start` 버튼을 눌러야 Apache와 MySQL이 요청을 받을 수 있습니다.

**Apache를 실행하는 목적**

- 브라우저의 HTTP 요청 수신
- `htdocs` 폴더의 웹 파일 제공
- PHP 파일을 PHP 실행 환경에 연결
- 생성된 HTML을 브라우저에 반환

예를 들어 다음 주소로 접속하면 Apache가 요청을 처리합니다.

```text
http://localhost/course-admin/students.php
```

**MySQL을 실행하는 목적**

- 데이터베이스 연결 요청 수신
- 테이블과 데이터 저장
- SQL 실행
- 조회 결과를 PHP 또는 DBeaver에 반환

전체 흐름은 다음과 같습니다.

```text
브라우저
  ↓
XAMPP의 Apache
  ↓
PHP 프로그램
  ↓
XAMPP의 MySQL
```

기본 포트는 일반적으로 다음과 같습니다.

- Apache HTTP: `80`
- Apache HTTPS: `443`
- MySQL: `3306`

다른 프로그램이 같은 포트를 사용하면 Apache나 MySQL이 시작되지 않을 수 있으며, 이 경우 충돌하는 프로그램을 종료하거나 포트를 변경해야 합니다.

**실행 결과**

Apache와 MySQL이 정상적으로 실행되면 XAMPP 제어판에서 두 모듈이 실행 중인 상태로 표시됩니다.

브라우저에서 다음 주소에 접속하여 Apache가 동작하는지 확인할 수 있습니다.

```text
http://localhost/
```

프로젝트와 데이터베이스 연결 설정이 올바르면 다음 주소도 열 수 있습니다.

```text
http://localhost/course-admin/
```

**질문**

XAMPP에서 Apache 또는 MySQL을 실행하지 않으면 어떤 문제가 발생하는가?

브라우저의 요청을 받을 웹 서버가 없으므로 다음과 같은 문제가 발생합니다.

PHP 파일은 PHP 실행 환경을 통해 처리해야 하며, XAMPP 실습에서는 Apache가 이 연결을 담당합니다.

- 학생 목록 조회 실패
- 로그인 실패

```text

정적인 화면 일부는 보일 수 있지만 데이터베이스를 사용하는 부분에서 오류가 발생합니다.

MySQL에 데이터가 정상적으로 저장되어 있어도 브라우저에 웹 페이지를 제공할 서버가 없으므로 웹사이트에 접속할 수 없습니다.

다만 DBeaver 같은 데이터베이스 클라이언트는 실행 중인 MySQL에 직접 접속할 수 있습니다.

**Apache와 MySQL이 모두 중지된 경우**

웹 페이지 제공과 데이터베이스 처리가 모두 불가능합니다.

따라서 PHP와 MySQL을 사용하는 로컬 웹 애플리케이션을 실습할 때는 일반적으로 Apache와 MySQL이 모두 실행 중이어야 합니다.
````

##### 15. 왜 인덱스는 무조건 좋지 않다는 내용을 배우는가?

**질문**

왜 “인덱스는 무조건 좋지 않다”는 내용을 배우는가?

**답변**

인덱스는 책의 **찾아보기**와 비슷합니다. 원하는 데이터를 빠르게 찾는 데 도움을 주지만, 무료로 사용할 수 있는 기능은 아닙니다.

인덱스를 만들면 다음과 같은 장점이 있습니다.

- 적은 수의 데이터를 빠르게 찾을 수 있습니다.
- `WHERE` 조건이나 `JOIN` 조건에 맞는 행을 찾는 시간이 줄어들 수 있습니다.
- 인덱스의 정렬 순서를 이용하면 `ORDER BY` 작업이 줄어들 수 있습니다.

하지만 다음과 같은 비용도 발생합니다.

- 새로운 데이터를 `INSERT`할 때 테이블과 인덱스를 함께 갱신해야 합니다.
- 인덱스 컬럼을 `UPDATE`하면 인덱스도 수정해야 합니다.
- 데이터를 `DELETE`할 때도 인덱스를 정리해야 합니다.
- 인덱스를 저장할 디스크 공간이 필요합니다.
- 결과로 가져올 행이 많으면 인덱스보다 테이블 전체를 읽는 `Seq Scan`이 더 빠를 수 있습니다.

예를 들어 100,000행 중 10행만 찾는다면 인덱스가 유리할 가능성이 큽니다. 반면 100,000행 중 30,000행을 가져온다면 인덱스를 따라 테이블의 여러 위치를 방문하는 것보다 테이블을 처음부터 한 번 읽는 편이 더 효율적일 수 있습니다.

따라서 인덱스는 많이 만드는 것이 목적이 아닙니다. 다음 내용을 측정한 뒤 인덱스의 유지 여부를 결정해야 합니다.

- 실제로 자주 실행되는 조회인가?
- 전체 데이터 중 몇 행을 반환하는가?
- 인덱스 생성 후 읽는 데이터의 양이 줄었는가?
- 실행 시간이 반복해서 줄어드는가?
- 읽기 성능 향상이 쓰기 및 저장 비용보다 가치가 있는가?

---

##### 16. 인덱스와 실행 계획을 나중에 어디에 활용할 수 있는가?

**질문**

인덱스와 실행 계획을 나중에 어디에 활용할 수 있는가?

**답변**

쇼핑몰, 예약 시스템, 게시판, 온라인 강의 서비스처럼 많은 데이터를 저장하는 프로그램에서 조회 속도가 느릴 때 활용할 수 있습니다.

예를 들어 다음과 같은 조회가 반복될 수 있습니다.

- 이메일 주소로 특정 회원 찾기
- 특정 회원의 최근 주문 20건 찾기
- 특정 상품의 미배송 주문 찾기
- 특정 학생의 수강신청 내역 찾기
- 특정 강사의 진행 중인 강의 찾기

서비스가 느려졌을 때 무조건 인덱스부터 만드는 것이 아니라 다음 순서로 문제를 해결합니다.

1. 느리거나 자주 실행되는 SQL을 찾습니다.
2. `EXPLAIN (ANALYZE, BUFFERS)`로 현재 실행 계획을 확인합니다.
3. 데이터 규모와 조회 결과의 비율을 확인합니다.
4. 조회 조건에 맞는 후보 인덱스를 설계합니다.
5. 인덱스를 생성합니다.
6. 같은 데이터와 같은 SQL로 다시 측정합니다.
7. 결과 행, 실행 계획, `Buffers`, 실행 시간을 비교합니다.
8. 효과가 충분하면 유지하고, 효과가 작으면 보류하거나 제거합니다.

**예시 코드**

다음은 특정 학생의 수강신청 기록을 찾는 SQL입니다.

```sql
EXPLAIN (ANALYZE, BUFFERS)
SELECT *
FROM performance_lab.enrollments
WHERE student_id = 5000;
```

현재 실행 계획을 저장한 다음 후보 인덱스를 만듭니다.

```sql
CREATE INDEX idx_performance_enrollments_student_id
ON performance_lab.enrollments (student_id);
```

인덱스를 만든 뒤 똑같은 SQL을 다시 실행합니다.

```sql
EXPLAIN (ANALYZE, BUFFERS)
SELECT *
FROM performance_lab.enrollments
WHERE student_id = 5000;
```

**실행 결과**

정확한 비용과 시간은 컴퓨터, PostgreSQL 버전, 캐시 상태 등에 따라 달라질 수 있습니다.

인덱스 생성 전에는 다음과 같은 계획이 나타날 수 있습니다.

```text
Seq Scan on enrollments
  Filter: (student_id = 5000)
  Rows Removed by Filter: ...
```

이는 테이블을 순서대로 읽으면서 `student_id = 5000`인 행을 골랐다는 뜻입니다.

인덱스 생성 후에는 다음과 같은 계획이 나타날 수 있습니다.

```text
Index Scan using idx_performance_enrollments_student_id
  Index Cond: (student_id = 5000)
```

이는 `student_id` 인덱스를 이용해 필요한 행의 위치를 찾았다는 뜻입니다.

하지만 `Index Scan`이 나타났다는 사실만으로 성공이라고 판단하지 않습니다. 다음 항목을 함께 비교해야 합니다.

- 전후의 결과 행이 같은가?
- `Index Cond`에 실제 조회 조건이 표시되는가?
- `Buffers hit/read`가 줄었는가?
- 읽는 범위가 줄었는가?
- 실행 시간 감소가 반복 측정에서도 나타나는가?

---

##### 17. PostgreSQL 16과 PostgreSQL 18 이상의 다중 컬럼 B-tree 인덱스 사용 차이는 무엇인가?

**질문**

“PostgreSQL 16에서는 다중 컬럼 B-tree 인덱스에서 선두 컬럼 조건이 없고 후행 컬럼만 조건으로 사용될 때 탐색 범위를 줄이기 어렵고, PostgreSQL 18 이상에서는 B-tree Skip Scan이 추가되었다”는 게 무슨 뜻이야?

**답변**

다음과 같은 복합 인덱스가 있다고 가정해 보겠습니다.

```text
(course_id, status)
```

이 인덱스는 데이터를 다음 순서로 정리합니다.

1. 먼저 `course_id` 순서로 정리합니다.
2. `course_id`가 같으면 그 안에서 `status` 순서로 정리합니다.

따라서 `course_id`는 **선두 컬럼**, `status`는 **후행 컬럼**입니다.

이 인덱스는 다음과 같은 조회에 잘 맞을 수 있습니다.

```sql
WHERE course_id = 1500
```

```sql
WHERE course_id = 1500
  AND status = '수강중'
```

먼저 `course_id = 1500`인 위치로 이동한 다음, 그 안에서 필요한 상태를 찾을 수 있기 때문입니다.

하지만 다음처럼 후행 컬럼인 `status`만 사용하면 상황이 달라집니다.

```sql
WHERE status = '수강중'
```

인덱스가 `status`부터 정리된 것이 아니기 때문에 여러 `course_id` 묶음을 확인해야 할 수 있습니다.

전화번호부가 `(성, 이름)` 순서로 정리되어 있다고 생각하면 이해하기 쉽습니다.

- 김씨 전부 찾기: 쉽습니다.
- 김씨 중 철수 찾기: 쉽습니다.
- 성을 모르고 이름이 철수인 사람 전부 찾기: 어렵습니다.

PostgreSQL 16에서는 선두 컬럼 조건이 없고 후행 컬럼만 사용하면 이 인덱스로 탐색 범위를 효과적으로 줄이기 어려울 수 있습니다. 이 경우 `Seq Scan`이 선택될 수 있습니다.

PostgreSQL 18 이상에서 추가된 B-tree Skip Scan은 선두 컬럼 조건이 없어도 선두 컬럼의 값들을 건너가며 후행 컬럼을 찾을 수 있게 하는 기능입니다.

예를 들면 다음과 같은 방식입니다.

```text
course_id 1 묶음 → 수강중 확인
course_id 2 묶음 → 수강중 확인
course_id 3 묶음 → 수강중 확인
```

다만 PostgreSQL 18 이상이라고 해서 항상 Skip Scan이 선택되는 것은 아닙니다.

다음과 같은 경우에는 여전히 `Seq Scan`이 더 유리할 수 있습니다.

- 선두 컬럼의 서로 다른 값이 매우 많은 경우
- 조회 결과로 반환할 행이 많은 경우
- 테이블 전체를 읽는 비용이 더 저렴한 경우
- 데이터 분포나 통계상 Skip Scan이 불리한 경우

따라서 PostgreSQL 버전이 다르면 같은 데이터와 같은 SQL이라도 서로 다른 실행 계획이 나올 수 있습니다.

**예시 코드**

복합 인덱스를 생성합니다.

```sql
CREATE INDEX idx_performance_enrollments_course_status
ON performance_lab.enrollments (course_id, status);
```

선두 컬럼만 사용하는 조회입니다.

```sql
EXPLAIN (ANALYZE, BUFFERS)
SELECT *
FROM performance_lab.enrollments
WHERE course_id = 1500;
```

선두 컬럼과 후행 컬럼을 함께 사용하는 조회입니다.

```sql
EXPLAIN (ANALYZE, BUFFERS)
SELECT *
FROM performance_lab.enrollments
WHERE course_id = 1500
  AND status = '수강중';
```

후행 컬럼만 사용하는 조회입니다.

```sql
EXPLAIN (ANALYZE, BUFFERS)
SELECT *
FROM performance_lab.enrollments
WHERE status = '수강중';
```

**실행 결과**

PostgreSQL 버전, 통계 및 데이터 분포에 따라 실제 결과는 달라질 수 있습니다.

`course_id` 조건이 포함된 조회에서는 다음과 같은 내용이 나타날 수 있습니다.

```text
Index Cond: (course_id = 1500)
```

두 컬럼을 함께 사용하면 다음과 같이 나타날 수 있습니다.

```text
Index Cond: ((course_id = 1500) AND (status = '수강중'))
```

반면 PostgreSQL 16에서 `status`만 조회하고 결과가 약 30%처럼 많다면 다음 계획이 나타날 수 있습니다.

```text
Seq Scan on enrollments
  Filter: (status = '수강중')
```

이는 인덱스 오류가 아닙니다. PostgreSQL이 인덱스를 따라 많은 위치를 방문하는 것보다 테이블 전체를 한 번 읽는 것이 더 저렴하다고 판단한 결과일 수 있습니다.

---

##### 18. 다중 컬럼 B-tree 인덱스란 무엇인가?

**질문**

다중 컬럼 B-tree 인덱스가 뭐야? 초등학생에게 설명하듯 쉽게 알려줘.

**답변**

다중 컬럼 B-tree 인덱스는 **두 개 이상의 기준을 정해 순서대로 정리한 찾아보기 표**입니다.

학교에서 학생들을 다음과 같이 정리한다고 생각해 보겠습니다.

1. 먼저 반별로 정리합니다.
2. 같은 반 안에서는 이름순으로 정리합니다.

```text
1반
├─ 김민수
├─ 박지수
└─ 이철수

2반
├─ 김민수
├─ 박지수
└─ 최영희
```

이 명단은 `(반, 이름)` 순서로 정리된 다중 컬럼 인덱스와 비슷합니다.

이 명단을 이용하면 다음 질문에는 빠르게 답할 수 있습니다.

- 2반 학생을 모두 찾아줘.
- 2반에 있는 김민수를 찾아줘.

먼저 2반으로 이동한 다음 그 안에서 이름을 찾으면 되기 때문입니다.

하지만 다음 질문은 상대적으로 어렵습니다.

- 몇 반인지는 모르지만 이름이 김민수인 학생을 모두 찾아줘.

명단이 이름부터 정리된 것이 아니기 때문에 1반, 2반, 3반을 각각 살펴봐야 할 수 있습니다.

B-tree의 `B`를 단순히 **가지가 여러 개로 나뉘는 정리 구조**라고 생각해도 됩니다. 데이터베이스는 처음부터 모든 값을 읽는 대신 큰 범위에서 작은 범위로 이동하며 원하는 값을 찾습니다.

```text
전체 데이터
├─ 작은 course_id 값
├─ 중간 course_id 값
└─ 큰 course_id 값
```

다중 컬럼 B-tree 인덱스에서 컬럼 순서가 중요한 이유도 여기에 있습니다. 먼저 적은 컬럼을 기준으로 큰 묶음이 나뉘고, 그 안에서 다음 컬럼이 정리되기 때문입니다.

**예시 코드**

다음 코드는 `course_id`를 먼저 정리하고, 같은 `course_id` 안에서 `status`를 정리하는 인덱스를 만듭니다.

```sql
CREATE INDEX idx_course_status
ON performance_lab.enrollments (course_id, status);
```

다음 조회에는 잘 맞을 가능성이 있습니다.

```sql
SELECT *
FROM performance_lab.enrollments
WHERE course_id = 1500
  AND status = '수강중';
```

**실행 결과**

인덱스가 정상적으로 생성되면 PostgreSQL은 다음 메시지를 반환합니다.

```text
CREATE INDEX
```

이 메시지는 인덱스가 생성되었다는 뜻일 뿐, 해당 조회가 반드시 빨라졌다는 뜻은 아닙니다.

실제 효과는 다음과 같이 실행 계획을 측정하여 확인해야 합니다.

```sql
EXPLAIN (ANALYZE, BUFFERS)
SELECT *
FROM performance_lab.enrollments
WHERE course_id = 1500
  AND status = '수강중';
```

실행 계획에서 다음과 같은 내용이 나타나는지 확인할 수 있습니다.

```text
Index Cond: ((course_id = 1500) AND (status = '수강중'))
```

마지막으로 `Buffers`, 실제 반환 행 수, `Execution Time` 등을 인덱스 생성 전과 비교해야 합니다.
