### 1. CSS 기본 — 선택자, 색상, 폰트, 상속과 우선순위

[4강 학습 자료](https://blog.naver.com/dev-dog/224441060768)

1. HTML은 제목, 회원 목록, 등록 버튼 등 화면에 들어갈 내용과 구조를 만듭니다.

2. CSS는 글자색, 배경색, 크기, 글꼴 등 화면에 보이는 모습을 꾸밉니다.

3. CSS에서는 선택자를 사용하여 스타일을 적용할 HTML 요소를 지정합니다.

4. `color: red;`에서 `color`는 바꿀 항목을 나타내는 속성이고, `red`는 적용할 내용을 나타내는 값입니다.

5. 별도의 CSS 파일을 만들어 HTML 파일과 연결하면 외부에 작성한 CSS를 사용할 수 있습니다.

6. CSS의 상속과 우선순위에 따라 같은 요소에 여러 스타일이 지정되었을 때 실제로 적용되는 스타일이 결정됩니다.


### 2. CSS Box Model과 기본 배치

[5강 학습 자료](https://blog.naver.com/dev-dog/224441064261)

1. CSS Box Model을 이용하여 화면 요소의 크기와 간격을 조절하는 방법을 배웠습니다.

2. 앞에서 배운 색상과 글꼴 설정을 활용하여 회원관리 페이지를 여백과 테두리가 있는 카드 형태로 꾸몄습니다.

3. 브라우저에서 `F12`를 누르고 `Elements` 탭을 열면 각 요소가 어떤 박스 구조로 이루어져 있는지 확인할 수 있습니다.

4. HTML의 버튼, 텍스트 등의 요소는 각각 하나의 네모난 박스로 표현됩니다.

5. 박스 모델은 안쪽부터 다음과 같은 순서로 구성됩니다.

   - `content`: 글자나 이미지 같은 실제 내용
   - `padding`: 내용과 테두리 사이의 안쪽 여백
   - `border`: 요소를 둘러싼 테두리
   - `margin`: 테두리 바깥쪽의 여백

6. `padding`과 `margin`은 모두 공간을 만드는 속성이지만, `padding`은 요소 안쪽에, `margin`은 요소 바깥쪽에 공간을 만듭니다.

7. 각 영역의 크기는 `px` 등의 단위로 지정하며, 작은 수치 차이도 실제 화면 배치에 큰 영향을 줄 수 있습니다.

8. HTML 태그는 기본적인 표시 방식을 가지고 있습니다.

   - `div`는 블록 요소이므로 한 줄 전체를 차지하고 다음 요소가 새로운 줄에 표시됩니다.
   - `span`은 인라인 요소이므로 내용이 차지하는 영역만 사용하고 다른 요소와 같은 줄에 표시될 수 있습니다.


### 3. AI와 실행 증거로 데이터베이스 설계 검증하기

[Chapter 13 확장 실습 자료](https://blog.naver.com/dev-dog/224441098260)

1. 인터넷에서 가져온 외부 데이터를 기존 데이터와 연결하는 방법을 배웠습니다.

2. 데이터를 단순히 다운로드하는 데서 끝내지 않고, 기존 데이터와 연결해 분석에 사용할 수 있는지 확인했습니다.

3. 예를 들어 기존 데이터에 `월별 주문 금액`이 있다면 외부에서 가져온 `월별 공휴일 수`를 연결할 수 있습니다.

4. 두 데이터를 연결하면 “공휴일이 많은 달에는 주문 금액도 높은가?”와 같은 질문을 분석할 수 있습니다.

5. 외부 데이터를 사용할 때에는 데이터의 출처와 구조를 확인하고, 기존 데이터와 올바르게 연결되었는지 실행 결과를 통해 검증해야 합니다.

6. 웹 데이터에서 원하는 항목을 가져올 때 선택자를 사용할 수 있습니다. 이때 `item_info`는 보통 클래스 이름이며, CSS 선택자로 표현할 때는 `.item_info`처럼 앞에 마침표를 붙입니다.
