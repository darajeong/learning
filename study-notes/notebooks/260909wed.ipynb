````markdown
##### 1. pandas 기초와 데이터 선택·확인·집계

- **Series(시리즈)**는 인덱스가 붙은 1차원 데이터이며, 데이터프레임의 열 하나를 선택하면 보통 Series가 됩니다.
- **DataFrame(데이터프레임)**은 행과 열로 구성된 2차원 표이며, 열마다 서로 다른 자료형을 담을 수 있습니다.
- `df["amount"]`는 열 이름 하나를 전달해 Series로 선택하고, `df[["amount"]]`는 열 이름을 리스트로 전달해 열이 하나여도 DataFrame으로 선택합니다.
- `df[["customer", "amount"]]`처럼 열 이름 여러 개를 리스트로 전달하면 해당 열들로 구성된 DataFrame을 얻습니다.
- `df.head()`는 처음 **최대 5개 행**을 보여주므로, 예시처럼 데이터가 4행이면 4행 모두 나옵니다.
- `df.shape`는 `(행 개수, 열 개수)`를 반환하며, 예시에서는 `(4, 3)`이고 메서드가 아닌 속성이므로 `()`를 붙이지 않습니다.
- `df.info()`는 행·열 개수와 자료형 등의 구조를 출력하며, 예시의 `amount`가 `3 non-null`이면 전체 4행 중 값이 있는 행이 3개이고 결측치는 1개라는 뜻입니다.
- **불리언 마스크**는 조건에 맞으면 `True`, 아니면 `False`로 표시한 데이터이며, `df[df["status"] == "완료"]`처럼 사용하면 완료 상태인 민수와 현우의 행만 선택합니다.
- `df["status"].value_counts()`는 상태별 등장 횟수를 세어 완료 2개·대기 1개·취소 1개를 보여주며, 기본적으로 제외되는 결측치까지 세려면 `dropna=False`를 넣습니다.
- `df["amount"].sum()`은 기본적으로 결측치를 제외하고 금액을 더하므로, 예시의 결과는 `170000.0`입니다.
- `df["amount"].isna()`는 금액이 결측치인 곳을 `True`로 표시하므로, 예시에서는 순서대로 `False, False, True, False`가 나옵니다.
- `df["amount"].isna().sum()`은 결측치를 뜻하는 `True`를 세어 `1`을 반환하고, `df.isna().sum()`은 모든 열의 결측치 개수를 각각 계산합니다.
- `df[df["amount"].isna()]`는 금액이 결측치인 행만 선택하므로, 예시에서는 현우의 행이 나옵니다.
- `groupby`를 이용한 집계는 **같은 기준값끼리 나누기(Split) → 각 그룹에 함수 적용하기(Apply) → 결과 합치기(Combine)**의 3단계로 이해할 수 있습니다.
- `df.groupby("status")["amount"].sum()`은 상태별로 금액을 더해 완료 `45000.0`, 대기 `72000.0`, 취소 `53000.0`을 구합니다.
- `agg`는 집계 메서드이며, `df.groupby("status")["amount"].agg(["sum", "mean", "count"])`는 상태별 금액의 합계·평균·결측치를 제외한 개수를 한 번에 계산합니다.
- 집계 함수 `count`는 결측치를 제외한 값의 개수를 세므로 예시의 완료 그룹에서는 `1`이고, 결측치와 관계없이 그룹의 전체 행 수를 세는 `size`는 `2`입니다.

##### 2. 괄호의 표기와 용도

**질문**

소괄호, 대괄호, 중괄호는 어떻게 표기하고 Python에서 어디에 사용하나요?

**답변**

| 이름 | 표기 | 대표 용도 |
|---|---|---|
| 소괄호 | `()` | 함수·메서드 호출, 연산 순서 지정 |
| 대괄호 | `[]` | 리스트 생성, 원소 접근, 데이터프레임의 열이나 행 선택 |
| 중괄호 | `{}` | 딕셔너리 또는 집합 생성 |

- 딕셔너리는 `{"키": 값}` 형태로 만듭니다.
- 집합은 `{"서울", "부산"}`처럼 값을 나열해 만듭니다.
- 빈 중괄호 `{}`는 빈 딕셔너리이며, 빈 집합은 `set()`으로 만듭니다.

**예시 코드**

```python
customer = {"name": "민수", "amount": 45000}
cities = ["서울", "부산", "인천"]

print(customer["name"])
print(cities[0])
print((2 + 3) * 4)
```

**실행 결과**

```text
민수
서울
20
```

##### 3. agg와 그룹별 집계 과정

**질문**

`agg`는 무엇이며, 그룹별 집계의 3단계는 무엇인가요?

**답변**

`agg`는 **aggregate(집계하다)**의 줄임말로, 하나 이상의 집계 함수를 적용하는 메서드입니다.

| 집계 함수 | 의미 |
|---|---|
| `sum` | 합계 |
| `mean` | 평균 |
| `count` | 결측치를 제외한 값의 개수 |
| `min` | 최솟값 |
| `max` | 최댓값 |

그룹별 집계는 다음 3단계로 이해할 수 있습니다.

1. **Split(분할)**: 같은 기준값을 가진 행끼리 나눕니다.
2. **Apply(적용)**: 각 그룹에 집계 함수를 적용합니다.
3. **Combine(결합)**: 계산 결과를 하나의 결과로 합칩니다.

`df.groupby("status")["amount"].agg(["sum", "mean", "count"])`의 의미는 다음과 같습니다.

- `groupby("status")`: 상태별로 묶습니다.
- `["amount"]`: 계산할 금액 열을 선택합니다.
- `agg(["sum", "mean", "count"])`: 합계, 평균, 개수를 계산합니다.

**예시 코드**

```python
import pandas as pd

df = pd.DataFrame({
    "status": ["완료", "대기", "완료"],
    "amount": [45000, 72000, 53000]
})

result = df.groupby("status", sort=False)["amount"].agg(
    ["sum", "mean", "count"]
)

print(result)
```

**실행 결과**

| status | sum | mean | count |
|---|---:|---:|---:|
| 완료 | 98000 | 49000.0 | 2 |
| 대기 | 72000 | 72000.0 | 1 |

##### 4. pandas 주요 기능의 의미·예시·결과

**질문**

불리언 마스크, `value_counts()`, `info()`, `isna()` 등의 기능을 예시와 결과까지 표로 정리하면 어떻게 되나요?

**답변**

- **불리언 마스크**는 각 행이 조건을 만족하는지 `True`와 `False`로 나타냅니다.
- **`value_counts()`**는 각 값의 등장 횟수를 세며, 금액 합계를 계산하지 않습니다.
- **`info()`**는 행·열 개수, 자료형, 결측이 아닌 값의 개수 등 전체 구조를 보여줍니다.
- **`isna()`**는 결측치 여부를 표시하며, 개수를 구하려면 `.sum()`을 붙입니다.

**예시 코드**

다음 데이터프레임을 기준으로 아래 표의 코드를 실행합니다.

```python
import pandas as pd

df = pd.DataFrame({
    "customer": ["민수", "지영", "현우", "수진"],
    "status": ["완료", "대기", "완료", "취소"],
    "amount": [45000, 72000, None, 53000]
})
```

**실행 결과**

아래 결과는 출력 내용을 읽기 쉽게 요약한 것입니다.

| 기능 | 사용 예시 | 실행 결과 |
|---|---|---|
| 완료 여부 확인 | `df["status"] == "완료"` | 순서대로 `True, False, True, False` |
| 완료 행만 선택 | `df[df["status"] == "완료"]` | 민수와 현우의 행 |
| 상태별 개수 | `df["status"].value_counts()` | 완료 2개, 대기 1개, 취소 1개 |
| 전체 금액 합계 | `df["amount"].sum()` | `170000.0` |
| 상태별 금액 합계 | `df.groupby("status")["amount"].sum()` | 완료 45000, 대기 72000, 취소 53000 |
| 행·열 개수 | `df.shape` | `(4, 3)` |
| 전체 구조 확인 | `df.info()` | 4행, 3열이며 `amount`는 3개가 Non-Null |
| 금액의 결측 여부 | `df["amount"].isna()` | 순서대로 `False, False, True, False` |
| 금액의 결측치 개수 | `df["amount"].isna().sum()` | `1` |
| 모든 열의 결측치 개수 | `df.isna().sum()` | customer 0, status 0, amount 1 |
| 금액이 결측인 행 선택 | `df[df["amount"].isna()]` | 현우의 행 |
| 처음 5개 행 확인 | `df.head()` | 데이터가 4행이므로 전체 4행 |

- `sum()`은 기본적으로 결측치를 제외하고 계산합니다.
- `value_counts()`는 기본적으로 결측치를 제외합니다. 포함하려면 `value_counts(dropna=False)`를 사용합니다.
- `info()`의 **Non-Null Count**는 결측이 아닌 값의 개수입니다. 전체 4행 중 3개가 Non-Null이면 결측치는 1개입니다.
- 데이터가 클 때도 Non-Null Count를 표시하려면 `df.info(show_counts=True)`를 사용할 수 있습니다.

##### 5. LLM의 개념과 활용

**질문**

LLM이란 무엇이며, 어떤 일에 사용할 수 있나요?

**답변**

LLM은 **Large Language Model(대규모 언어 모델)**의 약자입니다. 많은 텍스트 데이터를 학습하여 질문의 문맥을 처리하고 답변이나 글을 생성하는 AI 모델입니다.

텍스트를 생성할 때는 앞의 문맥을 바탕으로 다음에 올 **토큰**을 예측하는 과정을 반복합니다. 토큰은 단어 전체이거나 단어의 일부일 수 있습니다.

| 활용 | 입력 예시 | 생성되는 답변 예시 |
|---|---|---|
| 개념 설명 | “불리언 마스크가 뭐야?” | “조건에 맞는지 True와 False로 표시한 데이터입니다.” |
| 번역 | “‘안녕하세요’를 영어로 번역해 줘.” | “Hello.” |
| 요약 | “이 회의록을 세 문장으로 요약해 줘.” | 핵심 논의와 결정 사항을 정리한 세 문장 |
| 코드 작성 | “금액 열의 합계를 구하는 pandas 코드를 써 줘.” | `df["amount"].sum()` |
| 문장 분류 | “‘배송이 빨라서 만족해요’는 긍정이야, 부정이야?” | “긍정입니다.” |

LLM은 그럴듯하지만 틀린 답을 생성할 수 있으므로, 계산 결과나 중요한 사실은 실제 데이터와 도구로 확인해야 합니다.
````
