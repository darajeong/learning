##### 1. PostgreSQL, MySQL, DBeaver의 역할과 차이

**질문**

PostgreSQL, MySQL, DBeaver는 각각 어떤 역할을 하며 PostgreSQL과 MySQL은 무엇이 다른가요?

**답변**

PostgreSQL과 MySQL은 데이터를 실제로 저장하고 관리하는 **DBMS(데이터베이스 관리 시스템)**입니다. 둘 다 회원, 주문, 게시글 같은 데이터를 저장하고 SQL로 조회할 수 있습니다.

DBeaver는 데이터를 직접 저장하는 DBMS가 아니라, PostgreSQL이나 MySQL에 접속하여 데이터를 편리하게 관리하는 프로그램입니다.

쉽게 비유하면 다음과 같습니다.

- **PostgreSQL·MySQL:** 데이터를 보관하는 창고
- **DBeaver:** 창고 안의 물건을 확인하고 관리하는 관리실 화면
- **SQL:** 관리실에서 창고에 전달하는 명령어

PostgreSQL과 MySQL은 기본적인 역할은 같지만 특징에 차이가 있습니다.

- **PostgreSQL:** 복잡한 검색과 데이터 처리, 다양한 고급 기능에 강합니다.
- **MySQL:** 비교적 접근하기 쉽고 일반적인 웹 서비스에서 널리 사용됩니다.
- **DBeaver:** PostgreSQL과 MySQL을 포함한 여러 DBMS에 접속할 수 있습니다.

따라서 DBeaver와 PostgreSQL을 비교하는 것은 VS Code와 Python을 비교하는 것과 비슷합니다.

- **Python:** 프로그래밍 언어
- **VS Code:** Python 코드를 작성하는 도구
- **PostgreSQL·MySQL:** 데이터를 저장하고 SQL을 실행하는 DBMS
- **DBeaver:** SQL을 작성하고 데이터베이스를 관리하는 도구

**예시 코드**

DBeaver에서 다음 SQL을 작성해 PostgreSQL이나 MySQL에 전달할 수 있습니다.

```sql
SELECT *
FROM members;
```

**실행 결과**

`members` 테이블에 다음 데이터가 저장되어 있다면:

| member_id | name |
|---:|---|
| 1 | 철수 |
| 2 | 영희 |

조회 결과는 다음과 같습니다.

| member_id | name |
|---:|---|
| 1 | 철수 |
| 2 | 영희 |

DBeaver는 PostgreSQL이나 MySQL이 보내준 결과를 표 형태로 보여줍니다.

##### 2. SQL의 작성 순서와 실제 실행 순서

**질문**

SQL은 `SELECT`로 작성하는데 실제로는 왜 `FROM`부터 실행되나요?

**답변**

SQL을 사람이 작성하는 순서와 데이터베이스가 실제로 처리하는 순서는 다릅니다.

일반적인 작성 순서는 다음과 같습니다.

1. `SELECT`
2. `FROM`
3. `WHERE`
4. `GROUP BY`
5. `HAVING`
6. `ORDER BY`

그러나 일반적인 실제 실행 순서는 다음과 같습니다.

1. `FROM`: 데이터를 가져올 테이블을 정합니다.
2. `WHERE`: 조건에 맞는 행만 남깁니다.
3. `GROUP BY`: 데이터를 그룹으로 묶습니다.
4. `HAVING`: 만들어진 그룹을 조건에 따라 거릅니다.
5. `SELECT`: 최종적으로 보여줄 열을 선택합니다.
6. `ORDER BY`: 결과의 순서를 정렬합니다.

데이터베이스는 먼저 어떤 테이블을 사용할지 알아야 그 안에서 데이터를 찾을 수 있습니다. 따라서 SQL 문장은 `SELECT`로 시작하지만 실제 처리는 `FROM`부터 시작합니다.

**예시 코드**

```sql
SELECT name, COUNT(*) AS question_count
FROM questions
WHERE created_at >= '2026-01-01'
GROUP BY name
HAVING COUNT(*) >= 2
ORDER BY name;
```

**실행 결과**

데이터베이스는 위 SQL을 다음 순서로 처리합니다.

```text
FROM questions
→ WHERE created_at >= '2026-01-01'
→ GROUP BY name
→ HAVING COUNT(*) >= 2
→ SELECT name, COUNT(*)
→ ORDER BY name
```

예를 들어 질문 데이터가 다음과 같다면:

| name | question |
|---|---|
| 철수 | 질문 1 |
| 철수 | 질문 2 |
| 영희 | 질문 3 |

결과는 다음과 같습니다.

| name | question_count |
|---|---:|
| 철수 | 2 |

영희는 질문이 1개이므로 `HAVING COUNT(*) >= 2` 조건을 통과하지 못합니다.

##### 3. 기준 데이터와 파생 데이터의 차이

**질문**

기준 데이터와 파생 데이터는 무엇이며 어떤 차이가 있나요?

**답변**

**기준 데이터**는 특정 업무에서 공식적인 판단의 근거로 사용하는 데이터입니다. 실제로 발생한 개별 주문, 회원 가입, 출석, 결제 기록 등이 기준 데이터가 될 수 있습니다.

**파생 데이터**는 기준 데이터를 계산하거나 가공해서 만든 결과입니다. 일일 매출, 월별 주문 수, 회원별 출석률, 대시보드 지표 등이 파생 데이터에 해당합니다.

쉽게 말하면 다음과 같습니다.

- **기준 데이터:** 계산에 사용하는 재료
- **파생 데이터:** 재료를 계산하거나 가공해서 만든 결과

예를 들어 주문별 결제 기록은 기준 데이터이고, 그 기록을 모두 더해서 만든 일일 매출은 파생 데이터입니다.

기준 데이터가 수정되면 파생 데이터도 다시 계산해야 합니다. 파생 데이터가 없어져도 기준 데이터가 남아 있다면 다시 만들 수 있지만, 기준 데이터가 없어지면 원래 기록을 복구하기 어려울 수 있습니다.

**예시 코드**

다음 SQL은 개별 주문 기록이라는 기준 데이터를 이용해 날짜별 매출이라는 파생 데이터를 계산합니다.

```sql
SELECT
    order_date,
    SUM(amount) AS daily_sales
FROM orders
GROUP BY order_date
ORDER BY order_date;
```

**실행 결과**

기준 데이터인 개별 주문 기록이 다음과 같다고 가정합니다.

| order_id | order_date | amount |
|---:|---|---:|
| 1 | 2026-09-01 | 4000 |
| 2 | 2026-09-01 | 5000 |
| 3 | 2026-09-02 | 4000 |

SQL로 계산한 파생 데이터는 다음과 같습니다.

| order_date | daily_sales |
|---|---:|
| 2026-09-01 | 9000 |
| 2026-09-02 | 4000 |

`2026-09-01`의 개별 주문 금액인 `4,000원`과 `5,000원`은 기준 데이터이고, 두 금액을 더한 일일 매출 `9,000원`은 파생 데이터입니다.

##### 4. Figma, Figma Make, Claude Code의 차이와 연결 과정

**질문**

Figma와 Figma Make는 무엇이 다르며, Figma Make가 생성한 작업물이나 코드를 Claude Code로 옮기고 연결하는 과정은 어떻게 이루어지나요?

**답변**

일반적인 **Figma Design**은 화면의 모양과 구조를 만드는 디자인 도구입니다. 버튼의 색상, 글꼴, 간격, 화면 배치 등을 설계할 수 있습니다.

**Figma Make**는 설명이나 디자인을 바탕으로 실제로 작동하는 웹 프로토타입과 코드를 만들어주는 도구입니다. 버튼을 누르면 화면이 바뀌는 것과 같은 동작도 구현할 수 있습니다.

**Claude Code**는 전달받은 소스 코드를 읽고 수정하거나 다른 프로젝트에 통합하는 개발 도구입니다.

세 가지의 역할을 집에 비유하면 다음과 같습니다.

- **Figma Design:** 집을 어떻게 지을지 보여주는 설계도
- **Figma Make:** 설계도를 바탕으로 만든 작동 가능한 모델하우스
- **Claude Code:** 모델하우스의 구조를 분석하고 실제 건물에 맞게 공사하는 개발자

Figma Make가 생성한 전체 코드가 있다면 Claude Code는 그 코드를 읽을 수 있습니다. 하지만 Figma Make 코드와 기존 프로젝트가 서로 다른 기술이나 구조를 사용한다면 코드를 그대로 복사하는 것만으로는 통합되지 않을 수 있습니다.

예를 들어 Figma Make는 React와 TypeScript로 만들어졌지만 기존 대시보드는 순수 HTML, CSS, JavaScript로 만들어졌을 수 있습니다. 이 경우 Claude Code가 코드를 현재 프로젝트의 구조에 맞게 변환해야 합니다.

다만 작업물을 대시보드에 통합할 필요 없이 다른 사람에게 보여주는 것이 목적이라면 Claude Code를 사용할 필요가 없습니다.

Figma Make에서 다음 순서로 공개할 수 있습니다.

1. Figma Make 파일을 엽니다.
2. 오른쪽 위의 `Publish` 버튼을 누릅니다.
3. 제목과 공개 설정을 확인합니다.
4. 다시 `Publish`를 누릅니다.
5. 생성된 `https://이름.figma.site/` 주소를 복사해 전달합니다.

즉, 두 가지 요청을 구분해야 합니다.

- **결과물을 링크로 보여주기:** Figma Make에서 Publish하여 `figma.site` 링크를 생성합니다.
- **실제 서비스에 적용하기:** Figma Make의 전체 코드를 Claude Code에 전달하고 기존 프로젝트에 맞게 변환·통합합니다.

Figma Make의 미리보기 링크를 Claude Code에 전달하는 것과 전체 소스 코드를 전달하는 것도 서로 다릅니다. Claude Code가 코드를 직접 분석해야 한다면 Figma Make에서 전체 코드를 다운로드하거나 GitHub 저장소로 전달하는 것이 가장 확실합니다.
